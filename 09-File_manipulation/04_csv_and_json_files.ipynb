# Модуль 9. Работа с файлами

## Часть 4. Обработка файлов CSV и JSON

Помимо обычных текстовых файлов, Python поддерживает обработку CSV (Comma-Separated Values, значения, разделённые запятыми) и JSON (JavaScript Object Notation, формат записи объектов JavaScript) средствами стандартной библиотеки. В этом разделе мы разберём, как читать и записывать данные в этих форматах.

### 4.1. Обработка файлов CSV

CSV-файлы часто используют для хранения табличных данных. Каждая строка представляет запись, а значения в ней отделены разделителем, например запятой. Для чтения и записи таких файлов в Python есть модуль `csv`.

Прочитать данные из CSV-файла можно с помощью функции `csv.reader()`.

```python
import csv

with open("data.csv", "r") as file:
    csv_reader = csv.reader(file)
    for row in csv_reader:
        print(row)
```

В этом примере мы открываем CSV-файл `data.csv` для чтения и создаём объект `csv_reader` с помощью функции `csv.reader()`. Затем перебираем его в цикле и получаем каждую строку как список значений.

Чтобы записать данные в CSV-файл, можно использовать функцию `csv.writer()`.

```python
import csv

data = [
    ["Name", "Age", "Country"],
    ["John", 25, "USA"],
    ["Alice", 30, "Canada"],
    ["Bob", 28, "UK"]
]

with open("data.csv", "w", newline="") as file:
    csv_writer = csv.writer(file)
    csv_writer.writerows(data)
```

В этом примере данные для записи в CSV-файл представлены списком списков. Мы открываем файл `data.csv` для записи и создаём объект `csv_writer` с помощью функции `csv.writer()`. Затем методом `writerows()` записываем в файл все строки данных.

### 4.2. Указание разделителя в CSV-файлах

В некоторых CSV-файлах значения разделены другим символом вместо запятой. При чтении или записи файла можно задать нужный разделитель.

Например, чтобы прочитать CSV-файл, где разделителем служит точка с запятой (`;`), измените код следующим образом.

```python
import csv

with open("data.csv", "r") as file:
    csv_reader = csv.reader(file, delimiter=";")
    for row in csv_reader:
        print(row)
```

В этом примере мы передаём функции `csv.reader()` аргумент `delimiter=";"`. Так мы указываем, что значения в CSV-файле разделены точкой с запятой.

Аналогично при записи файла можно указать нужный разделитель в вызове `csv.writer()`.

```python
import csv

data = [
    ["Name", "Age", "Country"],
    ["John", 25, "USA"],
    ["Alice", 30, "Canada"],
    ["Bob", 28, "UK"]
]

with open("data.csv", "w", newline="") as file:
    csv_writer = csv.writer(file, delimiter=";")
    csv_writer.writerows(data)
```

В этом примере мы передаём функции `csv.writer()` аргумент `delimiter=";"`, чтобы значения в CSV-файле записывались через точку с запятой.

Выбирая подходящий разделитель, можно обрабатывать CSV-файлы с разными форматами разделения значений.

### 4.3. Обработка файлов JSON

JSON-файлы используют для хранения структурированных данных в читаемом текстовом формате. Для работы с ними Python предоставляет модуль `json`.

Прочитать данные из JSON-файла можно с помощью функции `json.load()`.

```python
import json

with open("data.json", "r") as file:
    data = json.load(file)
    print(data)
```

В этом примере мы открываем файл `data.json` для чтения и с помощью функции `json.load()` загружаем данные из файла в объект Python. Затем можно работать с данными как со словарём, списком или другой подходящей структурой данных.

Чтобы записать данные в JSON-файл, можно использовать функцию `json.dump()`.

```python
import json

data = {
    "name": "John",
    "age": 25,
    "country": "USA"
}

with open("data.json", "w") as file:
    json.dump(data, file)
```

В примере данные для записи в JSON-файл представлены словарём. Мы открываем файл `data.json` для записи и с помощью функции `json.dump()` записываем данные в формате JSON.

### 4.4. Итоги

В этом разделе мы разобрали, как обрабатывать файлы CSV и JSON в Python. Модуль `csv` предоставляет функции чтения и записи CSV-файлов, что позволяет работать с табличными данными. Модуль `json` предназначен для чтения и записи JSON-файлов, которые часто используют для хранения структурированных данных. Эти навыки помогают работать с данными разных форматов в программах на Python.
