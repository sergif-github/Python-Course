# Модуль 9. Работа с файлами

## Часть 1. Чтение и запись текстовых файлов

Чтение и запись текстовых файлов часто встречаются в программах на Python. В этом разделе мы разберём, как открывать и читать текстовые файлы и записывать в них данные средствами Python.

### 1.1. Открытие текстового файла

Чтобы открыть текстовый файл в Python, можно воспользоваться встроенной функцией `open()`. Она принимает путь к файлу и возвращает файловый объект, с помощью которого можно выполнять операции с файлом.

Откроем файл `example.txt` в режиме чтения.

````python
file = open("example.txt", "r")
````

В этом примере аргумент `"r"` открывает файл в режиме чтения. Другие режимы — `"w"` для записи, `"a"` для добавления в конец и `"x"` для создания нового файла при условии, что его ещё нет.

### 1.2. Чтение из текстового файла

После открытия текстового файла его содержимое можно прочитать с помощью методов файлового объекта `read()` и `readline()`.

Метод `read()` считывает всё содержимое файла в одну строку.

````python
file = open("example.txt", "r")
content = file.read()
print(content)
````

Метод `readline()` считывает из файла по одной строке за вызов.

````python
file = open("example.txt", "r")
line = file.readline()
print(line)
````

Можно также перебрать файловый объект в цикле, чтобы прочитать файл построчно.

````python
file = open("example.txt", "r")
for line in file:
    print(line)
````

### 1.3. Запись в текстовый файл

Чтобы записать текст в файл, откройте его в режиме записи с аргументом `"w"` или в режиме добавления в конец с аргументом `"a"`.

Откроем файл `output.txt` в режиме записи и запишем в него текст.

````python
file = open("output.txt", "w")
file.write("Hello, World!")
file.close()
````

В этом примере аргумент `"w"` открывает файл в режиме записи, а метод `write()` записывает в файл указанный текст.

Чтобы дописать текст в конец существующего файла, откройте его в режиме добавления с аргументом `"a"`.

````python
file = open("output.txt", "a")
file.write("Appending new content.")
file.close()
````

### 1.4. Закрытие файла

Завершив работу с файлом, закройте его с помощью метода файлового объекта `close()`. Это освобождает ресурсы, связанные с файлом.

````python
file = open("example.txt", "r")
file.close()
````

### 1.5. Операции с файлами в блоке `with`

Вместо явного закрытия можно использовать инструкцию `with`, которая автоматически закрывает файл после завершения работы.

```python
with open("example.txt", "r") as file:
    # Выполняем операции с файлом
```

Инструкцию `with` рекомендуется использовать, чтобы файл закрывался даже при исключении внутри блока.
