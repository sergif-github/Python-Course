# Модуль 9. Работа с файлами

## Часть 5. Обработка ошибок при работе с файлами

При работе с файлами важно учитывать ошибки, которые могут возникнуть во время операций. Например, файл может отсутствовать, у программы может не быть нужных прав или файл может быть недоступен, потому что его открыл другой процесс. В Python есть средства, которые помогают корректно обработать такие ситуации.

### 5.1. Обработка ошибок при работе с файлами

Чтобы обработать ошибки, связанные с файлами, можно использовать конструкцию `try-except`. В блоке `try` находится код, который может вызвать исключение. Если исключение возникает, его обрабатывает блок `except`.

Пример обработки исключения `FileNotFoundError`.

```python
try:
    with open("data.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("Файл не существует.")
```

В примере мы пытаемся открыть файл `data.txt` для чтения. Если файл отсутствует, возникает исключение `FileNotFoundError`. Мы перехватываем это исключение в блоке `except` и показываем своё сообщение об ошибке.

Аналогично можно обработать другие исключения при работе с файлами, например `PermissionError` и `IOError`, добавив блоки `except`.

### 5.2. Освобождение ресурсов с помощью `finally`

Блок `finally` позволяет задать код, который выполнится независимо от того, возникло исключение или нет. Его часто используют для освобождения ресурсов, например для закрытия файлов.

Пример закрытия файла с помощью `finally`.

```python
file = None
try:
    file = open("data.txt", "r")
    content = file.read()
    print(content)
except FileNotFoundError:
    print("Файл не существует.")
finally:
    if file is not None:
        file.close()
```

В этом примере блок `finally` выполнится даже при возникновении исключения и закроет файл, если его удалось открыть.

### 5.3. Итоги

В этом разделе мы рассмотрели обработку ошибок при работе с файлами. С помощью блоков `try-except` можно корректно обрабатывать ошибки и показывать подходящие сообщения. Блок `finally` позволяет освобождать ресурсы, например закрывать открытые файлы.
