# Модуль 9. Работа с файлами

## Часть 2. Работа с файловыми объектами

Для работы с файловыми объектами Python предоставляет методы и атрибуты для чтения и записи текстовых файлов и других операций. В этом разделе рассмотрим некоторые из них.

### 2.1. Методы файловых объектов

У файловых объектов в Python есть методы для работы с файлом. Вот несколько часто используемых методов.

- `read(size)`

    Читает и возвращает до `size` единиц данных из файла: символов в текстовом режиме или байтов в двоичном. Если размер не указан, читает всё оставшееся содержимое файла.

- `readline()`

    Читает и возвращает следующую строку из файла.

- `readlines()`

    Читает все оставшиеся строки и возвращает их в виде списка.

- `write(string)`

    Записывает указанную строку в файл.

- `writelines(lines)`

    Записывает в файл строки из списка `lines`.

- `seek(offset)`

    Перемещает текущую позицию в файле. В двоичном режиме смещение задаётся в байтах.

- `tell()`

    Возвращает текущую позицию в файле.

- `close()`

    Закрывает файл.

В следующем примере показано, как использовать некоторые из этих методов.

```python
file = open("example.txt", "r")
content = file.read()  # Читаем весь файл
line = file.readline()  # Читаем следующую строку
lines = file.readlines()  # Читаем оставшиеся строки в список
file.close()  # Закрываем файл
```

### 2.2. Атрибуты файловых объектов

У файловых объектов есть полезные атрибуты, которые предоставляют сведения о файле. Вот несколько часто используемых атрибутов.

- `name`

    Возвращает имя файла.

- `mode`

    Возвращает режим, в котором открыт файл.

- `closed`

    Возвращает `True`, если файл закрыт, и `False` в противном случае.

Обращаться к этим атрибутам можно через точку.

```python
file = open("example.txt", "r")
print(file.name)  # Выводим имя файла
print(file.mode)  # Выводим режим открытия файла
print(file.closed)  # Выводим признак закрытия файла
file.close()  # Закрываем файл
```

### 2.3. Использование инструкции `with`

Чтобы файл корректно закрывался, удобно использовать инструкцию `with`. Она автоматически закрывает файл, даже если возникает исключение.

Вот пример использования инструкции `with`.

```python
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

В этом примере файл автоматически закрывается при выходе из блока `with`, в том числе если выполнение блока прервалось исключением.

### 2.4. Итоги

В этом разделе мы рассмотрели некоторые методы и атрибуты файловых объектов Python. С их помощью можно читать и записывать данные, менять текущую позицию и получать информацию о файле. Закрывайте файл после завершения работы, чтобы освободить системные ресурсы. Инструкция `with` позволяет сделать это автоматически.
