# Модуль 9. Работа с файлами

## Часть 3. Режимы открытия файлов и пути

При работе с файлами в Python важно понимать режимы открытия и уметь работать с путями. В этом разделе рассмотрим оба понятия.

### 3.1. Режимы открытия файлов

Режим определяет, как открыть файл и какие операции с ним доступны. Чтобы указать режим, передайте его вторым аргументом функции `open()`. Ниже приведены распространённые режимы.

- `'r'` — чтение.

    Открывает файл для чтения. Если файл не существует, возникает ошибка.

- `'w'` — запись.

    Открывает файл для записи. Если файла нет, создаёт его. Если файл существует, очищает его содержимое.

- `'a'` — добавление в конец.

    Открывает файл для добавления данных. Позиция записи находится в конце файла, и новые данные добавляются туда.

- `'x'` — создание нового файла.

    Создаёт новый файл. Если файл уже существует, возникает ошибка.

- `'b'` — двоичный режим.

    Открывает файл в двоичном режиме, в котором можно читать или записывать двоичные данные.

- `'t'` — текстовый режим.

    Открывает файл в текстовом режиме, заданном по умолчанию. В этом режиме программа работает со строками Юникода, которые преобразуются в байты и обратно с помощью кодировки.

Некоторые обозначения режимов можно сочетать в одной строке. Например, `'rb'` открывает файл для чтения в двоичном режиме.

### 3.2. Пути к файлам

Путь к файлу указывает его расположение в файловой системе. В Python можно указывать относительный или абсолютный путь.

- Относительный путь `"example.txt"` указывает на файл в текущем каталоге.
- Абсолютный путь `"C:/Documents/example.txt"` задаёт полное расположение файла.

При работе с путями учитывайте операционную систему. Разделители в путях могут отличаться. Например, в Windows используется обратная косая черта (`\`), а в Unix-подобных системах — прямая (`/`).

Модуль `os` предоставляет функции для работы с путями независимо от операционной системы.

```python
import os

# Объединяем части пути к файлу
file_path = os.path.join("./", "example.txt")

# Проверяем, существует ли файл
if os.path.exists(file_path):
    print("Файл существует")
else:
    print("Файл не существует")
```

В этом примере функция `os.path.join()` объединяет путь к каталогу и имя файла. Функция `os.path.exists()` проверяет, существует ли файл по указанному пути.

Путь указывает расположение файла в файловой системе и может быть относительным или абсолютным. Модуль `os` предоставляет функции для работы с путями независимо от операционной системы.
