# Модуль 12. Тестирование и отладка

## Часть 2. Разработка через тестирование (TDD)

Разработка через тестирование (Test-Driven Development, TDD) — подход, при котором тесты пишут до кода. Сначала пишут тест, который не проходит, затем реализуют код, чтобы тест прошёл, и при необходимости выполняют рефакторинг. Такой подход помогает повысить качество и сопровождаемость кода и создавать надёжные приложения.

### 2.1. Цикл TDD

Обычно процесс состоит из трёх шагов.

1. **Напишите тест.** Начните с тестового случая, который описывает желаемое поведение определённой части кода. Сначала этот тест должен завершаться неудачей, так как код с таким поведением ещё не реализован.

2. **Напишите код.** После написания непроходящего теста реализуйте код, необходимый для его успешного прохождения. Старайтесь написать максимально простой код, удовлетворяющий требованиям теста.

3. **Выполните рефакторинг.** Когда тест проходит, можно улучшить структуру, производительность или читаемость кода, сохранив наблюдаемое поведение. Существующие тесты помогают убедиться, что изменения не привели к регрессиям.

### 2.2. Преимущества разработки через тестирование

Подход TDD даёт несколько преимуществ в процессе разработки.

- **Качество кода.** TDD побуждает писать читаемый, модульный и пригодный для тестирования код. Написание тестов помогает выявлять возможные проблемы на ранних этапах и поддерживать кодовую базу понятной и удобной для сопровождения.

- **Быстрая обратная связь.** Благодаря регулярным запускам тестов ошибки и регрессии обнаруживаются сразу, что позволяет своевременно их устранять и тратить меньше времени на отладку.

- **Улучшение проектирования.** Тесты описывают желаемое поведение кода и помогают принимать проектные решения. С ними проще определить зависимости, спроектировать интерфейсы и учесть крайние случаи. В результате код получается хорошо структурированным и слабо связанным.

- **Уверенность при изменениях.** Полноценный набор тестов помогает уверенно выполнять рефакторинг и менять существующий код. Если правки вызывают непредвиденные побочные эффекты, тесты помогают их обнаружить.

- **Документация.** Набор тестов служит живой документацией. Он показывает, как должен вести себя код, и содержит примеры использования. По тестам можно понять ожидаемые возможности кодовой базы.

Типичный рабочий цикл TDD выглядит так.

- **Красный.** Напишите тест, который сначала не проходит, поскольку проверяемая функциональность ещё не реализована.

- **Зелёный.** Реализуйте самый простой код, удовлетворяющий требованиям теста, чтобы тест прошёл.

- **Рефакторинг.** Улучшите код и следите, чтобы существующие тесты продолжали проходить. На этом шаге можно улучшить структуру, убрать дублирование и оптимизировать производительность.

- **Повторение.** Начните новый цикл с очередного непроходящего теста и реализуйте код, после которого тест пройдёт. Повторяйте этот процесс для каждой новой возможности или разновидности поведения.

### 2.3. Итоги

TDD — практика разработки, при которой тесты пишут до реализации. Цикл «непроходящий тест, реализация, рефакторинг» улучшает качество и сопровождаемость кода и повышает уверенность в работе программы. Он обеспечивает обратную связь, помогает принимать проектные решения и создаёт живую документацию. Применение TDD помогает разрабатывать надёжные и удобные для сопровождения приложения.
