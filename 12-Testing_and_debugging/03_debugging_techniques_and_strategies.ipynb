# Модуль 12. Тестирование и отладка

## Часть 3. Приёмы и стратегии отладки

Отладка — важный для разработчика навык. Она позволяет систематически находить и устранять ошибки и причины неожиданного поведения кода. В этом разделе рассмотрим приёмы и стратегии, которые помогают диагностировать и устранять проблемы в программах на Python.

### 3.1. Инструменты отладки

- **Вызовы `print()`.** Размещайте их в подходящих точках кода, чтобы отслеживать ход выполнения и просматривать значения переменных на разных этапах.

- **Отладчики.** В Python есть встроенный отладчик `pdb` (Python Debugger), а многие интегрированные среды разработки (IDE) предоставляют свои средства отладки. Они позволяют устанавливать точки останова, выполнять код пошагово и подробно изучать значения переменных.

### 3.2. Стратегии отладки

- **Воспроизведите проблему.** Добейтесь, чтобы проблема воспроизводилась стабильно. Определите шаги или условия, которые приводят к её появлению, и сузьте область поиска.

- **Изолируйте проблему.** Последовательно локализуйте проблемный участок. Временно закомментируйте или уберите части кода, чтобы выяснить, какая из них вызывает проблему. Так можно сократить число возможных причин.

- **Проверьте входные и выходные данные.** Убедитесь, что входные данные имеют ожидаемый формат и находятся в допустимом диапазоне. Сравните фактический результат с ожидаемым, чтобы обнаружить расхождения.

- **Изучите сообщения об ошибках и трассировки стека.** Они дают информацию о месте и характере ошибки. Обратите внимание на номера строк, типы ошибок и сопровождающие сообщения.

- **Примените двоичный поиск.** При работе с большой кодовой базой временно разделите код пополам и проверьте, возникает ли проблема. По результату выберите нужную половину и повторяйте процесс, пока не найдёте проблемный участок.

- **Тестируйте поэтапно.** Проверяйте код после добавления каждой новой возможности или изменения. Так можно рано обнаружить ошибки и быстрее выяснить их причину.

- **Попробуйте метод резиновой уточки (rubber duck debugging).** Объясните свой код и проблему неодушевлённому предмету, например резиновой уточке, или другому человеку. Формулируя проблему вслух, вы можете заметить то, что раньше упускали.

- **Документация и журналирование.** Добавляйте записи в журнал, чтобы отслеживать ход выполнения и сохранять важные сведения о работе программы. Журнал помогает выяснить значения переменных, результаты проверок условий и последовательность выполнения инструкций.

- **Интернет-ресурсы и сообщества.** При сложных проблемах обращайтесь к онлайн-ресурсам, форумам и сообществам разработчиков. Другие люди могли столкнуться с похожей проблемой и подсказать способ её решения.

### 3.3. Чтение и разбор сообщений об ошибках

- Внимательно читайте сообщения об ошибках. Они часто подсказывают, что произошло и где. Разберите тип ошибки и текст сообщения, чтобы точнее локализовать проблему.

- Обращайте внимание на номера строк в сообщениях. Они указывают место, где возникла ошибка, и помогают найти проблемный код.

- Ищите в интернете текст сообщения или тип исключения, чтобы найти подходящие обсуждения и документацию с пояснениями.

### 3.4. Тестирование и проверка входных данных

- Проверяйте, соответствуют ли введённые пользователем данные ожидаемому формату и диапазону. Применяйте условные инструкции и подходящие способы проверки, чтобы обработать разные ситуации.

- Тестируйте код на разных входных значениях. Включайте крайние случаи и граничные условия, чтобы выявить проблемы, возникающие только при отдельных значениях или особых обстоятельствах.

### 3.5. Журналирование и отладочный вывод

- Записывайте в журнал ход выполнения, значения переменных и важные сведения о работе программы.

- Размещайте записи в журнале в ключевых точках кода, чтобы получить информацию о состоянии переменных и результатах проверки условий.

- Используйте отладочные вызовы `print()` или уровни журналирования, чтобы по необходимости выводить отладочные сведения или отключать их.

### 3.6. Парное программирование и ревью кода

- Работайте в паре или попросите коллег проверить код. Свежий взгляд и совместная работа помогают заметить ошибки, которые можно пропустить самостоятельно.

- Проверка чужого кода развивает и ваши навыки отладки. Она знакомит с другими подходами и приёмами решения задач.

### 3.7. Документация и организация кода

- Пишите читаемый, модульный и хорошо организованный код. Поддерживайте документацию и комментарии в актуальном состоянии, чтобы было проще понять назначение и поведение кода.

- Давайте переменным и функциям понятные имена, отражающие их назначение. Это помогает быстрее находить и устранять проблемы.

### 3.8. Итоги

Отладка — важный навык разработчика. Инструменты отладки, воспроизведение и локализация проблем, разбор сообщений об ошибках, проверка входных данных и ведение журнала помогают находить и устранять ошибки. Поддержка сообществ и ревью кода тоже развивают навыки отладки. Подходите к отладке системно, обращайте внимание на детали и пользуйтесь доступными средствами, чтобы эффективно диагностировать и решать проблемы.
