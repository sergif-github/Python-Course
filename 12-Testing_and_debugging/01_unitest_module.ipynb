# Модуль 12. Тестирование и отладка

## Часть 1. Модульное тестирование с `unittest`

Модульное тестирование (unit testing) — важная практика разработки, при которой отдельные части кода проверяют на корректность и соответствие ожидаемому поведению. Модуль `unittest` в Python предоставляет средства для написания и запуска модульных тестов, в том числе набор проверок для создания полноценных тестовых случаев.

### 1.1. Основные компоненты модуля `unittest`

- **Тестовый случай (test case).** Класс тестового случая наследуется от `unittest.TestCase` и представляет проверку отдельной части кода. Обычно он содержит несколько тестовых методов, проверяющих разные аспекты поведения кода.

- **Тестовый метод.** Метод класса тестового случая, который выполняет отдельную проверку участка кода. Чтобы средство запуска `unittest` автоматически находило и выполняло тестовый метод, его имя должно начинаться с префикса `test_`.

- **Утверждения (assertions).** Проверки того, что заданное условие выполняется. В `unittest` есть методы для проверки ожидаемого результата, сравнения значений, проверки возникновения исключений и других задач.

### 1.2. Написание модульных тестов с `unittest`

Чтобы создать модульные тесты с помощью `unittest`, выполните следующие шаги.

- **Импортируйте модуль `unittest`.** Начните с импорта модуля в тестовом скрипте.

- **Определите тестовый случай.** Создайте подкласс `unittest.TestCase` и определите в нём тестовые методы. Имя каждого такого метода должно начинаться с префикса `test_`. В теле метода используйте утверждения для проверки ожидаемого поведения.

- **Запустите тесты.** Используйте средство запуска `unittest`, чтобы найти и выполнить тестовые методы. Для этого можно вызвать `unittest.main()` в конце тестового скрипта или использовать другие средства модуля `unittest`.

```python
import unittest

# Пример тестируемого кода
def add_numbers(a, b):
    return a + b

# Класс тестового случая
class TestAddNumbers(unittest.TestCase):

    def test_add_numbers(self):
        result = add_numbers(2, 3)
        self.assertEqual(result, 5)  # Проверяем, что результат равен 5

        result = add_numbers(10, -5)
        self.assertEqual(result, 5)  # Проверяем, что результат равен 5

        result = add_numbers(0, 0)
        self.assertEqual(result, 0)  # Проверяем, что результат равен 0

if __name__ == '__main__':
    unittest.main()
```

В этом примере `TestAddNumbers` наследуется от `unittest.TestCase` и содержит один тестовый метод `test_add_numbers`. Он вызывает функцию `add_numbers` с разными входными значениями и использует `self.assertEqual()`, чтобы проверить совпадение результатов с ожидаемыми значениями.

При запуске скрипта средство запуска `unittest` выполняет тестовые методы класса `TestAddNumbers` и выводит результаты.

### 1.3. Преимущества модульного тестирования с `unittest`

- **Проверка корректности кода.** Модульные тесты помогают рано обнаруживать и исправлять ошибки и подтверждают, что код ведёт себя как ожидается.
- **Упрощение рефакторинга.** Тесты помогают увереннее менять код, поскольку быстро выявляют регрессии, возникшие в ходе рефакторинга.
- **Упрощение сопровождения.** Хорошо написанные модульные тесты служат документацией и помогают понять ожидаемое поведение кода.
- **Поддержка разработки через тестирование (TDD).** С помощью `unittest` разработчики могут придерживаться TDD и писать тесты до реализации кода.

### 1.4. Итоги

Модуль `unittest` позволяет писать модульные тесты на Python. Тестовые случаи с достаточным охватом повышают качество и надёжность кода и упрощают отладку и разработку.
