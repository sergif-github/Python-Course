## Часть 3. Функции высшего порядка

Функции высшего порядка могут принимать другие функции в качестве аргументов, возвращать функции в качестве результата или делать и то и другое. В Python это возможно, потому что функции являются объектами первого класса. Рассмотрим свойства функций высшего порядка и то, как с их помощью делать код гибким и модульным.

### 3.1. Приём функций в качестве аргументов

Функция высшего порядка может принимать другие функции в качестве аргументов. Так можно выделить часть поведения и предоставить вызывающему коду возможность задавать или настраивать его. Внутри функции высшего порядка переданную функцию можно вызвать или использовать другим способом, чтобы выполнить нужную операцию или преобразование.

```python
def apply_operation(func, x, y):
    return func(x, y)

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

result = apply_operation(add, 5, 3)
print(result)  # Вывод: 8

result = apply_operation(subtract, 10, 4)
print(result)  # Вывод: 6
```

В этом примере функция высшего порядка `apply_operation` принимает функцию `func` и два значения `x` и `y`. Она применяет переданную функцию к этим значениям и возвращает результат. В первом вызове в `apply_operation` передаётся функция `add` для сложения, а во втором — `subtract` для вычитания.

### 3.2. Возврат функций в качестве результата

Функции высшего порядка могут также создавать новые функции и возвращать их. Это позволяет получать специализированные функции для конкретных задач или настроек. Возвращённую функцию можно присвоить переменной или сразу использовать в дальнейших вычислениях.

```python
def create_multiplier(n):
    def multiplier(x):
        return x * n
    return multiplier

double = create_multiplier(2)
triple = create_multiplier(3)

print(double(5))  # Вывод: 10
print(triple(5))  # Вывод: 15
```

В этом примере функция высшего порядка `create_multiplier` принимает аргумент `n` и возвращает новую функцию `multiplier`. Возвращённая функция умножает свой аргумент на `n`. Вызывая `create_multiplier` с разными значениями, мы получаем специализированные функции `double` и `triple`. Они умножают числа на 2 и 3 соответственно.

### 3.3. Композиция функций

С помощью функций высшего порядка можно выполнять композицию функций, то есть объединять несколько функций в одну новую. Композиция позволяет последовательно связывать операции, упрощая чтение и понимание сложного кода. Для композиции можно передавать функции в качестве аргументов или возвращать функции из других функций.

```python
def compose(func1, func2):
    def composed_function(x):
        return func1(func2(x))
    return composed_function

def add_one(x):
    return x + 1

def multiply_by_two(x):
    return x * 2

composed = compose(add_one, multiply_by_two)

result = composed(5)
print(result)  # Вывод: 11
```

В этом примере функция высшего порядка `compose` принимает две функции, `func1` и `func2`, и возвращает новую функцию `composed_function`. Возвращённая функция применяет `func2` к аргументу `x`, а затем применяет `func1` к полученному результату. Объединив `add_one` и `multiply_by_two` с помощью `compose`, мы получаем функцию `composed`, которая сначала умножает число на два, а затем прибавляет единицу.

### 3.4. Итоги

Функции высшего порядка позволяют работать с функциями как с объектами. Они помогают писать гибкий и модульный код, в котором поведение можно выделять, настраивать и комбинировать. Понимание функций высшего порядка позволяет писать выразительный код, пригодный для повторного использования, и применять более сложные приёмы, в том числе приёмы функционального программирования.
