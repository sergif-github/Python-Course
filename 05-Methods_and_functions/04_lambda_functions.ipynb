## Часть 4. Лямбда-функции

Лямбда-функции, или анонимные функции, — это короткие функции, которые можно определять прямо в выражении и оставлять без имени. Их часто используют с функциями высшего порядка, например с `map()`, `filter()`, `sorted()` и `reduce()`. В этой части рассмотрим лямбда-функции и их применение вместе с такими функциями.

### 4.1. Лямбда-функции

Лямбда-функция определяется с помощью ключевого слова `lambda`, за которым следуют параметры и одно выражение. Количество параметров может быть любым, а тело состоит из одного выражения.

Синтаксис:
```python
lambda arguments: expression
```

````python
add = lambda x, y: x + y
result = add(2, 3)
print(result)  # Вывод: 5
````

В этом примере определена лямбда-функция с двумя параметрами `x` и `y`. Функция возвращает сумму двух аргументов. Лямбда-функция присваивается переменной `add`. При её вызове с аргументами `2` и `3` выводится `5`.

### 4.2. Лямбда-функции в функциях высшего порядка

#### 4.2.1. map(function, iterable)

Применяет переданную функцию к каждому элементу итерируемого объекта и возвращает итератор с преобразованными значениями.

````python
numbers = [1, 2, 3, 4, 5]
squares = map(lambda x: x ** 2, numbers)
print(list(squares))  # Вывод: [1, 4, 9, 16, 25]
````

С помощью `map()` лямбда-функция `lambda x: x ** 2` применяется к каждому элементу списка `numbers`. После преобразования итератора в список получаем квадраты исходных чисел.

#### 4.2.2. filter(function, iterable)

Фильтрует итерируемый объект с помощью переданной функции и возвращает итератор по элементам, которые удовлетворяют условию фильтрации.

````python
numbers = [1, 2, 3, 4, 5]
even_numbers = filter(lambda x: x % 2 == 0, numbers)
print(list(even_numbers))  # Вывод: [2, 4]
````

Лямбда-функция `lambda x: x % 2 == 0` передаётся в `filter()`, чтобы отобрать чётные числа из списка `numbers`. Полученный итератор преобразуется в список, содержащий только чётные числа.

#### 4.2.3. sorted(iterable, key=None, reverse=False)

Возвращает новый отсортированный список из элементов итерируемого объекта. Можно задать собственный ключ сортировки и указать, нужен ли обратный порядок.

````python
numbers = [5, 2, 8, 1, 3]
sorted_numbers = sorted(numbers, key=lambda x: x % 3)
print(sorted_numbers)  # Вывод: [3, 1, 5, 2, 8]
````

В `sorted()` лямбда-функция `lambda x: x % 3` используется как ключ сортировки. Элементы списка `numbers` упорядочиваются по остатку от деления каждого числа на 3.

#### 4.2.4. reduce(function, iterable[, initializer])

Свёртывает итерируемый объект в одно значение. Функция с двумя аргументами последовательно применяется слева направо к накопленному результату и очередному элементу.

````python
from functools import reduce

numbers = [1, 2, 3, 4, 5]
product = reduce(lambda x, y: x * y, numbers)
print(product)  # Вывод: 120
````

Лямбда-функция `lambda x, y: x * y` передаётся в `reduce()`, чтобы вычислить произведение всех чисел из списка `numbers`.

### 4.3. Итоги

Лямбда-функции позволяют компактно определять небольшие однострочные функции. Вместе с функциями высшего порядка, такими как `map()`, `filter()`, `sorted()` и `reduce()`, их можно использовать для преобразования, фильтрации, сортировки и агрегирования данных. Умение применять лямбда-функции вместе с функциями высшего порядка помогает писать более выразительный и компактный код на Python.
