# Модуль 5. Методы и функции

## Часть 5. Упражнения

### Упражнение 1
Напишите функцию `greet()` на Python, которая выводит сообщение с приветствием.

### Упражнение 2
Напишите функцию `calculate_area()` на Python, которая вычисляет площадь прямоугольника. У функции должно быть два параметра, `length` и `width`, со значениями по умолчанию `10`.

### Упражнение 3
Напишите функцию `add_numbers()` на Python, которая принимает три параметра (`num1`, `num2` и `num3`) и возвращает сумму этих трёх чисел.

### Упражнение 4
Напишите функцию `divide_numbers()` на Python, которая принимает два параметра (`numerator` и `denominator`) и возвращает и частное, округлённое вниз, и остаток от деления.

### Упражнение 5
Напишите функцию `calculate_average` на Python, которая принимает список чисел в качестве аргумента и возвращает среднее арифметическое этих чисел.

### Упражнение 6
Напишите функцию `change_global_variable` на Python, которая показывает, как использовать ключевое слово `global` для изменения глобальной переменной внутри функции.

### Упражнение 7
Напишите функцию `apply_operation` на Python, которая принимает два числа и функцию в качестве аргументов. Она должна применить переданную функцию к двум числам и вернуть результат.

### Упражнение 8
Напишите программу на Python, которая с помощью лямбда-функции и функции `map()` возводит в квадрат каждое число в заданном списке.
