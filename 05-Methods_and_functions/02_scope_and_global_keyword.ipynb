## Часть 2. Область видимости переменных и ключевое слово `global`

В Python область видимости переменной — это часть кода, в которой её имя известно и к ней можно обратиться. Знание областей видимости помогает понять, где доступны переменные и как долго они существуют в контексте вызова функции.

Разберём области видимости и использование ключевого слова `global` в Python.

### 2.1. Локальные переменные

Переменные, определённые внутри функции, имеют локальную область видимости. Они доступны в пределах функции, в которой определены. После завершения вызова функции её локальные переменные исчезают.

Рассмотрим пример:

```python
def my_function():
    x = 10
    print(x)

my_function()
```

В этом примере переменная `x` — локальная. Она определена в функции `my_function()`, и обратиться к ней можно только в области видимости этой функции. При вызове функции значение `x` выводится на экран.

### 2.2. Глобальные переменные

Переменные, определённые вне функций, имеют глобальную область видимости в своём модуле. К ним можно обращаться из разных частей этого модуля, в том числе из его функций. В глобальных переменных можно хранить значения, нужные в нескольких частях кода.

Рассмотрим пример:

```python
x = 10

def my_function():
    print(x)

my_function()
```

В этом примере `x` — глобальная переменная, определённая вне функции `my_function()`. Обратиться к ней можно и внутри функции. При вызове функции значение `x` выводится на экран.

### 2.3. Ключевое слово `global`

Чтобы присвоить глобальной переменной новое значение внутри функции, используйте ключевое слово `global`. Оно указывает, что присваивание относится к глобальной переменной и не создаёт новую локальную переменную.

Рассмотрим пример:

```python
x = 10

def modify_global():
    global x
    x = 20

modify_global()
print(x)
```

В этом примере ключевое слово `global` в функции `modify_global()` указывает, что присваивание относится к глобальной переменной `x`. Её значение меняется на `20`. Вызов `print(x)` вне функции выводит обновлённое значение.

### 2.4. Итоги

В этой части вы изучили области видимости переменных и ключевое слово `global` в Python. Локальные переменные определяются внутри функций и видны в пределах соответствующей функции. Глобальные переменные определяются вне функций и доступны из разных частей модуля. Для присваивания глобальной переменной внутри функции используется ключевое слово `global`. Понимание областей видимости и правил работы с глобальными переменными помогает писать модульный код, который проще сопровождать.
