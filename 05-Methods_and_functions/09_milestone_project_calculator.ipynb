# Модуль 5. Методы и функции

## Часть 9. Итоговый проект «Калькулятор»

### Проект «Простой калькулятор»

В этом проекте вы на практике поработаете с пользовательским вводом, базовыми арифметическими операциями, условными инструкциями, циклами и функциями в Python. Вы также попрактикуетесь в организации программы и написании понятного кода.

Описание

Создайте программу, которая имитирует работу простого калькулятора и выполняет основные арифметические операции.

Рабочая заготовка проекта находится в файле [`Projects/01-Calculator/calculator.py`](../Projects/01-Calculator/calculator.py). Реализуйте функции в этом файле, а затем запустите базовую проверку командой `python3 Projects/01-Calculator/check_calculator.py`.

Требования

- Программа должна выводить меню с доступными операциями: сложение, вычитание, умножение, деление и выход.
- Пользователь должен иметь возможность выбрать операцию, введя соответствующий номер или символ.
- Если пользователь выбрал недопустимую операцию или ввёл некорректные данные, программа должна вывести подходящее сообщение об ошибке.
- После выбора операции программа должна предложить пользователю ввести два числа.
- Программа должна выполнить выбранную операцию над введёнными числами и вывести результат.

Дополнительные возможности (необязательно)

- Предусмотрите возможность продолжать вычисления без выхода из программы после каждой операции.
- Добавьте операцию возведения в степень.

Что учесть

- Разбейте программу на небольшие функции или методы, чтобы код было проще организовать и читать.
- Выберите подходящие типы данных для вычислений и корректно обработайте граничные случаи.
- Проверьте точность вычислений калькулятора на разных входных данных.
