# Модуль 5. Методы и функции

## Часть 7. Итоговый проект «Простой калькулятор оценок»

### Проект «Калькулятор оценок»

В этом проекте вы поработаете с переменными, типами данных, операторами и инструкциями управления потоком выполнения в Python. Вы также попрактикуетесь в математических вычислениях, обработке ввода и форматировании вывода.

Описание

Создайте программу, которая вычисляет среднюю оценку учащегося по баллам за разные предметы. Программа должна предложить пользователю ввести балл за каждый предмет и вычислить общую среднюю оценку.

Требования

- Предложите пользователю ввести количество предметов.
- Запросите оценку по каждому предмету.
- Выведите пользователю среднюю оценку.

Дополнительные возможности (необязательно)

- По средней оценке определите и выведите буквенную оценку, например A, B или C.

Что учесть

- Выберите подходящие типы данных для хранения оценок и вычисления среднего значения.
- Обрабатывайте разные виды оценок, например целые числа и числа с плавающей точкой.
- Используйте арифметические операторы для вычислений и операторы сравнения для сопоставления значений.
- Используйте инструкции управления потоком выполнения, например циклы `for` или `while`, чтобы обработать несколько предметов.
- Отформатируйте вывод так, чтобы средняя оценка и любые дополнительные показатели были хорошо читаемы.
