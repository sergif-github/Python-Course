# Модуль 5. Методы и функции

## Часть 1. Определение и вызов функций

В Python функции — это многократно используемые блоки кода, каждый из которых выполняет определённую задачу. Функции позволяют разбить программу на более мелкие части. Так код проще организовать и понять.

Рассмотрим, как определять и вызывать функции в Python.

### 1.1. Определение функции

Для определения функции используется ключевое слово `def`. После него указывают имя функции, круглые скобки и двоеточие. Код, который следует за строкой определения с отступом, составляет тело функции.

Вот пример простой функции, которая приветствует пользователя:

```python
def greet():
    print("Здравствуйте, добро пожаловать!")
```

В этом примере функция `greet()` определена без параметров. Её тело состоит из одной инструкции, которая выводит приветствие.

### 1.2. Вызов функции

Определив функцию, можно вызвать её, чтобы выполнить код из её тела. Для вызова укажите имя функции, а после него — круглые скобки.

Вот как можно вызвать определённую ранее функцию `greet()`:

```python
greet()
```

При вызове функции `greet()` выполняется код из её тела и на экране появляется приветствие `"Здравствуйте, добро пожаловать!"`.

### 1.3. Параметры функции

Функции могут принимать параметры. Параметры — это имена, которым при вызове функции присваиваются значения переданных аргументов.
Благодаря параметрам функция может работать с разными входными данными, а код становится гибче и пригоднее для повторного использования.

Вот пример функции с одним параметром:

```python
def greet(name):
    print("Здравствуйте,", name, "добро пожаловать!")
```

В этом примере у функции `greet()` один параметр — `name`. При вызове функции передайте аргумент для этого параметра. Функция подставит полученное значение в персональное приветствие.

### 1.4. Значения параметров по умолчанию

В определении функции можно задать значения параметров по умолчанию. Такое значение используется, если при вызове функции аргумент для соответствующего параметра не передан.

Рассмотрим пример:

```python
def greet(name="guest"):
    print("Здравствуйте,", name, "добро пожаловать!")
```

В этом примере параметру `name` функции `greet()` задано значение по умолчанию `"guest"`. При вызове функции без аргумента будет использовано это значение. Переданный аргумент имеет приоритет перед значением по умолчанию.

### 1.5. Несколько параметров

У функции может быть несколько параметров. Это позволяет передавать в неё несколько входных значений. В круглых скобках параметры разделяются запятыми.

Рассмотрим пример:

```python
def add_numbers(a, b):
    return a + b
```

### 1.6. Инструкция `return`

С помощью инструкции `return` функция может возвращать значения. Функция вычисляет результат и передаёт его вызвавшему её коду.

Рассмотрим пример:

```python
def add_numbers(a, b):
    return a + b
```

В этом примере функция `add_numbers()` принимает два параметра, `a` и `b`, и возвращает их сумму с помощью инструкции `return`.
Возвращаемое значение можно сохранить в переменной или использовать в другой части программы.

Если в функции нет инструкции `return` либо выполняется `return` без значения, функция возвращает `None`. Это специальное значение Python, которое обозначает отсутствие значения.

Рассмотрим пример:

```python
def say_hello():
    print("Привет!")

result = say_hello()
print(result)
```

В этом примере в функции `say_hello()` нет инструкции `return`. При вызове она выводит `"Привет!"` и не возвращает значение явно. Переменной `result` присваивается `None`, после чего это значение выводится на экран.

### 1.7. Итоги

В этой части вы научились определять и вызывать функции в Python. Функции — это многократно используемые блоки кода. С их помощью можно разбить программу на небольшие части, которыми проще управлять. У функции может не быть параметров, или их может быть один или несколько. Чтобы вернуть значение из функции, используйте инструкцию `return`. Функции помогают организовать код, сделать его модульным и более понятным.
