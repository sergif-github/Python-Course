### 5.1. Измерение времени выполнения

Измерение времени выполнения помогает выявить узкие места и оценить эффективность разных реализаций. В Python время можно измерять, например, функцией `time.time()` из модуля `time` и средствами модуля `timeit`.

Чтобы измерить время выполнения конкретного блока кода, можно использовать модуль `time` следующим образом.

```python
import time

start_time = time.time()

# Код, время выполнения которого измеряем
for i in range(1000000):
    pass

end_time = time.time()
execution_time = end_time - start_time

print(f"Время выполнения: {execution_time} с")
```

В этом примере мы импортируем модуль `time` и запоминаем время начала с помощью `time.time()`. Затем выполняем код, время которого измеряем, здесь это простой цикл. После выполнения вычисляем прошедшее время, вычитая время начала из текущего времени, полученного через `time.time()`. Наконец, выводим время выполнения в секундах.

Для более точных измерений времени выполнения можно воспользоваться модулем `timeit`.

```python
import timeit

def func():
    i = 1000
    for x in range(1000):
        i -= 1

execution_time = timeit.timeit(func, number=5)
print(f"Время выполнения: {execution_time} с")
```

В этом примере мы помещаем измеряемый код в функцию `func()`. Время его выполнения измеряет функция `timeit.timeit()`. Аргумент `number` задаёт число повторных запусков, которые помогают получить более точную оценку времени.

### 5.2. Оптимизация кода

Оптимизация кода направлена на повышение производительности программы за счёт сокращения времени выполнения или расхода памяти. Распространённые подходы включают оптимизацию алгоритмов, структур данных и отдельных участков кода. Подбор алгоритма и структуры данных для конкретной задачи может существенно повысить производительность.

Вот несколько примеров эффективных алгоритмов и структур данных.

- Хеш-таблицы, например словари Python, обеспечивают быстрый поиск значения по ключу.
- Двоичный поиск позволяет эффективно искать элементы в отсортированных списках и массивах.
- Динамическое программирование помогает избежать повторных вычислений в рекурсивных задачах.
- Кучи и очереди с приоритетом позволяют эффективно добавлять и извлекать элементы.
- Понимание алгоритмической сложности помогает сравнивать алгоритмы и выбирать подходящий для конкретной задачи. Нотация «большое O» (Big O) задаёт асимптотическую верхнюю границу временной или пространственной сложности. Так можно описать, например, сложность в худшем случае. Нотация показывает, как меняется время выполнения или расход памяти с ростом размера входных данных. При этом учитывается доминирующий член, а постоянные множители и члены меньшего порядка опускаются.
    - `O(1)` — константное время. Время выполнения не зависит от размера входных данных.
    - `O(log n)` — логарифмическое время. Время выполнения растёт логарифмически относительно размера входных данных.
    - `O(n)` — линейное время. Время выполнения растёт линейно с размером входных данных.
    - `O(n log n)` — линейно-логарифмическое время. Время выполнения растёт пропорционально размеру входных данных, умноженному на его логарифм.
    - `O(n^2)` — квадратичное время. Время выполнения растёт пропорционально квадрату размера входных данных.
    - `O(2^n)` — экспоненциальное время. Время выполнения растёт экспоненциально с размером входных данных.

Помимо улучшения алгоритмов можно применять несколько приёмов на уровне кода.

- Сводите к минимуму лишние операции и избегайте повторных вычислений.
- Используйте встроенные функции и библиотеки там, где они уже решают задачу.
- Оптимизируйте циклы, уменьшая число итераций или применяя векторные операции.
- Снижайте расход памяти, подбирая структуры данных и используя генераторы или итераторы там, где это подходит.
- Обрабатывайте строки эффективно. Например, объединяйте их с помощью `join()`, чтобы избежать конкатенации в цикле.
- Профилируйте и анализируйте код, чтобы выявить узкие места и сосредоточить оптимизацию на критических участках.

Оптимизируйте код обдуманно. Преждевременная оптимизация усложняет код и его сопровождение, поэтому стоит сосредоточиться на участках, которые существенно влияют на производительность.

```python
def sum_of_squares(n):
    return sum([i**2 for i in range(1, n+1)])

print(sum_of_squares(100))
```

Здесь мы определяем функцию `sum_of_squares`, вычисляющую сумму квадратов чисел от 1 до `n`. В первой реализации списковое включение создаёт список квадратов чисел от 1 до `n`, а функция `sum()` суммирует их. Временная сложность этой реализации равна `O(n)`. В оптимизированном варианте можно вычислить сумму квадратов по формуле за `O(1)` арифметических операций.

```python
def process_data(data):
    result = []

    for item in data:
        # Выполняем обработку элемента
        processed_item = item * 2
        result.append(processed_item)

    return result

data = [1, 2, 3, 4, 5]
processed_data = process_data(data)
```

Функция `process_data` обрабатывает элементы списка данных. Текущая реализация использует цикл `for` и добавляет каждый обработанный элемент в список результатов. В оптимизированной версии можно использовать списковое включение или генераторное выражение, чтобы убрать явный цикл. Генераторное выражение позволяет снизить расход памяти, если результаты нужны по одному.
