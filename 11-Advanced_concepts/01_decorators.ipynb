## Часть 1. Декораторы функций и классов

Декораторы — средство Python, позволяющее изменять поведение функций и классов без правок в их исходном коде. С их помощью можно компактно добавить новые возможности, изменить входные данные или результаты, а также выполнять действия до или после вызова функции или класса. В этом разделе разберём декораторы функций и классов.

### 1.1. Декораторы функций

Декоратор функции принимает другую функцию и оборачивает её, меняя её поведение. Запись `@decorator_name` помещают над определением функции. Так к обёрнутой функции можно добавить, например, журналирование, измерение времени, кэширование или проверку входных данных.

```python
def uppercase_decorator(func):
    def wrapper(text):
        result = func(text.upper())
        return result

    return wrapper

@uppercase_decorator
def greet(name):
    return f"Hello, {name}!"

print(greet("John"))  # Вывод: Hello, JOHN!
```

Здесь мы определяем декоратор функции `uppercase_decorator`. Он принимает функцию `func` и определяет функцию-обёртку `wrapper`, которая переводит входной текст в верхний регистр перед вызовом `func`. Декоратор возвращает обёртку. Применим `uppercase_decorator` к функции `greet` с помощью синтаксиса `@`. При вызове `greet("John")` декоратор преобразует имя в `"JOHN"` до того, как функция вернёт приветствие.

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        # Выполняем действия перед вызовом функции
        print("Декоратор: перед вызовом функции")

        # Вызываем обёрнутую функцию
        result = func(*args, **kwargs)

        # Выполняем действия после вызова функции
        print("Декоратор: после вызова функции")

        # Возвращаем результат обёрнутой функции
        return result

    return wrapper

@decorator
def greeting(name):
    print("Привет,", name)

greeting("John")
```

В этом фрагменте определён декоратор `decorator`. Он принимает функцию `func`, определяет обёртку `wrapper`, выполняющую действия до и после вызова функции, и возвращает её. Записью `@decorator` мы применяем декоратор к функции `greeting`. При вызове `greeting` с именем `"John"` действия декоратора выполняются перед вызовом исходной функции и после него.

### 1.2. Декораторы классов

Декораторы классов похожи на декораторы функций и работают с классами. Они оборачивают класс, меняют его поведение или добавляют новые возможности. Запись `@decorator_name` располагается перед определением класса. Декораторы могут подключать примеси (mixins), накладывать ограничения на уровне класса или изменять его атрибуты и методы.

```python
def add_property(cls):
    cls.new_property = "New Property"
    return cls

@add_property
class MyClass:
    pass

obj = MyClass()
print(obj.new_property)  # Вывод: New Property
```

Здесь мы определяем декоратор класса `add_property`. Он принимает класс `cls`, добавляет ему атрибут `new_property` и возвращает изменённый класс. Применим `add_property` к классу `MyClass` с помощью синтаксиса `@`. Если создать экземпляр `MyClass` и обратиться к атрибуту `new_property`, получим значение `"New Property"`, добавленное декоратором.

```python
def decorator(cls):
    class WrapperClass:
        def __init__(self, *args, **kwargs):
            self.wrapped = cls(*args, **kwargs)

        def greet(self):
            print("Декоратор: перед приветствием")
            self.wrapped.greet()
            print("Декоратор: после приветствия")

    return WrapperClass

@decorator
class Greeting:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Привет,", self.name)

obj = Greeting("John")
obj.greet()
```

В этом примере декоратор `decorator` принимает класс `cls`. Он определяет класс-обёртку `WrapperClass`, который оборачивает исходный класс и изменяет его поведение. Метод `__init__` класса `WrapperClass` создаёт экземпляр обёрнутого класса, а его метод `greet` выполняет действия перед вызовом метода `greet` обёрнутого класса и после него. Запись `@decorator` применяет декоратор к классу `Greeting`. Когда мы создаём экземпляр `Greeting` и вызываем его метод `greet`, действия декоратора выполняются до и после метода исходного класса.

### 1.3. Итоги

Декораторы функций и классов позволяют гибко изменять поведение этих объектов в Python. Декораторы функций оборачивают функции и позволяют добавить возможности или изменить входные данные и результаты. Декораторы классов оборачивают классы и позволяют менять их поведение, атрибуты или методы. Декораторы помогают повторно использовать код, разделять ответственность и расширять программу. С ними можно дополнить возможности функций и классов, сохранив исходный код.
