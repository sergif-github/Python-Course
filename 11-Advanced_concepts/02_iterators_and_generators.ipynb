## Часть 2. Итераторы и генераторы

Итераторы и генераторы позволяют удобно работать с последовательностями данных и экономить память. С их помощью можно перебирать коллекцию или получать значения по мере необходимости, без загрузки всех данных в память сразу. В этом разделе рассмотрим применение и преимущества итераторов и генераторов.

### 2.1. Итераторы

Итератор — объект, реализующий протокол итератора. Он позволяет последовательно получать элементы коллекции или другие значения по одному.

```python
numbers = [1, 2, 3, 4, 5]
iterator = iter(numbers)

print(next(iterator))  # Вывод: 1
print(next(iterator))  # Вывод: 2
print(next(iterator))  # Вывод: 3
```

В этом фрагменте мы создаём список чисел. Функция `iter()` получает из списка объект-итератор. Функция `next()` извлекает очередное значение из итератора. При каждом следующем вызове `next()` возвращает очередное значение, пока элементы итератора не закончатся.

В протоколе итерации используются методы `__iter__()` и `__next__()`. Реализуя их, можно сделать объект итерируемым и определить, как он ведёт себя при переборе.

- Метод `__iter__()` возвращает объект-итератор. Он вызывается в начале перебора объекта. При необходимости метод инициализирует состояние и возвращает итератор. Для самого итератора этот метод возвращает его самого.
- Метод `__next__()` вызывается у объекта-итератора и возвращает очередное значение. Когда значения заканчиваются, метод должен вызвать исключение `StopIteration`.

Протокол итерации позволяет создавать собственные итерируемые объекты и определять порядок перебора и значения, возвращаемые на каждом шаге.

Знание протокола итерации необходимо для создания сложных итерируемых объектов и работы с собственными итераторами и генераторами в Python.

Рассмотрим пример собственного итератора для списка чисел.

```python
class NumberIterator:
    def __init__(self, numbers):
        self.numbers = numbers
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.numbers):
            raise StopIteration
        value = self.numbers[self.index]
        self.index += 1
        return value

numbers = [1, 2, 3, 4, 5]
iterator = NumberIterator(numbers)

for number in iterator:
    print(number)
```

В этом примере определён класс `NumberIterator`, принимающий список чисел. Его метод `__iter__()` возвращает сам объект-итератор, а метод `__next__()` извлекает очередное значение из списка. Когда значения заканчиваются, возникает исключение `StopIteration`. Объект `NumberIterator` можно передать циклу `for`, чтобы перебрать все числа списка.

### 2.2. Генераторы

Генераторы — разновидность итераторов. Их можно создать с помощью функций-генераторов или генераторных выражений. Функции-генераторы определяются как обычные функции, но используют ключевое слово `yield`, чтобы выдавать значения по одному, сохраняя состояние между выдачами. Генераторные выражения похожи на списковые включения (list comprehensions), но используют круглые скобки вместо квадратных и создают объект-генератор.

```python
numbers = [1, 2, 3, 4, 5]
squares = (x ** 2 for x in numbers)

for square in squares:
    print(square, end=" ")  # Вывод: 1 4 9 16 25
```

В этом фрагменте мы создаём список чисел. Генераторное выражение `(x ** 2 for x in numbers)` вычисляет квадрат каждого числа. Оно создаёт объект-генератор, который выдаёт квадраты при переборе. Цикл `for` последовательно получает и выводит каждый квадрат.

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

for num in countdown(5):
    print(num, end=" ")  # Вывод: 5 4 3 2 1
```

Здесь мы определяем функцию-генератор `countdown`, которая принимает число `n`. Внутри функции цикл `while` с инструкцией `yield` выдаёт числа в убывающем порядке от `n` до 1. В цикле `for` генератор выдаёт очередное значение последовательности, пока перебор не завершится.

Ещё один пример функции-генератора выдаёт последовательность чисел Фибоначчи.

```python
def fibonacci_generator():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

fibonacci = fibonacci_generator()

for _ in range(10):
    print(next(fibonacci))
```

В этом примере мы определяем функцию `fibonacci_generator` с ключевым словом `yield`. Функция присваивает переменным `a` и `b` первые числа Фибоначчи. Затем она входит в бесконечный цикл, выдаёт текущее число Фибоначчи и обновляет переменные так, чтобы они содержали следующие числа последовательности. В цикле `for` мы вызываем `next()` для генератора и получаем очередное значение.

Генераторы экономят память, поскольку вычисляют очередное значение по запросу. Это особенно полезно для больших наборов данных или бесконечных последовательностей.

### 2.3. Итоги

Итераторы и генераторы позволяют эффективно перебирать последовательности данных в Python. Итераторы дают последовательный доступ к элементам коллекции, а генераторы создают значения по запросу. Это особенно полезно при работе с большими или бесконечными наборами данных, которые нецелесообразно целиком загружать в память. Зная итераторы и генераторы, вы сможете писать более эффективные программы с меньшим расходом памяти.
