## Часть 4. Работа с отладчиком

Отладка — необходимый навык программиста. Она помогает находить и исправлять ошибки в коде, делать программы устойчивее и надёжнее.

### 4.1. Модуль `pdb`

Название модуля `pdb` происходит от Python Debugger. Это встроенный в Python отладчик для интерактивной отладки программ. Он позволяет приостанавливать выполнение программы, просматривать значения переменных, вычислять выражения и отслеживать ход выполнения программы, чтобы находить и исправлять ошибки. В этом разделе разберём, как пользоваться модулем `pdb`.

### 4.2. Команды отладчика

Модуль `pdb` предоставляет команды для перемещения по коду и его анализа во время отладки. Вот несколько часто используемых команд.

- `s` или `step` выполняет текущую строку и останавливается на следующей доступной точке, при необходимости заходя внутрь вызываемой функции.
- `n` или `next` выполняет текущую строку и продолжает до следующей строки в текущей функции.
- `l` или `list` показывает текущую строку кода и строки вокруг неё.
- `p` или `print` выводит значение переменной или выражения.
- `q` или `quit` завершает сеанс отладки и выполнение программы.

```python
import pdb

def calculate_sum(a, b):
    result = a + b
    pdb.set_trace()  # Устанавливаем точку останова
    return result

num1 = 5
num2 = 10
sum_result = calculate_sum(num1, num2)
print("Сумма:", sum_result)
```

В этом примере мы импортируем модуль `pdb` и размещаем вызов `pdb.set_trace()` там, где нужна точка останова. Когда выполнение кода доходит до неё, программа приостанавливается и появляется приглашение отладчика. В нём можно вводить команды.

### 4.3. Итоги

Встроенный модуль `pdb` позволяет интерактивно отлаживать программы на Python. Точки останова, пошаговое выполнение и просмотр значений переменных помогают находить и исправлять ошибки в коде. Команды отладчика позволяют изучить поведение программы и выяснить причины ошибок или неожиданных результатов. Этот инструмент особенно полезен при работе со сложным кодом и ошибками, которые трудно воспроизвести.
