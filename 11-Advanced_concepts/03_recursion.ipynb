## Часть 3. Рекурсия

Рекурсия — приём программирования, при котором функция вызывает себя, разбивая задачу на более мелкие подзадачи. Это альтернатива итеративным решениям с циклами, особенно полезная для задач с рекурсивной структурой. В этом разделе рассмотрим принципы рекурсии и примеры её применения.

### 3.1. Принципы рекурсии

Рекурсивное решение сводит задачу к более простому варианту той же задачи. Оно включает два основных компонента.

- **Базовый случай** определяет условие завершения рекурсии. Это простейший вариант задачи, который можно решить напрямую.
- **Рекурсивный случай** описывает часть задачи, которую можно разбить на более мелкие подзадачи. Чтобы решить их, функция вызывает себя с изменёнными входными данными.

### 3.2. Типичные задачи на рекурсию

Рекурсию часто применяют для решения самых разных задач программирования. Вот несколько распространённых примеров.

- **Факториал.** Факториал неотрицательного целого числа `n`, обозначаемый `n!`, равен произведению всех положительных целых чисел, не превосходящих `n`.
- **Двоичный поиск.** Алгоритм поиска элементов в отсортированном массиве по принципу «разделяй и властвуй».
- **Ханойская башня.** Головоломка, в которой нужно перемещать диски с одного стержня на другой, используя три стержня.
- **Поиск в глубину.** Алгоритм обхода графа, который проходит как можно дальше по каждой ветви перед возвратом.
- **Сортировка слиянием.** Алгоритм сортировки, который разбивает массив на более мелкие части, сортирует их и объединяет.
- **Перестановки и сочетания.** Задачи, в которых нужно получить все способы расположить или выбрать элементы.
- **Поиск с возвратом.** Общий алгоритмический приём поиска решений, при котором перебирают варианты и отменяют неудачные выборы.

#### 3.2.1. Факториал

Функция вычисления факториала рекурсивно вызывает себя, сводя задачу к более простым подзадачам. В базовом случае (`n == 0`) она возвращает 1. При возврате из рекурсивных вызовов значения перемножаются, образуя факториал `n`.

````python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
````

Пример использования:

````python
print(factorial(5))  # Вывод: 120
print(factorial(0))  # Вывод: 1
````

#### 3.2.2. Двоичный поиск

Двоичный поиск определяет позицию искомого значения в отсортированном массиве. Алгоритм делит область поиска пополам, пока не найдёт нужное значение или не установит, что его нет.

````python
def binary_search(arr, target, low, high):
    if low > high:
        return -1
    else:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            return binary_search(arr, target, mid + 1, high)
        else:
            return binary_search(arr, target, low, mid - 1)
````

Пример использования:

````python
arr = [1, 3, 5, 7, 9, 11, 13]
target = 7
result = binary_search(arr, target, 0, len(arr) - 1)
print(result)  # Вывод: 3
````

#### 3.2.3. Ханойская башня

Ханойская башня — математическая головоломка с тремя стержнями и несколькими дисками разного размера. Нужно перенести всю стопку дисков с одного стержня на другой. За один ход можно переместить только один диск, а больший диск нельзя класть на меньший.

````python
def tower_of_hanoi(n, source, destination, auxiliary):
    if n > 0:
        tower_of_hanoi(n - 1, source, auxiliary, destination)
        print(f"Переместите диск {n} со стержня {source} на стержень {destination}")
        tower_of_hanoi(n - 1, auxiliary, destination, source)
````

Пример использования:

````python
tower_of_hanoi(3, 'A', 'C', 'B')
````

Вывод:
```
Переместите диск 1 со стержня A на стержень C
Переместите диск 2 со стержня A на стержень B
Переместите диск 1 со стержня C на стержень B
Переместите диск 3 со стержня A на стержень C
Переместите диск 1 со стержня B на стержень A
Переместите диск 2 со стержня B на стержень C
Переместите диск 1 со стержня A на стержень C
```

#### 3.2.4. Поиск в глубину (DFS)

Поиск в глубину — алгоритм обхода графа. Он проходит как можно дальше по каждой ветви, прежде чем вернуться назад. Такой обход можно реализовать рекурсивно.

````python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()
    visited.add(start)
    print(start)
    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)
````

Пример использования:

````python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

dfs(graph, 'A')
````

Вывод:
```
A
B
D
E
C
F
```

#### 3.2.5. Сортировка слиянием

Сортировка слиянием следует принципу «разделяй и властвуй». Алгоритм рекурсивно делит исходный массив на две половины, сортирует их отдельно и затем сливает отсортированные половины.

````python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    result = []
    i, j = 0, 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result
````

Пример использования:

````python
arr = [9, 5, 1, 3, 7, 2, 8, 6, 4]
sorted_arr = merge_sort(arr)
print(sorted_arr)  # Вывод: [1, 2, 3, 4, 5, 6, 7, 8, 9]
````

#### 3.2.6. Перестановки и сочетания

Перестановки и сочетания — математические понятия, с помощью которых подсчитывают возможные способы упорядочить или выбрать элементы заданного множества. Их можно вычислять рекурсивно.

````python
import itertools

# Перестановки
items = ['A', 'B', 'C']
permutations = list(itertools.permutations(items))
print(permutations)  # Вывод: [('A', 'B', 'C'), ('A', 'C', 'B'), ('B', 'A', 'C'), ('B', 'C', 'A'), ('C', 'A', 'B'), ('C', 'B', 'A')]

# Сочетания
combinations = list(itertools.combinations(items, 2))
print(combinations)  # Вывод: [('A', 'B'), ('A', 'C'), ('B', 'C')]
````

#### 3.2.7. Поиск с возвратом

Поиск с возвратом (backtracking) — алгоритмический приём перебора возможных решений. Алгоритм постепенно строит варианты и отбрасывает каждый из них, как только становится ясно, что он не подходит. Этот приём часто используют для поиска всех возможных перестановок или сочетаний и для решения головоломок.

Реализация поиска с возвратом зависит от конкретной задачи.

### 3.3. Итоги

При рекурсии функция вызывает себя, чтобы решить задачу, разбив её на более мелкие подзадачи. В базовом случае простейший вариант задачи решается напрямую. В рекурсивном случае задача сводится к более мелким подзадачам. Рекурсия применима к широкому кругу задач и часто даёт компактное решение. При этом важно продумать условие завершения, иначе возможно переполнение стека вызовов. Понимание рекурсии помогает решать сложные задачи и разрабатывать эффективные алгоритмы в Python.
