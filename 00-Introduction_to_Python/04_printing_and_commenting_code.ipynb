## Часть 4. Вывод и комментирование кода

Вывод и комментарии помогают передавать информацию, форматировать результаты и пояснять код. В этом разделе мы рассмотрим способы вывода, форматирования и комментирования кода на Python.

### 4.1. Вывод данных

Функция `print()` выводит данные в консоль. С её помощью можно выводить строки, значения переменных и результаты выражений.

Рассмотрим пример.

```python
print("Привет, мир!")
```

### 4.2. Форматирование вывода

Форматирование позволяет настроить представление данных при выводе. В Python для этого есть несколько способов, в том числе конкатенация строк, подстановка значений в строки (интерполяция) и метод `format()`.

Рассмотрим несколько примеров.

```python
# Конкатенация строк

name = "Alice"
age = 25
print("Меня зовут " + name + ". Мой возраст — " + str(age) + ".")
```

```python
# Интерполяция строк (f-строки)

name = "Alice"
age = 25
print(f"Меня зовут {name}. Мой возраст — {age}.")
```

```python
# Метод format()

name = "Alice"
age = 25
print("Меня зовут {}. Мой возраст — {}.".format(name, age))
```

### 4.3. Форматирование чисел

Для чисел есть дополнительные параметры форматирования. Можно задавать точность, количество знаков после десятичной точки и выравнивание.

Рассмотрим несколько способов форматирования чисел.

#### 4.3.1. Форматирование чисел с плавающей точкой

Для форматирования чисел с плавающей точкой можно использовать метод `format()` или f-строки. Они позволяют задать количество знаков после десятичной точки и управлять отображением конечных нулей.

Рассмотрим пример.

```python
pi = 3.14159
formatted_pi = "{:.2f}".format(pi)  # Форматируем pi с двумя знаками после десятичной точки
print(formatted_pi)  # Результат: 3.14

formatted_pi = f"{pi:.2f}"  # Используем f-строку
print(formatted_pi)  # Результат: 3.14
```

#### 4.3.2. Форматирование целых чисел

Для форматирования целых чисел также можно использовать метод `format()` или f-строки.

Рассмотрим пример.

```python
num = 42
formatted_num = "{:4d}".format(num)  # Форматируем num в поле шириной не менее 4 символов
print(formatted_num)  # Результат:   42

formatted_num = f"{num:4d}"  # Используем f-строку
print(formatted_num)  # Результат:   42
```

#### 4.3.3. Форматирование с разделителями групп разрядов

Большие числа удобнее читать с разделителями тысяч. Метод `format()` позволяет использовать запятые как разделители групп разрядов.

Рассмотрим пример.

```python
population = 5000000
formatted_pop = "{:,}".format(population)  # Форматируем population с разделителями групп разрядов
print(formatted_pop)  # Результат: 5,000,000
```

### 4.4. Комментирование кода

Комментарии в Python поясняют и документируют код, облегчая его чтение. Интерпретатор не выполняет текст после `#` до конца строки. Для нескольких строк комментария ставьте `#` в начале каждой. Отдельно стоящие строки в тройных кавычках иногда используют как многострочные пояснения. С точки зрения синтаксиса это строковые литералы.

Рассмотрим примеры.

```python
# Это однострочный комментарий.
```

```python
"""
Это пояснение в строке с тройными кавычками.
Оно может занимать несколько строк.
"""
'''
Такое пояснение тоже можно записать
на нескольких строках.
'''
```

### 4.5. Итоги

Вывод и комментарии помогают общаться с пользователями и документировать код. Функция `print()` выводит данные, а приёмы форматирования позволяют настраивать их представление. Комментарии поясняют код, описывают его назначение и улучшают читаемость. Удачное оформление вывода и комментариев делает код понятнее и упрощает его сопровождение и совместную работу.
