## Часть 2. Запуск программ на Python

В этом разделе вы познакомитесь с разными способами запуска программ на Python. Мы рассмотрим способы и инструменты, которые позволяют запускать сценарии и взаимодействовать с программами. К концу раздела вы будете знать, как выполнять код Python в интерактивном режиме.

### 2.1. Запуск Python в интерактивном режиме

У Python есть интерактивный режим, который обычно называют оболочкой Python или REPL (Read-Eval-Print Loop, цикл «чтение — вычисление — вывод»). В этом режиме можно выполнять код построчно и сразу видеть результат. Он удобен для проверки фрагментов кода, экспериментов с возможностями языка и быстрых вычислений.

#### 2.1.1. Запуск оболочки Python

Чтобы запустить оболочку Python, откройте командную строку или терминал и введите команду `python`. Запустится интерпретатор Python и появится приглашение для ввода кода. В зависимости от системы для запуска интерпретатора Python 3 может понадобиться команда `python3`.

#### 2.1.2. Выполнение инструкций Python в интерактивном режиме

После запуска оболочки Python можно вводить инструкции и выражения, которые интерпретатор выполнит сразу. Например, попробуйте ввести простые математические выражения.

```python
>>> 2 + 3  # Результат: 5
```

```python
>>> 10 / 2  # Результат: 5.0
```

Также можно определять переменные и функции и выполнять более сложный код.

```python
>>> x = 5
>>> y = 2 * x + 3
>>> y  # Результат: 13
```

```python
>>> def square(n):
...     return n ** 2
...
>>> square(4)
```

#### 2.1.3. Выход из оболочки Python

Выйти из оболочки Python можно с помощью функции `exit()` или `quit()` либо сочетания клавиш, принятого в вашей операционной системе, например Ctrl+Z или Ctrl+D.

Интерактивный режим удобен, если нужно проверить код, испытать разные идеи и сразу увидеть результат. Он помогает изучать язык и знакомиться с его возможностями.

### 2.2. Запуск сценариев Python

Сценарии Python сохраняются в текстовых файлах с расширением `.py`. Их можно запускать разными способами. При запуске сценария выполняется последовательность инструкций или функций, сохранённых в файле.

#### 2.2.1. Создание сценария Python

Для создания сценария подойдёт любой текстовый редактор или интегрированная среда разработки (IDE). Создайте файл и сохраните его с расширением `.py`. Например, назовите его `my_script.py`.

#### 2.2.2. Запуск сценария Python из командной строки

Чтобы запустить созданный сценарий из командной строки, укажите имя файла сценария после команды запуска интерпретатора Python. Откройте командную строку или терминал, перейдите в каталог со сценарием и выполните команду.

```shell
python my_script.py
```

Интерпретатор Python прочитает файл сценария, последовательно выполнит инструкции, покажет результат или выполнит предусмотренные действия.

#### 2.2.3. Передача аргументов командной строки сценарию Python

При запуске сценария Python можно передавать ему аргументы командной строки. Это позволяет задавать входные данные и параметры во время запуска. Чтобы обратиться к аргументам внутри сценария, используйте список `sys.argv` из модуля `sys`.

Рассмотрим сценарий `script.py`, который принимает аргументы командной строки и выводит их.

```python
import sys

# Получаем аргументы командной строки
args = sys.argv

# Проверяем, были ли переданы аргументы
if len(args) > 1:
    # Перебираем аргументы с индекса 1 (индекс 0 занимает имя сценария)
    for i in range(1, len(args)):
        # Выводим каждый аргумент
        print("Аргумент", i, ":", args[i])
else:
    print("Аргументы не переданы.")
```

В общем случае аргументы указывают после имени сценария. Например, чтобы передать имя `John`, выполните вторую команду.

```shell
python script.py arg1 arg2 arg3 ...
```

```shell
python script.py John
```

Сценарии Python позволяют автоматизировать задачи, создавать приложения и повторно использовать код. В файл можно записать последовательность инструкций, а затем запустить её целиком. Так можно разрабатывать более сложные и объёмные программы.
