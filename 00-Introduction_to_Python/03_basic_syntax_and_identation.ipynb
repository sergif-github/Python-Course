## Часть 3. Основы синтаксиса и отступы

Python использует простой, читаемый синтаксис. Структуру блоков кода в нём задают отступы. Чтобы писать корректные программы, которые легко сопровождать, важно знать основные правила синтаксиса и правильно расставлять отступы.

### 3.1. Инструкции и строки

В Python инструкции — это команды, выполняющие определённые действия. Обычно инструкция занимает одну строку, но её можно разбить на несколько строк с помощью явного или неявного переноса.

Строка кода в Python заканчивается символом новой строки. Чтобы явно указать, что инструкция продолжается на следующей строке, поставьте обратную косую черту (`\`) в конце строки.

Рассмотрим пример.

````python
result = 10 + 20 + \
         30 + 40 + \
         50

print(result)  # Результат: 150
````

### 3.2. Отступы

В Python отступы определяют структуру и границы блоков кода. В некоторых языках блоки выделяют фигурными скобками или ключевыми словами. В Python начало и конец блоков обозначают за счёт одинаковых отступов для инструкций одного уровня. Для отступов используют пробелы или табуляцию. Важно придерживаться одного способа во всём коде. Обычно для каждого уровня отступа рекомендуют четыре пробела.

Рассмотрим пример.

```python
if condition:
    # Блок кода с отступом
    statement1
    statement2
else:
    # Другой блок кода с отступом
    statement3
    statement4
```

Правильные отступы улучшают читаемость кода и помогают интерпретатору Python корректно определить структуру программы.

### 3.3. Пустые строки

Интерпретатор Python пропускает пустые строки. Ими можно разделять части кода или логические блоки, чтобы улучшить читаемость. Пустые строки помогают визуально организовать код.

### 3.4. Блоки кода в управляющих конструкциях

Управляющие конструкции `if`, `for`, `while` и определения функций задают блоки кода. Блок определяется уровнем отступа. Инструкции одного уровня внутри блока имеют одинаковый отступ.

Рассмотрим пример.

```python
if condition:
    # Блок кода 1
    statement1
    statement2
else:
    # Блок кода 2
    statement3
    statement4
```

### 3.5. Итоги

Отступы определяют иерархию и вложенность блоков кода. Правильные отступы позволяют избежать синтаксических ошибок и точно выразить логику программы.

Соблюдайте основные правила синтаксиса и отступов Python, чтобы писать ясный и корректный код. Единообразие помогает другим разработчикам читать и сопровождать ваши программы.
