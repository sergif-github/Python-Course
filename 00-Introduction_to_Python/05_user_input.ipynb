## Часть 5. Ввод данных пользователем

### 5.1. Ввод данных с помощью `input()`

Функция `input()` позволяет программе запросить данные у пользователя. Полученное значение можно сохранить в переменной и использовать в дальнейшей обработке.

Используйте следующий синтаксис.

```python
variable_name = input(prompt)
```

Здесь `variable_name` — имя переменной, в которой будет сохранено введённое значение, а `prompt` — необязательная строка с приглашением к вводу.

Рассмотрим пример.

````python
name = input("Введите ваше имя: ")
print("Привет, " + name + "! Приятно познакомиться.")
````

В этом примере функция `input()` показывает приглашение «Введите ваше имя» и сохраняет введённое имя в переменной `name`. Затем функция `print()` выводит персональное приветствие с этим именем.

Функция `input()` всегда возвращает строку. Чтобы выполнять математические операции с введённым значением, преобразуйте строку в нужный тип данных с помощью `int()`, `float()` или другой подходящей функции.

Рассмотрим пример преобразования введённого значения в целое число.

````python
age = int(input("Введите ваш возраст: "))
years_remaining = 100 - age
print("До вашего столетия осталось примерно столько лет: " + str(years_remaining) + ".")
````

В этом примере функция `input()` получает возраст пользователя в виде строки. Функция `int()` преобразует эту строку в целое число. Результат сохраняется в переменной `age`, после чего с ним выполняются вычисления.
