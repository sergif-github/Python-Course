# Модуль 0. Введение в Python

## Часть 6. Упражнения

### Упражнение 1
Напишите программу на Python, которая выводит `Hello, Python!`.

### Упражнение 2
Напишите программу на Python, которая запрашивает имя пользователя и выводит приветствие.

### Упражнение 3
Напишите программу на Python, которая запрашивает у пользователя два числа и выводит их сумму.

### Упражнение 4
Напишите программу на Python, которая запрашивает у пользователя три слова и выводит их в обратном порядке, разделяя пробелами.
