# Модуль 4. Циклы

## Часть 6. Упражнения

### Упражнение 1
Напишите программу, которая запрашивает число, а затем выводит все числа от 1 до введённого числа включительно с помощью цикла `while`.

### Упражнение 2
Напишите программу, которая запрашивает строку и выводит каждый её символ с новой строки с помощью цикла `for`.

### Упражнение 3
Напишите программу, которая запрашивает последовательность чисел. Перебирайте числа и выводите «Найдено», если встретилось число 5. Когда пользователь вводит 0, завершите цикл.

### Упражнение 4
Напишите программу, которая выводит треугольник из звёздочек (`*`). Число строк треугольника определяет пользователь. Чтобы выводить текст без перехода на новую строку, используйте `print("string", end="")`.

### Упражнение 5
Напишите программу, которая хранит список словарей. Каждый словарь описывает человека и содержит такие сведения, как имя, возраст и страна. Переберите список и выведите сведения о каждом человеке.

### Упражнение 6
Напишите программу, которая хранит вложенный список чисел. Переберите его элементы и вычислите сумму всех чисел.

### Упражнение 7
Напишите программу, которая запрашивает строку и с помощью цикла подсчитывает число вхождений заданного символа.

### Упражнение 8
Напишите программу, которая запрашивает строку. Переберите и выведите её символы, пропуская латинскую букву `'e'` с помощью инструкции `continue`.

### Упражнение 9
Напишите программу, которая хранит словарь товаров и их цен. Переберите словарь и выведите товары, цена которых превышает заданный порог.

### Упражнение 10
Напишите программу, которая показывает, как перебирать данные разных типов.

- Переберите список и выведите каждый его элемент.
- Переберите строку и выведите каждый её символ.
- Переберите словарь с помощью методов `.keys()`, `.values()` и `.items()`. Выведите ключи, значения и пары «ключ — значение».
