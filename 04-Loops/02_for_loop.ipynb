## Часть 2. Цикл `for`

Цикл `for` в Python используют для перебора элементов последовательности или другого итерируемого объекта, например списка, кортежа, строки или объекта `range`. Для каждого элемента выполняется один и тот же блок кода.

Ниже приведён пример цикла `for`.

```python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits: 
    print(fruit)

print("Цикл завершён")
```

В этом примере цикл `for` перебирает элементы списка `fruits`. На каждой итерации переменной `fruit` присваивается значение очередного элемента, а вызов `print()` выводит название фрукта. После перебора всех элементов списка программа выводит сообщение «Цикл завершён».

Вместе с циклами `for` часто используют `range()`, чтобы получить последовательность чисел. В следующем примере показано, как это сделать.

```python
for i in range(5):
    print(i)

print("Цикл завершён")
```

В этом примере вызов `range(5)` создаёт последовательность чисел от 0 до 4. Цикл `for` перебирает эту последовательность, а вызов `print()` выводит значение `i`. Когда все числа выведены, программа выводит сообщение «Цикл завершён».

### Итоги

В этой части вы познакомились с циклом `for` в Python. Он позволяет перебирать последовательность или другой итерируемый объект и выполнять блок кода для каждого элемента. Текущее значение доступно через переменную цикла. Для перебора последовательности чисел часто используют `range()`.
