## Часть 3. Управление циклом (`break`, `continue`)

В Python ход выполнения цикла можно менять с помощью управляющих инструкций. Две распространённые инструкции — `break` и `continue`.

### 3.1. Инструкция `break`

Инструкция `break` досрочно завершает цикл, например при выполнении заданного условия. При выполнении `break` цикл сразу завершается и программа переходит к первой инструкции после него.

Пример

```python
numbers = [1, 2, 3, 4, 5]
 
for num in numbers:
    if num == 3:
        break
    print(num)

print("Цикл завершён")
```

В этом примере инструкция `break` завершает цикл, когда значение `num` становится равным 3. Как только условие выполняется, цикл прекращается и программа переходит к инструкции после цикла. До завершения цикла выводятся только числа 1 и 2.

### 3.2. Инструкция `continue`

Инструкция `continue` прерывает текущую итерацию и переводит цикл к следующей. Когда выполняется `continue`, остальной код текущей итерации пропускается.

Пример

```python
numbers = [1, 2, 3, 4, 5]

for num in numbers:
    if num == 3:
        continue
    print(num)

print("Цикл завершён")
```

В этом примере инструкция `continue` прерывает итерацию, на которой значение `num` равно 3. При выполнении условия вызов `print()` пропускается, и цикл переходит к следующей итерации. В результате выводятся все числа, кроме 3.

### 3.3. Итоги

В этой части вы познакомились с инструкциями управления циклом `break` и `continue`. Инструкция `break` досрочно завершает цикл при заданном условии, а `continue` пропускает оставшуюся часть текущей итерации и переходит к следующей. Используйте эти инструкции, когда ход цикла нужно изменить в зависимости от конкретного условия.
