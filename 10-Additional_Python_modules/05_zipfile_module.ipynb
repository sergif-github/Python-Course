## Часть 5. Архивы ZIP (модуль `zipfile`)

Модуль `zipfile` позволяет создавать архивы ZIP, извлекать их содержимое и изменять его. Этот модуль работает со сжатыми файлами и папками. Он полезен для сжатия данных, упаковки файлов и создания резервных копий. Рассмотрим его основные возможности.

### 5.1. Знакомство с модулем `zipfile`

Модуль `zipfile` входит в стандартную библиотеку Python и предоставляет интерфейс для работы с архивами ZIP. Он позволяет создавать архивы, извлекать из них файлы, добавлять файлы и выполнять другие операции с архивами и их содержимым.

```python
import zipfile

# Создать новый архив ZIP
with zipfile.ZipFile("archive.zip", "w") as zipf:
    zipf.write("file1.txt")

# Извлечь файлы из архива ZIP
with zipfile.ZipFile("archive.zip", "r") as zipf:
    zipf.extractall("extracted_files")
```

В этом фрагменте импортируем модуль `zipfile`. С помощью класса `ZipFile` создаём архив `archive.zip` в режиме `"w"` (запись). В блоке `with` методом `write()` добавляем файл `file1.txt` в архив. Степень сжатия определяется настройками архива. В этом примере по умолчанию файл добавляется без сжатия. Затем методом `extractall()` извлекаем файлы из архива в каталог `extracted_files`.

### 5.2. Добавление файлов в архив ZIP

Модуль `zipfile` позволяет добавлять файлы в существующий архив ZIP.

```python
import zipfile

# Открыть существующий архив ZIP
with zipfile.ZipFile("archive.zip", "a") as zipf:
    # Добавить файл в архив
    zipf.write("file1.txt")
```

В этом примере импортируем модуль `zipfile`. Открываем существующий архив `archive.zip` через класс `ZipFile` в режиме `"a"` (добавление). В блоке `with` методом `write()` добавляем в архив файл `file1.txt`. В этом примере такой файл уже есть в архиве, поэтому архив получит две записи с одним именем.

### 5.3. Извлечение отдельных файлов из архива ZIP

С помощью модуля `zipfile` можно извлечь из архива один файл или несколько выбранных файлов. Это удобно, когда требуется только часть содержимого архива.

```python
import zipfile

destination_folder = "./"

# Извлечь выбранный файл из архива ZIP
with zipfile.ZipFile("archive.zip", "r") as zipf:
    zipf.extract("file1.txt", destination_folder)
```

В этом фрагменте импортируем модуль `zipfile`. Открываем архив `archive.zip` через класс `ZipFile` в режиме `"r"` (чтение). В блоке `with` методом `extract()` извлекаем файл `file1.txt`. Папку назначения передаём через переменную `destination_folder`. В примере она указывает на текущий каталог, и файл будет записан в него.

### 5.4. Просмотр содержимого архива ZIP

Модуль `zipfile` позволяет получить список файлов в архиве без их извлечения. Это удобно для просмотра содержимого архива.

```python
import zipfile

# Получить список файлов в архиве ZIP
with zipfile.ZipFile("archive.zip", "r") as zipf:
    file_list = zipf.namelist()

# Вывести список файлов
print("Файлы в архиве:")
for file in file_list:
    print(file)
```

В этом примере импортируем модуль `zipfile` и открываем архив `archive.zip` через класс `ZipFile` в режиме `"r"` (чтение). В блоке `with` методом `namelist()` получаем список имён файлов в архиве и сохраняем его в переменной `file_list`. В конце выводим этот список.

### 5.5. Итоги

Модуль `zipfile` даёт удобный способ работы с архивами ZIP в Python. С его помощью можно создавать архивы, извлекать из них файлы, добавлять файлы и выполнять другие операции с архивами и их содержимым. Он пригодится для сжатия файлов, их упаковки для передачи и работы со сжатыми архивами.
