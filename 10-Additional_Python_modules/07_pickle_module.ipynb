## Часть 7. Сериализация и десериализация объектов

Сериализация преобразует объект в формат, пригодный для хранения или передачи. Десериализация — обратный процесс, при котором объект восстанавливается из сериализованного представления. В Python для этих задач можно использовать, например, модули `pickle`, `json` и `yaml`. Рассмотрим сериализацию и десериализацию на примере модуля `pickle`.

### 7.1. Сериализация объектов с помощью `pickle`

Модуль `pickle` сериализует объекты в двоичный формат. Он поддерживает многие объекты Python, в том числе экземпляры пользовательских классов, списки и словари. Сериализованные данные можно сохранить в файл или передать по сети.

```python
import pickle

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

person = Person("Alice", 25)

with open("person.pkl", "wb") as file:
    pickle.dump(person, file)
```

В этом примере объявляем класс `Person` с атрибутами `name` и `age`. Функция `pickle.dump()` сериализует объект `person` и записывает его в файл `person.pkl`, открытый в двоичном режиме.

### 7.2. Десериализация объектов с помощью `pickle`

Десериализация восстанавливает объект из сериализованного представления. Модуль `pickle` позволяет десериализовать объекты из данных, полученных из файла или сетевого потока.

```python
import pickle

with open("person.pkl", "rb") as file:
    person = pickle.load(file)
    print(person.name)
    print(person.age)
```

В этом примере открываем файл `person.pkl` в двоичном режиме и восстанавливаем хранящийся в нём объект с помощью функции `pickle.load()`. После десериализации объект `person` снова доступен для работы.

### 7.3. Другие форматы сериализации

Формат `pickle` удобен для объектов Python. Для обмена данными с программами на других языках или с другими системами могут понадобиться иные форматы. Модули `json` и `yaml` поддерживают сериализацию и десериализацию в форматах JSON и YAML соответственно.

Рассмотрим пример сериализации и десериализации с помощью `json`:

```python
import json

person = {"name": "Alice", "age": 25}

# Сериализация в JSON
json_data = json.dumps(person)
print(json_data)

# Десериализация из JSON
deserialized_person = json.loads(json_data)
print(deserialized_person["name"])
print(deserialized_person["age"])
```

В этом примере функция `json.dumps()` сериализует словарь `person` в строку формата JSON. Функция `json.loads()` десериализует эту строку обратно в объект Python.

### 7.4. Итоги

Сериализация преобразует объекты в формат, удобный для хранения или передачи, а десериализация восстанавливает объекты из этого формата. Для сериализации и десериализации в Python используются разные модули, например `pickle`, `json` и `yaml`. Модуль `pickle` удобен для объектов Python, а `json` и `yaml` подходят для работы с форматами JSON и YAML соответственно. Эти операции позволяют сохранять и восстанавливать состояние объектов и обмениваться данными между системами.
