## Часть 4. Работа с изображениями (Pillow)

Pillow (PIL) — популярная библиотека Python для обработки изображений. В ней есть функции и методы для открытия, преобразования и сохранения файлов изображений разных форматов. Pillow широко применяют для обработки изображений, компьютерного зрения и работы с графикой. Рассмотрим её основные возможности.

### 4.1. Знакомство с Pillow

Библиотека Pillow основана на PIL (Python Imaging Library). Она предоставляет удобный интерфейс для открытия изображений, изменения размеров, обрезки, применения фильтров и сохранения в разных форматах. Поддержка множества форматов позволяет работать с разными типами изображений.

```python
from PIL import Image

# Открыть файл изображения
image = Image.open("image.jpg")

# Вывести свойства изображения
print("Формат изображения:", image.format)
print("Размер изображения:", image.size)
print("Цветовой режим изображения:", image.mode)

# Показать изображение
image.show()
```

В этом фрагменте импортируем `Image` из пакета `PIL`. Методом `Image.open()` открываем файл `image.jpg` и сохраняем объект изображения в переменной `image`. Затем выводим его свойства: формат, размеры и цветовой режим. Наконец, отображаем изображение с помощью метода `show()`.

### 4.2. Преобразование изображений

В Pillow есть методы для изменения размера, обрезки, поворота и применения фильтров. Эти операции позволяют настроить внешний вид и характеристики изображения под конкретную задачу.

```python
from PIL import Image

# Открыть файл изображения
image = Image.open("image.jpg")

# Изменить размер изображения
resized_image = image.resize((400, 300))

# Вырезать область изображения
cropped_image = image.crop((100, 100, 300, 200))

# Повернуть изображение
rotated_image = image.rotate(90)

# Сохранить преобразованные изображения
resized_image.save("resized_image.jpg")
cropped_image.save("cropped_image.jpg")
rotated_image.save("rotated_image.jpg")
```

В этом примере импортируем `Image` из пакета `PIL`. Открываем файл `image.jpg` методом `Image.open()` и сохраняем объект в переменной `image`. Затем преобразуем изображение. Методом `resize()` меняем ширину на 400 пикселей, а высоту на 300 пикселей и сохраняем результат в `resized_image`. Методом `crop()` вырезаем область с координатами `(100, 100, 300, 200)` и сохраняем результат в `cropped_image`. Методом `rotate()` поворачиваем изображение на 90 градусов против часовой стрелки и сохраняем результат в `rotated_image`. Наконец, методом `save()` сохраняем преобразованные изображения в отдельных файлах.

### 4.3. Фильтрация изображений

Pillow предоставляет фильтры, которые позволяют улучшить вид изображения или выделить на нём отдельные признаки. Среди доступных операций — размытие, повышение резкости, выделение границ и другие.

```python
from PIL import Image, ImageFilter

# Открыть файл изображения
image = Image.open("image.jpg")

# Применить фильтры к изображению
blurred_image = image.filter(ImageFilter.BLUR)
sharpened_image = image.filter(ImageFilter.SHARPEN)
edge_detected_image = image.filter(ImageFilter.FIND_EDGES)

# Сохранить преобразованные изображения
blurred_image.save("blurred_image.jpg")
sharpened_image.save("sharpened_image.jpg")
edge_detected_image.save("edge_detected_image.jpg")
```

В этом примере импортируем `Image` и модуль `ImageFilter` из пакета `PIL`. Открываем файл `image.jpg` методом `Image.open()` и сохраняем объект в `image`. Затем методом `filter()` применяем различные фильтры: размытие `ImageFilter.BLUR`, повышение резкости `ImageFilter.SHARPEN` и выделение границ `ImageFilter.FIND_EDGES`. Каждое обработанное изображение сохраняем в отдельный файл методом `save()`.

### 4.4. Преобразование форматов изображений

Pillow позволяет преобразовывать изображения из одного формата в другой. Это полезно, когда нужно преобразовать существующий файл или сохранить изображение в конкретном формате.

```python
from PIL import Image

# Открыть файл изображения
image = Image.open("image.jpg")

# Преобразовать формат изображения
image.save("image.png")
image.save("image.bmp")
```

В этом фрагменте импортируем `Image` из пакета `PIL`. Методом `Image.open()` открываем файл `image.jpg` и сохраняем объект в переменной `image`. Затем методом `save()` сохраняем изображение в разных форматах. В этом примере создаём файлы PNG и BMP.

### 4.5. Итоги

Pillow (PIL) — библиотека для обработки изображений на Python. Она позволяет открывать, преобразовывать и сохранять изображения в разных форматах. С ней можно изменять изображения, применять фильтры, преобразовывать форматы и выполнять другие операции. Эти инструменты пригодятся в приложениях для работы с изображениями, в компьютерном зрении и при анализе изображений.
