## Часть 2. Математические операции (модуль `math`)

Модуль `math` содержит набор функций и констант для различных математических операций. Он дополняет базовые арифметические операции, которые доступны через встроенные операторы Python. Рассмотрим основные возможности модуля `math`.

### 2.1. Знакомство с модулем `math`

Модуль `math` входит в стандартную библиотеку Python и предоставляет доступ к математическим функциям и константам. Он особенно полезен для вычислений, выходящих за рамки базовой арифметики. С ним можно вычислять тригонометрические функции, возводить числа в степень, находить логарифмы и выполнять другие операции.

Рассмотрим пример:

```python
import math

# Вычислить квадратный корень из числа
x = 16
sqrt_x = math.sqrt(x)
print(sqrt_x)

# Получить значение числа π
pi = math.pi
print(pi)
```

В этом фрагменте импортируем модуль `math`. Затем с помощью функции `sqrt()` из этого модуля вычисляем квадратный корень из числа `x = 16`. Сохраняем результат в переменной `sqrt_x` и выводим его. Кроме того, получаем из модуля `math` константу `pi` и выводим её значение.

### 2.2. Тригонометрические функции

Модуль `math` предоставляет разные тригонометрические функции, например синус, косинус и тангенс. Они нужны для вычислений, связанных с углами, и применяются в геометрии, физике и инженерных задачах.

```python
import math

# Вычислить синус угла
angle = math.pi / 4
sin_angle = math.sin(angle)
print(sin_angle)

# Вычислить косинус угла
cos_angle = math.cos(angle)
print(cos_angle)

# Вычислить тангенс угла
tan_angle = math.tan(angle)
print(tan_angle)
```

Здесь мы импортируем модуль `math` и задаём угол π/4. Функции `sin()`, `cos()` и `tan()` из модуля `math` вычисляют соответственно синус, косинус и тангенс этого угла. Сохраняем результаты в переменных `sin_angle`, `cos_angle` и `tan_angle` и выводим их.

### 2.3. Возведение в степень и логарифмы

В модуле `math` есть функции для возведения в степень и вычисления логарифмов. Они упрощают вычисления, в которых встречаются степени и логарифмы.

```python
import math

# Возвести число в степень
base = 2
exponent = 3
power = math.pow(base, exponent)
print(power)

# Вычислить натуральный логарифм числа
x = 10
logarithm = math.log(x)
print(logarithm)

# Вычислить логарифм числа по заданному основанию
base = 5
logarithm_base = math.log(x, base)
print(logarithm_base)
```

В этом примере импортируем модуль `math` и возводим число в степень функцией `pow()` с основанием `base = 2` и показателем `exponent = 3`. Сохраняем результат в переменной `power` и выводим его. Затем вычисляем натуральный логарифм числа `x = 10` с помощью `log()`. Сохраняем результат в `logarithm` и выводим его. Наконец, той же функцией `log()` вычисляем логарифм `x` по заданному основанию `base = 5`. Сохраняем результат в `logarithm_base` и выводим его.

### 2.4. Округление и вычисление модуля числа

При работе с числовыми данными часто требуется округлить число или получить его абсолютное значение. Эти операции полезны для обработки и форматирования чисел.

```python
import math

# Округлить число
x = 3.7
#rounded = math.round(x)
rounded = round(x)
print(rounded)

# Получить абсолютное значение числа
y = -5
#absolute_value = math.abs(y)
absolute_value = abs(y)

print(absolute_value)
```

Здесь мы импортируем модуль `math`. Округляем число `x = 3.7` с помощью функции `round()`, сохраняем результат в `rounded` и выводим его. Затем вычисляем абсолютное значение числа `y = -5` с помощью `abs()`, сохраняем результат в `absolute_value` и выводим его.

Функции `round()` и `abs()` встроены в Python и доступны без импорта модуля `math`.

### 2.5. Дополнительные математические функции

Модуль `math` содержит и другие математические функции, например округление вниз и вверх, вычисление факториала и многое другое. Эти функции полезны для специализированных вычислений.

```python
import math

# Вычислить факториал числа
n = 5
factorial = math.factorial(n)
print(factorial)

# Округлить число вниз и вверх
x = 3.7
floor = math.floor(x)
ceiling = math.ceil(x)
print(floor, ceiling)
```

В этом примере мы импортируем модуль `math` и вычисляем факториал числа `n = 5` с помощью функции `factorial()`. Сохраняем результат в переменной `factorial` и выводим его. Затем округляем число `x = 3.7` вниз функцией `floor()` и вверх функцией `ceil()`. Сохраняем результаты в переменных `floor` и `ceiling` соответственно и выводим оба значения.

С его помощью можно эффективно выполнять сложные математические вычисления в программах на Python.
