## Часть 6. Регулярные выражения (модуль `re`)

Модуль `re` поддерживает регулярные выражения — инструмент для поиска по шаблону и обработки текста. Регулярные выражения позволяют искать, извлекать и преобразовывать текст по заданным правилам. Они полезны при проверке данных, разборе текста и обработке строк. Рассмотрим основные возможности модуля `re`.

### 6.1. Знакомство с регулярными выражениями

Регулярное выражение — последовательность символов, задающая шаблон поиска. Такой шаблон позволяет гибко и кратко описать правила сопоставления и обработки строк. Он может содержать обычные символы и специальные символы, называемые метасимволами.

```python
import re

# Найти совпадение с шаблоном в строке
pattern = r"apple"
text = "I have an apple"
match = re.search(pattern, text)
if match:
    print("Найдено совпадение!")
else:
    print("Совпадений не найдено.")
```

В этом фрагменте импортируем модуль `re` и задаём шаблон `"apple"`. Префикс `r` создаёт необработанную строку (raw string). Задаём текст `"I have an apple"` и с помощью функции `search()` из модуля `re` ищем в нём совпадение с шаблоном. При совпадении программа выводит сообщение «Найдено совпадение!», иначе — «Совпадений не найдено.».

### 6.2. Поиск по простому шаблону

Регулярные выражения используют метасимволы и специальные последовательности, чтобы описать, какой текст считать совпадением. Такие метасимволы, как `.` (точка), `*` (звёздочка), `+` (плюс) и `[ ]` (квадратные скобки), задают правила шаблона.

```python
import re

# Найти совпадения с шаблоном адреса электронной почты
pattern = r"\w+@\w+\.\w+"
text = "Contact us at info@example.com or support@example.com"
matches = re.findall(pattern, text)
print(matches)
```

В этом примере импортируем модуль `re` и задаём шаблон в виде необработанной строки. Шаблон `\w+@\w+\.\w+` находит строки с простой структурой адреса электронной почты. Разберём его:

- `\w+` соответствует одному или нескольким буквенно-цифровым символам или знакам подчёркивания.
- `@` соответствует символу `@`.
- `\.` соответствует точке. Обратная косая черта экранирует точку, поскольку в регулярных выражениях она является метасимволом.

Задаём строку, содержащую два адреса электронной почты. Функцией `findall()` ищем в тексте все совпадения с шаблоном. Сохраняем найденные адреса в переменной `matches` и выводим её значение.

### 6.3. Замена по шаблону

Модуль `re` позволяет заменять фрагменты, совпавшие с шаблоном, на заданный текст. Это полезно, когда нужно обработать фрагменты текста, соответствующие определённым условиям.

```python
import re

# Заменить фрагменты строки по шаблону
pattern = r"apple"
text = "I have an apple"
new_text = re.sub(pattern, "orange", text)
print(new_text)
```

В этом фрагменте импортируем модуль `re`. Задаём шаблон `"apple"` как необработанную строку, а в качестве текста выбираем `"I have an apple"`. Функцией `sub()` заменяем совпадения с шаблоном словом `"orange"`. Сохраняем полученную строку в переменной `new_text` и выводим её.

### 6.4. Разбиение строк по шаблону

Модуль `re` позволяет разбивать строку по шаблону, задающему разные варианты разделителя. Это полезно, если нужно получить подстроки, а разделители в исходном тексте различаются.

```python
import re

# Разбить строку по шаблону
pattern = r"\s+"
text = "Hello    World"
substrings = re.split(pattern, text)
print(substrings)
```

В этом примере импортируем модуль `re`. Задаём в виде необработанной строки шаблон `\s+`, которому соответствует один или несколько пробельных символов. Берём строку `"Hello    World"`, содержащую несколько пробелов. Функция `split()` разбивает строку на подстроки по этому шаблону. Сохраняем результат в переменной `substrings` и выводим его.

### 6.5. Итоги

Модуль `re` позволяет искать совпадения и обрабатывать текст с помощью регулярных выражений. Регулярные выражения задают шаблоны поиска, по которым можно сопоставлять текст, выполнять замены и разбивать строки. Эти возможности помогают проверять данные, разбирать текст и преобразовывать строки в программах на Python. Регулярные выражения применяются в разных областях, поэтому модуль `re` пригодится для многих задач с текстовыми данными.
