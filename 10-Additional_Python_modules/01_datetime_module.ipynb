## Часть 1. Работа с датами и временем (модуль `datetime`)

Работа с датами и временем — распространённая задача во многих приложениях на Python. Модуль `datetime` предоставляет классы и функции для работы с датами, временем и временными метками.

### 1.1. Модуль `datetime`

Модуль `datetime` позволяет создавать даты и время, выполнять с ними операции, форматировать их и вычислять разницу между ними.

```python
import datetime

now = datetime.datetime.now()
print(now)
```

В этом примере показано, как получить текущую дату и время. Импортируем модуль `datetime` и вызываем метод `now()` класса `datetime`. Сохраняем результат в переменной `now` и выводим его на экран.

### 1.2. Работа с датами

Дата обозначает конкретный календарный день и состоит из года, месяца и числа. Объекты даты создаются конструктором `date()`. С датами можно выполнять арифметические операции, например прибавлять или вычитать несколько дней. Метод `strftime()` позволяет представить дату в виде строки заданного формата.

```python
import datetime

today = datetime.date.today()
print(today)

# Арифметика дат
tomorrow = today + datetime.timedelta(days=1)
print(tomorrow)

# Форматирование дат
formatted_date = today.strftime("%Y-%m-%d")
print(formatted_date)
```

Здесь мы получаем текущую дату с помощью `date.today()` и сохраняем её в переменной `today`. При выводе по умолчанию дата отображается в формате `YYYY-MM-DD`. Затем с помощью `timedelta()` создаём интервал в один день и прибавляем его к текущей дате. Получаем завтрашнюю дату. Наконец, методом `strftime()` преобразуем текущую дату в строку формата `%Y-%m-%d`, например `2023-06-14`.

### 1.3. Работа со временем

Время суток обозначает конкретный момент в течение дня и состоит из часов, минут, секунд и микросекунд. Объекты времени создаются конструктором `time()`. Чтобы прибавить или вычесть интервал, в примере ниже мы сначала объединяем время с датой. Метод `strftime()` позволяет представить время в виде строки заданного формата.

```python
import datetime

current_time = datetime.datetime.now().time()
print(current_time)

# Арифметика времени
future_time = datetime.datetime.combine(datetime.date.today(), current_time) + datetime.timedelta(hours=2)
print(future_time.time())

# Форматирование времени
formatted_time = current_time.strftime("%H:%M:%S")
print(formatted_time)
```

В этом примере вызов `datetime.now().time()` возвращает текущее время, которое мы сохраняем в переменной `current_time`. При выводе оно отображается в 24-часовом формате `HH:MM:SS`, при наличии микросекунд — с дробной частью. Затем методом `combine()` объединяем текущую дату и время в объект `datetime`. Прибавляем к нему интервал `timedelta` в два часа и выводим полученное время. Наконец, с помощью метода `strftime()` форматируем текущее время как строку `%H:%M:%S`.

### 1.4. Объединение даты и времени

Объекты `datetime` объединяют дату и время. Их можно создавать конструктором `datetime()`. Как и объекты даты и времени по отдельности, объекты `datetime` можно преобразовывать в строки с помощью метода `strftime()`. С ними также можно выполнять арифметические операции, например вычислять разницу между двумя объектами `datetime`.

```python
import datetime

current_datetime = datetime.datetime.now()
print(current_datetime)

# Арифметика даты и времени
future_datetime = current_datetime + datetime.timedelta(days=1, hours=3)
print(future_datetime)

# Форматирование даты и времени
formatted_datetime = current_datetime.strftime("%Y-%m-%d %H:%M:%S")
print(formatted_datetime)
```

Здесь мы получаем текущую дату и время с помощью `datetime.now()`, сохраняем результат в `current_datetime` и выводим его. Он содержит и дату, и время. Затем с помощью `timedelta()` создаём интервал в один день и три часа, прибавляем его к текущей дате и времени и выводим полученный момент в будущем. Наконец, применяем метод `strftime()`, чтобы перевести текущую дату и время в строку формата `%Y-%m-%d %H:%M:%S`, например `2023-06-14 10:30:00`.

### 1.5. Работа с часовыми поясами

Часовые пояса — регионы с одинаковым стандартным временем. В модуле `datetime` с часовыми поясами позволяет работать класс `timezone`. Метод `astimezone()` переводит объекты `datetime` из одного часового пояса в другой. При таком переводе важно корректно учитывать переход на летнее время.

```python
import datetime
import pytz

utc_now = datetime.datetime.now(pytz.UTC)
print(utc_now)

# Перевод в другой часовой пояс
local_timezone = pytz.timezone("America/New_York")
local_time = utc_now.astimezone(local_timezone)
print(local_time)

# Учёт перехода на летнее время
dst_timezone = pytz.timezone("Europe/Paris")
dst_time = utc_now.astimezone(dst_timezone)
print(dst_time)
```

В этом примере мы импортируем модули `datetime` и `pytz`. Модуль `pytz` содержит сведения о часовых поясах. Вызовом `datetime.now(pytz.UTC)` получаем текущую дату и время по всемирному координированному времени (UTC), сохраняем значение в `utc_now` и выводим его. Затем с помощью `pytz.timezone()` создаём объект часового пояса `America/New_York` и переводим `utc_now` в местное время Нью-Йорка. Сохраняем результат в `local_time` и выводим его. Наконец, создаём объект часового пояса `Europe/Paris` и переводим `utc_now` в местное время Парижа. Сохраняем результат в `dst_time` и выводим его.

### 1.6. Работа с интервалами времени

Интервалы времени показывают разницу между двумя моментами. Их создают конструктором `timedelta()`. Интервалы можно прибавлять к датам и объектам `datetime` или вычитать из них. Для вычисления со временем суток его сначала объединяют с датой.

```python
import datetime

delta = datetime.timedelta(days=7, hours=3)
print(delta)

# Прибавление и вычитание интервалов времени
future_datetime = datetime.datetime.now() + delta
print(future_datetime)

past_datetime = datetime.datetime.now() - delta
print(past_datetime)
```

Здесь мы создаём объект `timedelta` с помощью `timedelta()` и сохраняем его в переменной `delta`. Он представляет интервал в семь дней и три часа. Выводим значение `delta`, а затем используем его в арифметических операциях. Прибавляем `delta` к текущей дате и времени с помощью `datetime.now() + delta` и выводим момент в будущем. Аналогично вычитаем `delta` с помощью `datetime.now() - delta` и выводим момент в прошлом.
