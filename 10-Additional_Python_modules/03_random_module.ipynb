## Часть 3. Генерация случайных чисел (модуль `random`)

Модуль `random` позволяет генерировать случайные числа, выбирать случайные элементы и перемешивать последовательности. Его применяют в моделировании, играх и статистическом анализе. Рассмотрим основные возможности модуля `random`.

### 3.1. Знакомство с модулем `random`

Модуль `random` входит в стандартную библиотеку Python и содержит функции для генерации псевдослучайных чисел и других случайных операций. Он подходит для моделирования, игр и учебных задач. Для паролей, токенов и других задач безопасности используйте модуль `secrets`.

Рассмотрим пример:

```python
import random

# Сгенерировать случайное целое число в заданном диапазоне
random_number = random.randint(1, 10)
print(random_number)

# Сгенерировать случайное число с плавающей точкой
random_float = random.random()
print(random_float)

# Выбрать случайный элемент последовательности
fruits = ["apple", "banana", "orange", "mango"]
random_fruit = random.choice(fruits)
print(random_fruit)
```

В этом фрагменте импортируем модуль `random`. Функция `randint()` генерирует случайное целое число от 1 до 10 включительно. Сохраняем его в переменной `random_number` и выводим. Затем функцией `random()` генерируем случайное число с плавающей точкой от 0 включительно до 1 исключительно, сохраняем его в `random_float` и выводим. Наконец, из списка фруктов выбираем случайный фрукт с помощью `choice()`, сохраняем его в `random_fruit` и выводим.

### 3.2. Случайное перемешивание последовательности

Модуль `random` содержит функцию для случайного перемешивания элементов последовательности. Она полезна, когда нужно изменить порядок элементов списка или другой изменяемой последовательности. Для перемешивания элементов кортежа или иного итерируемого объекта их можно сначала собрать в список.

```python
import random

cards = ["Ace", "King", "Queen", "Jack", "10", "9", "8", "7"]

# Перемешать карты
random.shuffle(cards)
print(cards)
```

Здесь мы импортируем модуль `random` и задаём список игральных карт. Функцией `shuffle()` модуля `random` случайным образом перемешиваем элементы списка `cards`. Порядок карт меняется, после чего мы выводим перемешанный список.

### 3.3. Случайный выбор нескольких элементов последовательности

В модуле `random` есть функция для случайного выбора нескольких элементов последовательности, например списка. Один и тот же элемент может попасть в результат несколько раз.

```python
import random

colors = ["red", "blue", "green", "yellow", "orange"]

# Случайно выбрать три цвета
random_choices = random.choices(colors, k=3)
print(random_choices)
```

В этом примере импортируем модуль `random`. Из списка цветов `colors` случайным образом выбираем три элемента с помощью функции `choices()` модуля `random`. Сохраняем результат в списке `random_choices` и выводим его.

### 3.4. Генерация случайных чисел в заданном диапазоне

В модуле `random` есть функции для генерации случайных чисел в заданном диапазоне. Они полезны, когда случайные значения должны попадать в определённый интервал.

```python
import random

# Сгенерировать случайное целое число в заданном диапазоне
random_number = random.randint(1, 100)
print(random_number)

# Сгенерировать случайное число с плавающей точкой в заданном диапазоне
random_float = random.uniform(0.0, 1.0)
print(random_float)
```

Здесь мы импортируем модуль `random`. Функцией `randint()` генерируем случайное целое число от 1 до 100 включительно, сохраняем его в переменной `random_number` и выводим. Затем функцией `uniform()` генерируем случайное число с плавающей точкой от 0.0 до 1.0, сохраняем его в переменной `random_float` и выводим.

### 3.5. Установка начального значения генератора случайных чисел

Модуль `random` позволяет задать начальное значение генератора (seed), чтобы получить ту же последовательность случайных чисел. Это полезно, когда требуется воспроизвести определённый набор случайных значений.

```python
import random

# Задать начальное значение генератора
random.seed(42)

# Сгенерировать случайные числа
random1 = random.random()
random2 = random.random()

print(random1)
print(random2)
```

В этом примере импортируем модуль `random`. Функцией `seed()` задаём начальное значение 42. Благодаря этому при каждом запуске этого кода генерируется одна и та же последовательность случайных чисел. Затем функцией `random()` генерируем два случайных числа с плавающей точкой, сохраняем их в переменных `random1` и `random2` и выводим. При каждом запуске этого кода значения будут одинаковыми.

Эти возможности пригодятся в моделировании, играх и статистическом анализе. Для криптографических и других задач безопасности применяйте модуль `secrets`.
