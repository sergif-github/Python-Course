## Часть 2. Операторы сравнения

Операторы сравнения сопоставляют значения и определяют, как они соотносятся между собой.

### 2.1. Операторы сравнения

В Python можно проверить, равны ли значения, различаются ли они и какое из них больше или меньше.

#### Оператор «равно» (==)

Оператор `==` проверяет, равны ли два значения. Он возвращает `True`, если они равны, и `False` в противном случае.

Пример

```python
x = 5
y = 3

result = x == y  # 5 == 3 -> False
print("Результат:", result)  # Вывод: Результат: False
```

#### Оператор «не равно» (!=)

Оператор `!=` проверяет, различаются ли два значения. Он возвращает `True`, если они различаются, и `False`, если они равны.

Пример

```python
x = 5
y = 3

result = x != y  # 5 != 3 -> True
print("Результат:", result)  # Вывод: Результат: True
```

#### Оператор «больше» (>)

Оператор `>` проверяет, больше ли левый операнд правого. Если условие выполняется, результат равен `True`, иначе результат равен `False`.

Пример

```python
x = 5
y = 3

result = x > y  # 5 > 3 -> True
print("Результат:", result)  # Вывод: Результат: True
```

#### Оператор «меньше» (<)

Оператор `<` проверяет, меньше ли левый операнд правого. Если условие выполняется, результат равен `True`, иначе результат равен `False`.

Пример

```python
x = 5
y = 3

result = x < y  # 5 < 3 -> False
print("Результат:", result)  # Вывод: Результат: False
```

#### Оператор «больше или равно» (>=)

Оператор `>=` проверяет, больше ли левый операнд правого или равен ему. Если условие выполняется, результат равен `True`, иначе результат равен `False`.

Пример

```python
x = 5
y = 3

result = x >= y  # 5 >= 3 -> True
print("Результат:", result)  # Вывод: Результат: True
```

#### Оператор «меньше или равно» (<=)

Оператор `<=` проверяет, меньше ли левый операнд правого или равен ему. Если условие выполняется, результат равен `True`, иначе результат равен `False`.

Пример

```python
x = 5
y = 3

result = x <= y  # 5 <= 3 -> False
print("Результат:", result)  # Вывод: Результат: False
```
