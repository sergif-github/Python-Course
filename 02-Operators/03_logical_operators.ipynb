## Часть 3. Логические операторы

Логические операторы объединяют условия и выполняют логические операции.

### 3.1. Логические операторы

В Python три логических оператора: `and`, `or` и `not`. Они позволяют принимать решения на основе нескольких условий.

#### Логическое «И» (and)

Если оба операнда логические, оператор `and` возвращает `True`, когда оба равны `True`, и `False` в остальных случаях.

Пример

```python
x = 5
y = 3
z = 7

result = (x > y) and (y < z)  # (5 > 3) and (3 < 7) -> True and True -> True
print("Результат:", result)  # Вывод: Результат: True
```

#### Логическое «ИЛИ» (or)

Если оба операнда логические, оператор `or` возвращает `True`, когда хотя бы один из них равен `True`. Если оба равны `False`, оператор возвращает `False`.

Пример

```python
x = 5
y = 3
z = 7

result = (x > y) or (y > z)  # (5 > 3) or (3 > 7) -> True or False -> True
print("Результат:", result)  # Вывод: Результат: True
```

#### Логическое «НЕ» (not)

Оператор `not` возвращает логическое отрицание своего операнда. Для значения `True` результат равен `False`, для `False` — `True`.

Пример

```python
x = True

result = not x  # not True -> False
print("Результат:", result)  # Вывод: Результат: False
```

В этой части вы познакомились с логическими операторами Python «И» (`and`), «ИЛИ» (`or`) и «НЕ» (`not`). Эти операторы помогают объединять условия и принимать решения на основе логических выражений.
