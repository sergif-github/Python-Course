## Часть 4. Операторы присваивания

Операторы присваивания позволяют присваивать значения переменным.

### 4.1. Операторы присваивания

Знак присваивания `=` можно объединять с другими операторами, чтобы выполнить операцию и присвоить её результат переменной за один шаг.

#### Простое присваивание (=)

Простое присваивание `=` связывает переменную со значением.

Пример

```python
x = 5
y = 3
```

### 4.2. Операторы составного присваивания

В Python есть операторы составного присваивания, которые объединяют арифметическую или побитовую операцию с присваиванием. Они выполняют операцию и обновляют значение переменной за один шаг.

#### Сложение с присваиванием (+=)

Оператор `+=` прибавляет правый операнд к левому и записывает результат в переменную слева.

Пример

```python
x = 5
x += 3  # x = x + 3 -> 5 + 3 -> 8
print("Результат:", x)  # Вывод: Результат: 8
```

#### Вычитание с присваиванием (-=)

Оператор `-=` вычитает правый операнд из левого и записывает результат в переменную слева.

Пример

```python
x = 5
x -= 3  # x = x - 3 -> 5 - 3 -> 2
print("Результат:", x)  # Вывод: Результат: 2
```

#### Умножение с присваиванием (*=)

Оператор `*=` умножает левый операнд на правый и записывает результат в переменную слева.

Пример

```python
x = 5
x *= 3  # x = x * 3 -> 5 * 3 -> 15
print("Результат:", x)  # Вывод: Результат: 15
```

#### Деление с присваиванием (/=)

Оператор `/=` делит левый операнд на правый и записывает результат в переменную слева.

Пример

```python
x = 10
x /= 2  # x = x / 2 -> 10 / 2 -> 5.0
print("Результат:", x)  # Вывод: Результат: 5.0
```

#### Остаток от деления с присваиванием (%=)

Оператор `%=` вычисляет остаток от деления левого операнда на правый и записывает результат в переменную слева.

Пример

```python
x = 10
x %= 3  # x = x % 3 -> 10 % 3 -> 1
print("Результат:", x)  # Вывод: Результат: 1
```

#### Возведение в степень с присваиванием (**=)

Оператор `**=` возводит левый операнд в степень, заданную правым, и записывает результат в переменную слева.

Пример

```python
x = 2
x **= 3  # x = x ** 3 -> 2 ** 3 -> 8
print("Результат:", x)  # Вывод: Результат: 8
```

#### Деление с округлением вниз и присваиванием (//=)

Оператор `//=` делит левый операнд на правый, округляет частное вниз и записывает результат в переменную слева.

Пример

```python
x = 10
x //= 3  # x = x // 3 -> 10 // 3 -> 3
print("Результат:", x)  # Вывод: Результат: 3
```

### 4.3. Итоги

В этой части вы изучили простое присваивание (`=`) и операторы составного присваивания (`+=`, `-=`, `*=`, `/=`, `%=`, `**=`, `//=`). Составное присваивание позволяет выполнить операцию и присвоить её результат переменной за один шаг.
