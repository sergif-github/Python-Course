## Часть 5. Побитовые операторы

Побитовые операторы выполняют операции над двоичным представлением целых чисел.

### 5.1. Побитовые операторы

Эти операторы работают с отдельными битами операндов. В Python для работы на уровне битов предусмотрено несколько операций.

#### Побитовое «И» (&)

Оператор `&` выполняет побитовое «И» над соответствующими битами двух операндов. Бит результата равен 1, если соответствующие биты обоих операндов равны 1.

Пример

```python
x = 5  # Двоичное представление: 0101
y = 3  # Двоичное представление: 0011

result = x & y  # Двоичное представление: 0001 -> Десятичное представление: 1
print("Результат:", result)  # Вывод: Результат: 1
```

#### Побитовое «ИЛИ» (|)

Оператор `|` выполняет побитовое «ИЛИ» над соответствующими битами двух операндов. Бит результата равен 1, если хотя бы один из двух соответствующих битов равен 1.

Пример

```python
x = 5  # Двоичное представление: 0101
y = 3  # Двоичное представление: 0011

result = x | y  # Двоичное представление: 0111 -> Десятичное представление: 7
print("Результат:", result)  # Вывод: Результат: 7
```

#### Побитовое исключающее «ИЛИ» (^)

Оператор `^` выполняет побитовое исключающее «ИЛИ» над соответствующими битами двух операндов. Бит результата равен 1, если ровно один из двух соответствующих битов равен 1.

Пример

```python
x = 5  # Двоичное представление: 0101
y = 3  # Двоичное представление: 0011

result = x ^ y  # Двоичное представление: 0110 -> Десятичное представление: 6
print("Результат:", result)  # Вывод: Результат: 6
```

#### Побитовое «НЕ» (~)

Оператор `~` инвертирует биты операнда: нули превращаются в единицы, единицы — в нули. Для целого числа `x` результат `~x` равен `-x - 1`.

Пример

```python
x = 5  # Двоичное представление: 0101

result = ~x  # Двоичное представление: 1010 (4-разрядный дополнительный код) (4-разрядный дополнительный код) (4-разрядный дополнительный код) -> Десятичное представление: -6
print("Результат:", result)  # Вывод: Результат: -6
```

#### Побитовый сдвиг влево (<<)

Оператор `<<` сдвигает биты левого операнда влево на число позиций, заданное правым операндом.

Пример

```python
x = 5  # Двоичное представление: 0101

result = x << 2  # Двоичное представление: 010100 -> Десятичное представление: 20
print("Результат:", result)  # Вывод: Результат: 20
```

#### Побитовый сдвиг вправо (>>)

Оператор `>>` сдвигает биты левого операнда вправо на число позиций, заданное правым операндом.

Пример

```python
x = 5  # Двоичное представление: 0101

result = x >> 2  # Двоичное представление: 0001 -> Десятичное представление: 1
print("Результат:", result)  # Вывод: Результат: 1
```

В этой части вы познакомились с побитовыми операторами Python: побитовыми «И» (`&`), «ИЛИ» (`|`), исключающим «ИЛИ» (`^`), «НЕ» (`~`), сдвигом влево (`<<`) и сдвигом вправо (`>>`). Они позволяют работать с отдельными битами двоичного представления целых чисел.
