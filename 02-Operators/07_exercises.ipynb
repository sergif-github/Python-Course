# Модуль 2. Операторы

## Часть 7. Упражнения

### Упражнение 1
Напишите программу, которая присваивает значение переменной с помощью оператора присваивания.

### Упражнение 2
Напишите программу, которая вычисляет сумму двух чисел с помощью оператора сложения.

### Упражнение 3
Напишите программу, которая вычисляет площадь прямоугольника с помощью оператора умножения.

### Упражнение 4
Напишите программу, которая с помощью оператора остатка от деления проверяет, делится ли одно число на другое без остатка.

### Упражнение 5
Напишите программу, которая вычисляет остаток от деления с помощью оператора `%`.

### Упражнение 6
Напишите программу, которая вычисляет среднее арифметическое трёх чисел с помощью оператора деления.

### Упражнение 7
Напишите программу, которая вычисляет квадратный корень числа с помощью оператора возведения в степень.

### Упражнение 8
Напишите программу, которая с помощью оператора `>` проверяет, больше ли одно число другого.

### Упражнение 9
Напишите программу, которая с помощью оператора `==` проверяет, равны ли два числа.

### Упражнение 10
Напишите программу, которая с помощью оператора `!=` проверяет, различаются ли два числа.

### Упражнение 11
Напишите программу, которая с помощью оператора `<=` проверяет, меньше ли одно число другого или равно ему.

### Упражнение 12
Напишите программу, которая применяет логическое «И» к двум логическим значениям.

### Упражнение 13
Напишите программу, которая применяет логическое «НЕ» к одному логическому значению.

### Упражнение 14
Напишите программу, которая применяет побитовое «ИЛИ» к двум целым числам.

### Упражнение 15
Напишите программу, которая применяет побитовое исключающее «ИЛИ» к двум целым числам.

### Упражнение 16
Напишите программу, которая с помощью оператора `==` проверяет, равны ли две строки.

### Упражнение 17
Напишите программу, которая увеличивает значение переменной на 1 с помощью оператора `+=`.

### Упражнение 18
Напишите программу, которая выполняет побитовый сдвиг целого числа.

### Упражнение 19
Напишите программу, которая вычисляет значение выражения со скобками и показывает, как приоритет операторов влияет на результат.

### Упражнение 20
Напишите программу, которая выполняет операцию побитового «И» над целыми числами 10 и 7, и посмотрите, какое значение получится.
