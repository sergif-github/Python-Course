## Часть 6. Приоритет операторов

Приоритет операторов определяет, в каком порядке выполняются операции в выражении. Если в одном выражении есть несколько операторов, операция с более высоким приоритетом обычно выполняется первой.

### 6.1. Приоритет операторов

В Python порядок вычислений подчиняется нескольким правилам.

- Операторы с более высоким приоритетом вычисляются раньше операторов с более низким приоритетом.
- Круглыми скобками можно изменить порядок вычислений, заданный по умолчанию.
- Операторы с одинаковым приоритетом обычно группируются слева направо. Возведение в степень группируется справа налево, а сравнения могут образовывать цепочки.

Ниже перечислены изученные операторы в порядке убывания приоритета.

1. Круглые скобки: `()`
2. Возведение в степень: `**`
3. Унарные плюс и минус: `+x`, `-x`
4. Умножение, деление, деление с округлением вниз, остаток от деления: `*`, `/`, `//`, `%`
5. Сложение и вычитание: `+`, `-`
6. Побитовые сдвиги: `<<`, `>>`
7. Побитовое «И»: `&`
8. Побитовое исключающее «ИЛИ»: `^`
9. Побитовое «ИЛИ»: `|`
10. Операторы сравнения: `<`, `<=`, `>`, `>=`, `==`, `!=`
11. Логическое «НЕ»: `not`
12. Логическое «И»: `and`
13. Логическое «ИЛИ»: `or`

Если порядок вычислений вызывает сомнения, обозначьте его круглыми скобками.

Пример

```python
result = 10 + 2 * 3  # Умножение имеет более высокий приоритет, чем сложение
print("Результат:", result)  # Вывод: Результат: 16

result = (10 + 2) * 3  # Скобки меняют порядок вычислений
print("Результат:", result)  # Вывод: Результат: 36
```

В первом примере скобок нет, поэтому первой выполняется операция умножения. Получается `10 + 6 = 16`. Во втором примере скобки задают другой порядок: сначала выполняется сложение, и получается `(10 + 2) * 3 = 36`.

Знание приоритета операторов помогает писать выражения, которые дают ожидаемый результат.
