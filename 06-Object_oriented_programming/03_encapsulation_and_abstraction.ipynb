## Часть 3. Инкапсуляция и абстракция

### 3.1. Инкапсуляция и сокрытие информации

Инкапсуляция — одно из основных понятий объектно-ориентированного программирования. Она объединяет данные (атрибуты) и поведение (методы) внутри класса. Это позволяет скрывать информацию и управлять доступом к внутреннему состоянию объекта из кода вне класса. Инкапсуляция помогает сохранять целостность данных, защищать их и упрощать сопровождение кода.

```python
class BankAccount:
    def __init__(self, account_number, balance):
        self._account_number = account_number  # Инкапсулированный атрибут
        self._balance = balance  # Инкапсулированный атрибут
        
    def deposit(self, amount):
        self._balance += amount
        
    def withdraw(self, amount):
        if self._balance >= amount:
            self._balance -= amount
        else:
            print("Недостаточно средств.")

# Создаём экземпляр класса BankAccount
account = BankAccount("1234567890", 1000)

# Обращение к атрибутам и вызов методов
account.deposit(500)
account.withdraw(200)
print(account._balance)   # Вывод: 1300
```

В этом примере атрибуты `_account_number` и `_balance` инкапсулированы в классе `BankAccount`. Одно подчёркивание в начале имени указывает, что атрибут предназначен для внутреннего использования. К таким атрибутам следует обращаться в методах класса, где их также можно изменять.

### 3.2. Абстракция и сокрытие данных

Абстракция тесно связана с инкапсуляцией. Абстракция позволяет предоставить внешнему коду только существенную информацию и скрыть лишние подробности. Благодаря ей можно сосредоточиться на основных свойствах и поведении объекта и скрыть детали реализации.

В Python абстракцию можно реализовать, определяя абстрактные классы и интерфейсы с помощью модуля `abc`. Абстрактные классы служат шаблоном для других классов. Нельзя создать экземпляр класса, пока у него остаются абстрактные методы. Абстрактные методы могут не содержать реализации и предназначены для переопределения в подклассах.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def calculate_area(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def calculate_area(self):
        return self.width * self.height

rectangle = Rectangle(4, 5)
print(rectangle.calculate_area())  # Вывод: 20
```

В этом примере `Shape` — абстрактный класс, в котором метод `calculate_area` объявлен абстрактным. Класс `Rectangle` — подкласс `Shape`, который реализует метод `calculate_area`. Абстрактные классы и методы позволяют задать единый интерфейс и потребовать, чтобы конкретные подклассы реализовали нужное поведение.

### 3.3. Уровни доступа — открытый, закрытый и защищённый

В Python приняты соглашения об именовании, которые указывают, как можно обращаться к членам класса — атрибутам и методам. Python не запрещает обращаться к ним напрямую, но на практике используют следующие соглашения.

- Открытый доступ (+)

    Открытые члены класса доступны отовсюду. По соглашению имена открытых атрибутов и методов начинаются без подчёркиваний.

- Закрытый доступ (-)

    Закрытые члены предназначены для использования внутри класса. По соглашению имена закрытых атрибутов и методов начинаются с двух подчёркиваний.

- Защищённый доступ (#)

    Защищённые члены предназначены для использования внутри класса и его подклассов. По соглашению имена защищённых атрибутов и методов начинаются с одного подчёркивания. Python не ограничивает доступ по этому соглашению.

```python
class Person:
    def __init__(self, name, age):
        self.name = name  # Открытый атрибут
        self._age = age  # Защищённый атрибут
        self.__address = '123 Street'  # Закрытый атрибут

    def __display_address(self):
        print(self.__address)  # Закрытый метод

person = Person('John', 25)
print(person.name)  # Вывод: John
print(person._age)  # Вывод: 25
#print(person.__address)  Ошибка: AttributeError
#person.__display_address()   Ошибка: AttributeError
```

В этом примере `name` — открытый атрибут, `_age` — защищённый, а `__address` — закрытый. Закрытые атрибуты и методы предназначены для использования внутри класса.

### 3.4. Управление доступом к атрибутам через свойства

Свойства позволяют управлять доступом к атрибутам, проверять значения, вычислять их и при необходимости защищать от изменения. Свойства поддерживают инкапсуляцию, предоставляя методы чтения и записи значения атрибута (геттеры и сеттеры).

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius
        
    @property
    def radius(self):
        return self._radius
    
    @radius.setter
    def radius(self, value):
        if value > 0:
            self._radius = value
        else:
            raise ValueError("Radius must be positive.")
            
circle = Circle(5)
print(circle.radius)  # Вывод: 5

circle.radius = 7
print(circle.radius)  # Вывод: 7

# circle.radius = -2  Ошибка: ValueError
```

В этом примере доступ к инкапсулированному атрибуту `_radius` осуществляется через свойство `radius` с декоратором `@property` и соответствующим методом записи. Метод записи проверяет, что присваиваемое значение положительно.

### 3.5. Итоги

В третьей части мы рассмотрели инкапсуляцию и сокрытие информации, а также связанное с ними понятие абстракции.

Инкапсуляция объединяет данные и поведение в классе, помогает сохранять целостность данных и упрощает сопровождение кода. Она позволяет скрывать информацию и контролировать доступ к внутреннему состоянию объекта из кода вне класса.

Абстракция позволяет выделять существенную информацию и скрывать лишние детали. Мы разобрали, как в Python использовать для этого абстрактные классы и методы. Абстрактный класс служит шаблоном для других классов. Чтобы создать экземпляр конкретного подкласса, нужно переопределить его абстрактные методы и задать их реализацию.

Мы также рассмотрели, как управлять доступом к атрибутам через свойства, чтобы проверять, вычислять и защищать их значения. Свойства позволяют инкапсулировать атрибуты и управлять доступом к ним через методы чтения и записи.

Понимание инкапсуляции и абстракции помогает писать надёжный код, который проще сопровождать. С помощью инкапсуляции можно организовать код и ограничить доступ к чувствительным данным. Абстракция позволяет сосредоточиться на существенных свойствах и скрывать детали реализации.
