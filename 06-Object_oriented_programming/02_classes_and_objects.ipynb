## Часть 2. Классы и объекты

В объектно-ориентированном программировании класс служит шаблоном для создания объектов. Класс объединяет данные (атрибуты) и поведение (методы) в единую сущность. Объекты — это экземпляры классов. Они обладают атрибутами и поведением, которые определяет их класс.

### 2.1. Определение классов и создание объектов

Вот пример определения класса на Python:

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius
        
    def calculate_area(self):
        return 3.14 * self.radius ** 2

# Создаём экземпляры класса Circle
circle1 = Circle(5)
circle2 = Circle(7)

# Обращение к атрибутам и вызов методов
print(circle1.radius)  # Вывод: 5
print(circle2.radius)  # Вывод: 7
print(circle1.calculate_area())  # Вывод: 78.5
print(circle2.calculate_area())  # Вывод: 153.86
```

В этом примере `Circle` — класс, а `circle1` и `circle2` — его экземпляры. Специальный метод `__init__` вызывается при создании объекта и инициализирует его атрибуты. Метод `calculate_area` вычисляет площадь круга по радиусу.

### 2.2. Атрибуты класса и атрибуты экземпляра

Атрибуты могут принадлежать классу или экземпляру. Атрибуты класса доступны всем его экземплярам, а атрибуты экземпляра принадлежат конкретному объекту.

```python
class Dog:
    species = 'Canine'  # Атрибут класса
    
    def __init__(self, name):
        self.name = name  # Атрибут экземпляра

# Создаём экземпляры класса Dog
dog1 = Dog('Buddy')
dog2 = Dog('Max')

# Обращение к атрибутам класса и экземпляра
print(dog1.species)  # Вывод: Canine
print(dog2.species)  # Вывод: Canine
print(dog1.name)  # Вывод: Buddy
print(dog2.name)  # Вывод: Max
```

В этом примере `species` — атрибут класса, общий для всех экземпляров `Dog`. Атрибут экземпляра `name` задаётся отдельно для каждого объекта этого класса.

### 2.3. Создание и инициализация объектов

Чтобы создать объект в Python, укажите имя класса и круглые скобки. При создании объекта Python автоматически вызывает специальный метод `__init__`, который инициализирует атрибуты объекта.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Создаём экземпляры класса Person
person1 = Person('John', 25)
person2 = Person('Alice', 30)

# Обращение к атрибутам
print(person1.name)  # Вывод: John
print(person2.age)  # Вывод: 30
```


В этом примере метод `__init__` класса `Person` принимает два параметра, `name` и `age`. Создавая экземпляры класса `Person`, мы передаём значения этих параметров, чтобы инициализировать атрибуты `name` и `age`.

### 2.4. Обращение к атрибутам и методам объекта

К атрибутам и методам объекта обращаются через точку. После имени объекта указываются точка и имя атрибута или метода.

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        
    def calculate_area(self):
        return self.width * self.height

# Создаём экземпляр класса Rectangle
rectangle = Rectangle(4, 5)

# Обращение к атрибутам и вызов методов
print(rectangle.width)  # Вывод: 4
print(rectangle.height)  # Вывод: 5
print(rectangle.calculate_area())  # Вывод: 20
```


В этом примере мы создаём экземпляр класса `Rectangle` и через точку обращаемся к его атрибутам `width` и `height` и методу `calculate_area`.

Мы определяли классы на Python и создавали их экземпляры (объекты). Мы также разобрали разницу между атрибутами класса, общими для всех экземпляров, и атрибутами экземпляра, которые принадлежат каждому объекту в отдельности.

Мы обращались к атрибутам и методам через точку. На примерах кругов, собак и прямоугольников мы увидели, как с помощью атрибутов и методов объектов представлять реальные сущности и описывать их поведение.
