# Модуль 6. Объектно-ориентированное программирование

## Часть 1. Введение в ООП

### 1.1. Что такое объектно-ориентированное программирование (ООП)?

Объектно-ориентированное программирование (ООП, англ. OOP) — это парадигма программирования, в основе которой лежит понятие объекта. Объект — экземпляр класса. Он объединяет данные (атрибуты) и поведение (методы) в единую сущность. ООП позволяет организовывать код, моделируя реальные сущности или абстрактные понятия в виде объектов.

### 1.2. Объекты, классы и экземпляры

В ООП объект — это экземпляр класса. Класс задаёт шаблон для создания объектов и определяет, какими атрибутами (данными) и методами (поведением) они будут обладать.

Вот пример простого класса на Python:

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        
    def start_engine(self):
        print("Двигатель автомобиля запускается.")

# Создаём экземпляр класса Car
my_car = Car("Honda", "Civic", 2022)

# Обращение к атрибутам и вызов методов
print(my_car.make)  # Вывод: Honda
print(my_car.model)  # Вывод: Civic
print(my_car.year)  # Вывод: 2022
my_car.start_engine()  # Вывод: Двигатель автомобиля запускается.
```


В этом примере `Car` — класс, а `my_car` — его экземпляр. Специальный метод `__init__` вызывается при создании объекта и инициализирует его атрибуты. Метод `start_engine` задаёт поведение, связанное с классом `Car`.

### 1.3. Основные понятия ООП — инкапсуляция, наследование и полиморфизм

В основе ООП лежат три понятия.

- Инкапсуляция

    Инкапсуляция объединяет данные и связанное с ними поведение внутри класса. Она позволяет скрывать информацию и управлять доступом к внутреннему состоянию объекта извне. Это помогает сохранять целостность данных и ограничивать доступ к ним.
- Наследование

    Наследование позволяет определять новые (производные) классы на основе существующих (базовых или родительских). Производный класс наследует атрибуты и методы родительского класса. Это позволяет повторно использовать код и выстраивать иерархии классов.
- Полиморфизм

    Полиморфизм позволяет работать с объектами разных классов через общий интерфейс базового класса. Один и тот же метод может вести себя по-разному в зависимости от объекта, у которого он вызван. Это помогает делать код гибким, расширяемым и модульным.

### 1.4. Итоги

ООП широко применяют при разработке программ благодаря удобной организации, модульности и сопровождаемости кода. Оно позволяет разбивать сложные задачи на небольшие части, представленные объектами. ООП способствует повторному использованию кода, помогает сократить дублирование и даёт участникам команды понятную структуру для совместной работы.

Понимание ООП помогает решать сложные задачи, создавать модули с кодом для повторного использования и налаживать совместную работу разработчиков. ООП лежит в основе многих языков программирования и фреймворков, поэтому умение применять его важно для разработчика.
