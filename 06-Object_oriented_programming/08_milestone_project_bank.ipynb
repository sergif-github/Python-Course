# Модуль 6. Объектно-ориентированное программирование

## Часть 9. Итоговый проект «Банковский счёт»

Описание

Создайте программу, которая имитирует работу простой системы управления банковскими счетами, используя принципы объектно-ориентированного программирования. Пользователь должен иметь возможность создавать несколько счетов, выполнять операции, например пополнение и снятие средств, и просматривать сведения о счёте.

Требования

- Создайте класс `BankAccount`, представляющий банковский счёт. Предусмотрите атрибуты для номера счёта, имени владельца и остатка на счёте.
- Реализуйте в классе `BankAccount` методы для выполнения операций, например пополнения счёта и снятия средств.
- Реализуйте метод, который выводит сведения о счёте: номер, имя владельца и текущий остаток.

Дополнительные возможности (необязательно)

- Предусмотрите возможность создавать несколько банковских счетов и выполнять операции с каждым из них.

Что учесть

- Тщательно продумайте структуру класса и его атрибуты, чтобы корректно представить банковские счета.
- Применяйте принципы объектно-ориентированного программирования.
- Разбейте программу на небольшие методы или функции, чтобы код было проще организовать и читать.
- Проверьте программу в разных сценариях, включая граничные случаи.
