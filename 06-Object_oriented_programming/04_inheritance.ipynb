## Часть 4. Наследование

Наследование — понятие объектно-ориентированного программирования, которое позволяет создавать новые (производные) классы на основе существующих (родительских). Производные классы наследуют атрибуты и поведение родительских классов. Это позволяет повторно использовать код и расширять возможности этих классов.

### 4.1. Введение в наследование

```python
# Родительский класс
class Vehicle:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color
        
    def accelerate(self):
        print("Транспортное средство разгоняется.")

# Производный класс наследуется от Vehicle
class Car(Vehicle):
    def __init__(self, brand, color, model):
        # Вызываем метод инициализации родительского класса
        super().__init__(brand, color)
        self.model = model
        
    def honk(self):
        print("Автомобиль сигналит.")

# Создаём экземпляр класса Car
my_car = Car("Toyota", "Red", "Camry")

# Обращение к атрибутам и вызов методов
print(my_car.brand)  # Вывод: Toyota
print(my_car.color)  # Вывод: Red
my_car.accelerate()  # Вывод: Транспортное средство разгоняется.
my_car.honk()  # Вывод: Автомобиль сигналит.
```

В этом примере `Vehicle` — родительский класс, а `Car` — производный класс, который от него наследуется.

Производный класс расширяет возможности родительского класса, добавляя атрибут `model` и метод `honk`.

С помощью `super()` производный класс может обращаться к методам родительского класса. В этом примере вызов `super().__init__()` инициализирует у объекта атрибуты `brand` и `color`.

### 4.2. Виды наследования

Python поддерживает несколько видов наследования, включая одиночное, множественное и многоуровневое наследование.

- Одиночное наследование

    Производный класс наследуется от одного родительского класса, получая его атрибуты и поведение.

- Множественное наследование

    Производный класс наследуется от нескольких родительских классов. Он получает атрибуты и поведение из нескольких источников, объединяя возможности этих классов.

- Многоуровневое наследование

    Производный класс наследуется от другого производного класса. Так образуется иерархия наследования, в которой класс получает атрибуты и поведение с нескольких уровней родительских классов.

```python
# Пример множественного наследования
class Animal:
    def __init__(self, name):
        self.name = name
        
    def eat(self):
        print("Животное ест.")

class Flyable:
    def fly(self):
        print("Животное летит.")

# Производный класс наследуется от нескольких родительских классов
class Bird(Animal, Flyable):
    def __init__(self, name, species):
        super().__init__(name)
        self.species = species

    def chirp(self):
        print("Птица щебечет.")

# Создаём экземпляр класса Bird
my_bird = Bird("Sparrow", "Passerine")

# Обращение к атрибутам и вызов методов
print(my_bird.name)  # Вывод: Sparrow
my_bird.eat()  # Вывод: Животное ест.
my_bird.fly()  # Вывод: Животное летит.
my_bird.chirp()  # Вывод: Птица щебечет.
```

В этом примере `Bird` наследуется одновременно от `Animal` и `Flyable`. Это пример множественного наследования. Производный класс получает атрибуты и поведение от обоих родительских классов и добавляет собственные атрибуты и методы.

Понимание наследования помогает строить иерархии классов, повторно использовать код и создавать сложные системы, опираясь на возможности родительских классов. В следующих частях мы рассмотрим другие понятия ООП, в том числе переопределение и перегрузку методов, а также полиморфизм.

### 4.3. Итоги

Мы рассмотрели наследование в объектно-ориентированном программировании. Наследование позволяет создавать производные классы, которые получают атрибуты и поведение родительских, повторно используют код этих классов и расширяют их возможности. При одиночном наследовании у производного класса один родительский класс, при множественном — несколько. При многоуровневом наследовании производный класс наследуется от другого производного класса. Понимание наследования помогает строить модульный, расширяемый и эффективный код.
