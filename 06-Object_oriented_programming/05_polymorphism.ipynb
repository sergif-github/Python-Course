## Часть 5. Полиморфизм

Полиморфизм в объектно-ориентированном программировании позволяет работать с объектами разных классов через интерфейс общего базового класса. Один и тот же код может работать с разными типами объектов. Это позволяет повторно использовать код и делает его гибче.

### 5.1. Введение в полиморфизм

```python
# Родительский класс
class Animal:
    def sound(self):
        pass

# Производные классы
class Dog(Animal):
    def sound(self):
        print("Собака лает!")

class Cat(Animal):
    def sound(self):
        print("Кошка мяукает!")

# Общая функция, принимающая объекты классов семейства Animal
def make_sound(animal):
    animal.sound()

# Создаём экземпляры разных классов
dog = Dog()
cat = Cat()

# Вызываем общую функцию для разных объектов
make_sound(dog)  # Вывод: Собака лает!
make_sound(cat)  # Вывод: Кошка мяукает!
```

В этом примере `Animal` — базовый класс, а `Dog` и `Cat` — производные классы. Каждый производный класс переопределяет метод `sound`, задавая собственную реализацию. Функция `make_sound` принимает объект семейства классов `Animal` и вызывает его метод `sound`. Это пример полиморфизма. Передавая разные объекты в одну функцию, можно получить разные действия в зависимости от класса объекта.

### 5.2. Переопределение методов

При переопределении подкласс задаёт собственную реализацию метода, который уже определён в родительском классе. Имя метода сохраняется, а реализация может отличаться.

```python
class Animal:
    def sound(self):
        print("Животное издаёт звук.")

class Dog(Animal):
    def sound(self):
        print("Собака лает.")

animal = Animal()
animal.sound()

dog = Dog()
dog.sound()
```

В этом примере в классе `Animal` есть метод `sound()`. Класс `Dog` наследуется от `Animal` и переопределяет `sound()`, задавая собственную реализацию. При вызове `sound()` у экземпляров `Animal` и `Dog` выполняется реализация соответствующего класса. Поэтому вывод различается.

### 5.3. Перегрузка методов

Перегрузка методов позволяет определить в одном классе несколько методов с одинаковым именем и разными параметрами. Python не поддерживает традиционную перегрузку методов, как некоторые другие языки. Похожего поведения можно добиться с помощью значений по умолчанию или переменного числа аргументов.

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

calculator = Calculator()
print(calculator.add(2, 3))
print(calculator.add(2, 3, 4))
```

В этом примере у класса `Calculator` есть метод `add()`, которому можно передавать два или три аргумента. У третьего параметра `c` есть значение по умолчанию, поэтому методу достаточно двух аргументов. Это даёт эффект, похожий на перегрузку. Один метод работает с разными наборами аргументов.
