# Модуль 6. Объектно-ориентированное программирование

## Часть 6. Упражнения

### Упражнение 1

Создайте иерархию классов для геометрических фигур. Она должна состоять из базового класса `Shape` и двух производных классов `Rectangle` и `Circle`. В классе `Shape` объявите абстрактный метод `calculate_area()`, который нужно переопределить в производных классах. В классе `Rectangle` предусмотрите атрибуты длины и ширины, а в классе `Circle` — атрибут радиуса. Реализуйте необходимые методы и продемонстрируйте полиморфизм, рассчитав площади прямоугольника и круга.

### Упражнение 2

Создайте иерархию классов для животных. Она должна состоять из базового класса `Animal` и двух производных классов `Dog` и `Cat`. В классе `Animal` объявите абстрактный метод `make_sound()`, который нужно переопределить в производных классах. Реализуйте необходимые методы и продемонстрируйте полиморфизм. Собака и кошка должны издавать разные, характерные для них звуки.

### Упражнение 3

Создайте иерархию классов для сотрудников. Она должна состоять из базового класса `Employee` и двух производных классов `Manager` и `Developer`. В классе `Employee` предусмотрите атрибуты имени и зарплаты и методы для получения и изменения зарплаты. В классе `Manager` добавьте атрибут отдела, а в классе `Developer` — атрибут языка программирования. Реализуйте необходимые методы и продемонстрируйте полиморфизм, наследование и инкапсуляцию. Для этого создайте экземпляры классов руководителя и разработчика и обратитесь к их атрибутам и методам.
