## Часть 2. Числовые типы данных (`int`, `float`)

В Python для целых чисел и чисел с плавающей точкой обычно используют два типа данных: `int` и `float` соответственно.

### 2.1. Целые числа (`int`)

Тип `int` представляет целые числа, то есть числа без дробной части. Они могут быть положительными, отрицательными или равными нулю.

Рассмотрим пример.

```python
age = 25
quantity = -10
population = 1000000
```

С числами типа `int` можно выполнять арифметические операции: сложение, вычитание, умножение и деление. Если оба операнда — целые числа, сложение, вычитание и умножение дают результат типа `int`. Оператор обычного деления `/` возвращает результат типа `float`.

Рассмотрим пример.

```python
x = 10
y = 3

sum = x + y
difference = x - y
product = x * y
quotient = x / y

print("Сумма:", sum)
print("Разность:", difference)
print("Произведение:", product)
print("Частное:", quotient)
```

### 2.2. Числа с плавающей точкой (`float`)

Тип `float` позволяет представлять числа с дробной частью. Он подходит для работы как с малыми, так и с большими числами с плавающей точкой.

Рассмотрим пример.

```python
pi = 3.14159
temperature = -10.5
price = 19.99
```

С числами с плавающей точкой также можно выполнять арифметические операции.

```python
x = 5.0
y = 2.0

sum = x + y
difference = x - y
product = x * y
quotient = x / y

print("Сумма:", sum)
print("Разность:", difference)
print("Произведение:", product)
print("Частное:", quotient)
```

Когда в арифметических операциях из примера один операнд имеет тип `int`, а другой — `float`, результат имеет тип `float`.

```python
x = 10
y = 3.0

result = x / y

print("Результат:", result)
```

### 2.3. Итоги

Вы познакомились с типами данных `int` и `float`. Они представляют соответственно целые числа и числа с плавающей точкой. Оба типа поддерживают арифметические операции и применяются в вычислениях.
