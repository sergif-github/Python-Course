# Модуль 1. Переменные и типы данных

## Часть 1. Переменные и присваивание

В Python переменные служат для хранения данных. Имя переменной связывается со значением: числом, строкой или другим объектом. Связывание имени со значением называется присваиванием.

### 1.1. Имена переменных

При выборе имён переменных учитывайте следующие правила.

- Имя может содержать буквы, цифры и символ подчёркивания (`_`). Например, можно использовать латинские буквы `a–z` и `A–Z`, а также цифры `0–9`. Python поддерживает и многие буквы Юникода.
- Имя не может начинаться с цифры.
- Регистр букв имеет значение. Например, `myVariable` и `myvariable` — разные имена.
- Выбирайте понятные имена, которые отражают назначение переменных.

Рассмотрим несколько допустимых имён.

```python
age = 25
name = "Alice"
is_student = True
```

### 1.2. Присваивание значения переменной

В Python для присваивания используют знак `=`. Имя слева от `=` связывается со значением выражения справа.

Рассмотрим пример.

```python
x = 10
y = 5.5
message = "Hello, World!"
```

В одной строке можно присвоить значения сразу нескольким переменным.

```python
a, b, c = 1, 2, 3
```

### 1.3. Повторное присваивание

В Python переменную можно связать с новым значением. Это называется повторным присваиванием. После него имя переменной указывает на новое значение.

Рассмотрим пример.

```python
x = 10
print(x)  # Результат: 10

x = 20
print(x)  # Результат: 20
```

### 1.4. Вывод значений переменных

Функция `print()` выводит значение переменной. Это позволяет увидеть её текущее значение во время выполнения программы.

Рассмотрим пример.

```python
name = "Alice"
print(name)  # Результат: Alice

age = 25
print("Возраст:", age)  # Результат: Возраст: 25
```

Функции `print()` можно передать один или несколько аргументов через запятую. Она преобразует каждый аргумент в строку и выводит её в консоль.

Рассмотрим пример.

```python
name = "Alice"
age = 25
country = "USA"

print(name, age, country)
```

### 1.5. Итоги
