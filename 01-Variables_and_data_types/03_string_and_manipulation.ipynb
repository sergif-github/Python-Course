## Часть 3. Строки и работа с ними

В Python строка — это последовательность символов. Строковый литерал записывают в одинарных (`'`) или двойных (`"`) кавычках. Строки используются для представления текста. Они неизменяемы: после создания строки изменить её содержимое нельзя.

### 3.1. Создание строк

Чтобы создать строку, заключите текст в одинарные или двойные кавычки.

```python
name = 'Alice'
message = "Hello, World!"
```

Для записи строк можно использовать как одинарные, так и двойные кавычки. Открывающая и закрывающая кавычки должны быть одного вида.

### 3.2. Операции со строками

Python предоставляет разные операции и методы для работы со строками. Рассмотрим распространённые приёмы.

#### 3.2.1. Конкатенация

Оператор `+` позволяет объединить две или несколько строк в одну. Это называется конкатенацией.

```python
first_name = "John"
last_name = "Doe"

full_name = first_name + " " + last_name
print(full_name)
```

#### 3.2.2. Длина строки и подсчёт вхождений

Функция `len()` возвращает длину строки, то есть количество её символов.

```python
message = "Hello, World!"
length = len(message)
print("Длина:", length)
```

Метод `count()` возвращает число неперекрывающихся вхождений подстроки. Например, так можно подсчитать, сколько раз в строке встречается заданное слово как последовательность символов.

```python
message = "Hello, World!"
word = "Hello"
occurrences = message.count(word)
print("Число вхождений:", occurrences)
```

#### 3.2.3. Обращение к символам

Чтобы получить отдельный символ строки, используйте его индекс. Индексация начинается с `0`: это индекс первого символа.

```python
message = "Hello, World"
first_char = message[0]
print("Первый символ:", first_char)
```

Отрицательные индексы позволяют обращаться к символам с конца строки. Индекс `-1` соответствует последнему символу.

```python
last_char = message[-1]
print("Последний символ:", last_char)
```

#### 3.2.4. Срезы

Чтобы извлечь подстроку, используйте срез. Срез задаёт диапазон индексов и возвращает соответствующую часть строки.

```python
message = "Hello, World!"
substring = message[7:12]
print("Подстрока:", substring)
```

### 3.3. Методы строк

У строк в Python есть много встроенных методов для обработки текста. Рассмотрим часто используемые методы.

```python
message = "Hello, World!"

# Переводим строку в верхний регистр
uppercase = message.upper()
print("Верхний регистр:", uppercase)  # Результат: Верхний регистр: HELLO, WORLD!

# Переводим строку в нижний регистр
lowercase = message.lower()
print("Нижний регистр:", lowercase)  # Результат: Нижний регистр: hello, world!

# Заменяем подстроку
replaced = message.replace("Hello", "Hi")
print("После замены:", replaced)  # Результат: После замены: Hi, World!

# Разбиваем строку на список
words = message.split(", ")
print("Слова:", words)  # Результат: Слова: ['Hello', 'World!']
```
