## Часть 6. Работа с коллекциями

Коллекции в Python — структуры данных, позволяющие хранить несколько значений вместе. В языке есть несколько встроенных типов коллекций. Каждый предоставляет свои способы организации и обработки данных.

### 6.1. Списки

Список — упорядоченная коллекция, которая может содержать элементы разных типов. Списки изменяемы: после создания можно менять их элементы. Для записи списка заключите элементы в квадратные скобки `[]` и разделите их запятыми.

Рассмотрим пример.

```python
fruits = ['apple', 'banana', 'cherry']
numbers = [1, 2, 3, 4, 5]
mixed = [1, 'apple', True]

print(fruits)  # Результат: ['apple', 'banana', 'cherry']
print(numbers)  # Результат: [1, 2, 3, 4, 5]
print(mixed)  # Результат: [1, 'apple', True]
```

Списки поддерживают разные операции и методы, с помощью которых можно добавлять, удалять и изменять элементы.

```python
fruits = ['apple', 'banana', 'cherry']
numbers = [1, 2, 3, 4, 5]
mixed = [1, 'apple', True]

# Добавляем элемент в конец списка
fruits.append('orange')
print(fruits)  # Результат: ['apple', 'banana', 'cherry', 'orange']

# Получаем длину списка
length = len(numbers)
print(length)  # Результат: 5

# Получаем элемент по индексу
item = mixed[1]
print(item)  # Результат: 'apple'

# Удаляем элемент из списка
numbers.remove(3)
print(numbers)  # Результат: [1, 2, 4, 5]

# Проверяем наличие элемента в списке
exists = 'apple' in fruits
print(exists)  # Результат: True

# Подсчитываем число вхождений элемента в список
count = mixed.count(1)
print(count)  # Результат: 2 (True == 1)

# Располагаем элементы списка в обратном порядке
fruits.reverse()
print(fruits)  # Результат: ['orange', 'cherry', 'banana', 'apple']

# Сортируем список по возрастанию
numbers.sort()
print(numbers)  # Результат: [1, 2, 4, 5]
```

### 6.2. Кортежи

Кортеж — упорядоченная коллекция, похожая на список. Кортеж неизменяем: после создания нельзя заменить его элементы. Кортежи обычно записывают в круглых скобках, разделяя элементы запятыми.

Рассмотрим пример.

```python
coordinates = (3, 4)
person = ('John', 25, 'Engineer')

print(coordinates)  # Результат: (3, 4)
print(person)  # Результат: ('John', 25, 'Engineer')
```

Кортежи часто используют, чтобы хранить несколько связанных значений, когда изменение состава коллекции не требуется.

Рассмотрим операции и методы кортежей.


```python
coordinates = (3, 4)
person = ('John', 25, 'Engineer')

# Получаем элемент по индексу
x = coordinates[0]
print(x)  # Результат: 3

# Получаем длину кортежа
length = len(person)
print(length)  # Результат: 3

# Объединяем два кортежа
combined = coordinates + person
print(combined)  # Результат: (3, 4, 'John', 25, 'Engineer')

# Подсчитываем число вхождений элемента в кортеж
count = person.count('Engineer')
print(count)  # Результат: 1

# Находим индекс элемента в кортеже
index = person.index('John')
print(index)  # Результат: 0

# Проверяем наличие элемента в кортеже
exists = 25 in person
print(exists)  # Результат: True
```

### 6.3. Множества

Множество — неупорядоченная коллекция уникальных элементов. Множества изменяемы: можно добавлять и удалять элементы. Чтобы создать непустое множество, запишите его элементы в фигурных скобках. Также можно воспользоваться конструктором `set()`.

Рассмотрим пример.

```python
fruits = {'apple', 'banana', 'cherry'}
numbers = set([1, 2, 3, 4, 5])

print(fruits)  # Пример результата, порядок элементов может отличаться: {'apple', 'banana', 'cherry'}
print(numbers)  # Результат: {1, 2, 3, 4, 5}
```

Множества полезны, когда нужно найти уникальные элементы, проверить принадлежность элемента множеству или выполнить операции над множествами: объединение, пересечение или разность.

Рассмотрим операции и методы множеств.


```python
fruits = {'apple', 'banana', 'cherry'}
numbers = set([1, 2, 3, 4, 5])

# Добавляем элемент в множество
fruits.add('orange')
print(fruits)  # Пример результата, порядок элементов может отличаться: {'apple', 'banana', 'cherry', 'orange'}

# Удаляем элемент из множества
numbers.remove(3)
print(numbers)  # Результат: {1, 2, 4, 5}

# Проверяем принадлежность элемента множеству
exists = 'apple' in fruits
print(exists)  # Результат: True

# Получаем число элементов множества
length = len(numbers)
print(length)  # Результат: 4

# Находим объединение двух множеств
combined = fruits.union(numbers)
print(combined)  # Пример результата, порядок элементов может отличаться: {1, 2, 4, 5, 'banana', 'apple', 'cherry', 'orange'}

# Находим пересечение двух множеств
intersection = fruits.intersection(numbers)
print(intersection)  # Результат: set()

# Находим разность двух множеств
difference = fruits.difference(numbers)
print(difference)  # Пример результата, порядок элементов может отличаться: {'banana', 'apple', 'cherry', 'orange'}
```

### 6.4. Словари

Словарь — коллекция пар «ключ — значение». Каждому уникальному ключу соответствует значение. Это позволяет быстро находить значения по ключам. Начиная с Python 3.7, порядок обхода словаря соответствует порядку добавления ключей. Чтобы создать словарь, запишите пары в фигурных скобках `{}` или вызовите конструктор `dict()`.

Рассмотрим пример.

```python
person = {'name': 'John', 'age': 25, 'occupation': 'Engineer'}

print(person)  # Результат: {'name': 'John', 'age': 25, 'occupation': 'Engineer'}
```

Словари используют для хранения данных и получения значений по определённым ключам.

Рассмотрим операции и методы словарей.

```python
person = {'name': 'John', 'age': 25, 'occupation': 'Engineer'}

# Получаем значение по ключу
name = person['name']
print(name)  # Результат: 'John'

# Получаем ключи словаря
keys = person.keys()
print(keys)  # Результат: dict_keys(['name', 'age', 'occupation'])

# Получаем значения словаря
values = person.values()
print(values)  # Результат: dict_values(['John', 25, 'Engineer'])

# Проверяем наличие ключа в словаре
exists = 'age' in person
print(exists)  # Результат: True

# Добавляем пару «ключ — значение» в словарь
person['city'] = 'New York'
print(person)  # Результат: {'name': 'John', 'age': 25, 'occupation': 'Engineer', 'city': 'New York'}

# Удаляем пару «ключ — значение» из словаря
removed_value = person.pop('occupation')
print(removed_value)  # Результат: 'Engineer'
print(person)  # Результат: {'name': 'John', 'age': 25, 'city': 'New York'}

# Удаляем все пары «ключ — значение» из словаря
person.clear()
print(person)  # Результат: {}
```

### 6.5. Различия между коллекциями и эффективность операций

У каждого типа коллекции свои свойства, которые влияют на эффективность и применимость в разных задачах. При небольшом объёме данных разница в скорости может быть незаметной. При выборе коллекции учитывайте объём данных, характер и частоту операций и требования к программе.

Рассмотрим свойства коллекций и операции, для которых они подходят.

Списки

- Свойства. Списки изменяемы, упорядочены и допускают повторяющиеся элементы.
- Эффективность. Доступ по индексу позволяет быстро прочитать или заменить элемент. Поиск элемента в большом списке может быть медленным: в худшем случае придётся перебрать весь список. Добавлять и удалять элементы в конце списка эффективно. При вставке или удалении в середине нужно сдвинуть следующие элементы.

Кортежи

- Свойства. Кортежи упорядочены, как и списки. Состав кортежа после создания изменить нельзя.
- Эффективность. Доступ по индексу примерно так же эффективен, как в списках. Кортеж можно использовать как ключ словаря или элемент множества, если все его элементы хешируемы. На хранение кортежа обычно уходит немного меньше памяти, чем на хранение списка с теми же элементами.

Множества

- Свойства. Множества изменяемы, не имеют заданного порядка и хранят только уникальные элементы.
- Эффективность. Множества подходят для быстрой проверки принадлежности и уникальности. Они используют хеширование, поэтому добавление, удаление и проверка принадлежности обычно работают быстро и на больших объёмах данных. Доступ к элементам по индексу у множества отсутствует.

Словари

- Свойства. Словари изменяемы, хранят пары «ключ — значение» и сохраняют порядок добавления ключей в современных версиях Python.
- Эффективность. Словари используют хеш-таблицы для хранения пар и быстрого поиска значений по уникальным ключам, в том числе на больших объёмах данных. Добавление, изменение и удаление пар обычно тоже выполняются быстро.

Выбор коллекции зависит от операций и задачи. Если существенны порядок элементов и возможность хранить повторяющиеся значения, подойдут списки или кортежи. Если нужны уникальные элементы и быстрая проверка принадлежности, удобно использовать множество. Для хранения соответствий «ключ — значение» и быстрого поиска по ключу подходят словари.
