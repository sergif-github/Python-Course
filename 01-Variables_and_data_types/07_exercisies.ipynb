# Модуль 1. Переменные и типы данных

## Часть 7. Упражнения

### Упражнение 1
Напишите программу на Python, которая присваивает переменной `x` значение `5` и выводит его.

### Упражнение 2
Напишите программу на Python, которая в одной строке присваивает переменным `a`, `b` и `c` значения `10`, `20` и `30` соответственно. Выведите значения этих переменных.

### Упражнение 3
Напишите программу на Python, которая запрашивает у пользователя два числа, вычисляет их сумму и выводит результат.

### Упражнение 4
Напишите программу на Python, которая запрашивает у пользователя два числа, вычисляет их разность и выводит результат.

### Упражнение 5
Напишите программу на Python, которая запрашивает у пользователя два числа, вычисляет их произведение и выводит результат.

### Упражнение 6
Напишите программу на Python, которая запрашивает у пользователя два числа, делит первое число на второе и выводит результат.

### Упражнение 7
Напишите программу на Python, которая запрашивает у пользователя имя и сохраняет его в переменной `name`. Выведите значение `name`.

### Упражнение 8
Напишите программу на Python, которая отдельно запрашивает у пользователя имя и фамилию. Объедините две строки и выведите полное имя.

### Упражнение 9
Напишите программу на Python, которая запрашивает у пользователя предложение и вычисляет его длину. Выведите длину предложения.

### Упражнение 10
Напишите программу на Python, которая запрашивает у пользователя предложение и слово для поиска. Подсчитайте, сколько раз это слово встречается в предложении, и выведите количество вхождений.

### Упражнение 11
Напишите программу на Python, которая запрашивает у пользователя слово и сохраняет его в переменной `word`. Получите первый символ слова и выведите его.

### Упражнение 12
Напишите программу на Python, которая запрашивает у пользователя предложение и сохраняет его в переменной `sentence`. С помощью среза выведите первые три символа предложения.

### Упражнение 13
Напишите программу на Python, которая запрашивает у пользователя слово. С помощью срезов извлеките первые три символа, последние три символа и подстроку с четвёртого символа до конца слова. Выведите полученные подстроки.

### Упражнение 14
Напишите программу на Python, которая запрашивает у пользователя слово. С помощью среза расположите символы слова в обратном порядке и выведите результат.

### Упражнение 15
Напишите программу на Python, которая запрашивает у пользователя слово и сохраняет его в переменной `word`. Методом `upper()` переведите слово в верхний регистр и выведите результат.

### Упражнение 16
Напишите программу на Python, которая запрашивает у пользователя предложение. Переведите предложение в нижний регистр и выведите результат.

### Упражнение 17
Напишите программу на Python, которая запрашивает у пользователя предложение. Методом `split()` разбейте предложение на слова и выведите полученный список.

### Упражнение 18
Напишите программу на Python, которая задаёт две логические переменные, `is_raining` и `is_sunny`. Выведите их значения.

### Упражнение 19
Напишите программу на Python, которая запрашивает у пользователя возраст. С помощью оператора `>` сравните введённый возраст с заранее заданным порогом совершеннолетия, например `18`. По результату сравнения выведите, считается ли пользователь совершеннолетним по этому условию.

### Упражнение 20
Напишите программу на Python, которая запрашивает у пользователя число. С помощью оператора сравнения `>` и логического оператора `and` проверьте, что число положительное и больше `10`. Выведите результат проверки.

### Упражнение 21
Напишите программу на Python, которая запрашивает у пользователя число. С помощью логического оператора И (`and`) проверьте, делится ли число одновременно на `2` и на `3`. Выведите результат проверки.

### Упражнение 22
Напишите программу на Python, которая задаёт три логические переменные: `is_student`, `has_homework` и `is_weekend`. Присвойте им подходящие значения. Напишите условную инструкцию с логическими операторами (`and`, `or`), чтобы определить, нужно ли студенту выполнять домашнее задание. Выведите результат.

### Упражнение 23
Напишите программу на Python, которая запрашивает у пользователя целое число. Преобразуйте число в строку функцией `str()` и сохраните результат в переменной. Выведите тип переменной.

### Упражнение 24
Напишите программу на Python, которая запрашивает у пользователя строку с числом. Преобразуйте строку в целое число функцией `int()` и сохраните результат в переменной. Выведите тип переменной.

### Упражнение 25
Напишите программу на Python, которая задаёт число с плавающей точкой. Преобразуйте его в целое число функцией `int()` и сохраните результат в переменной. Выведите тип переменной.

### Упражнение 26
Создайте пустой список `my_list`. Добавьте в него числа `1`, `2`, `3` и `4`. Удалите число `3` и выведите итоговое содержимое списка.

### Упражнение 27
Создайте кортеж `my_tuple` со значениями `'apple'`, `'banana'`, `'cherry'` и `'date'`. Получите второй элемент кортежа и выведите его.

### Упражнение 28
Создайте множество `set1` со значениями `1`, `2`, `3` и `4`, а также множество `set2` со значениями `3`, `4`, `5` и `6`. Выведите объединение и пересечение множеств `set1` и `set2`.

### Упражнение 29
Создайте словарь `my_dict` с ключами `'name'`, `'age'` и `'city'` и соответствующими значениями. Выведите значение по ключу `'age'`. Добавьте ключ `'country'` со значением `'USA'` и выведите обновлённый словарь.

### Упражнение 30
Создайте словарь `student_scores` с ключами `'Alice'`, `'Bob'` и `'Charlie'`. Каждому ключу сопоставьте оценку по 100-балльной шкале. Вычислите средний балл, сохраните его в переменной `average_score` и выведите.
