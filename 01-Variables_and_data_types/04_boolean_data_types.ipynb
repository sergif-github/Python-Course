## Часть 4. Логический тип данных

Тип `bool` в Python представляет логические значения. Возможны два значения: `True` (истина) и `False` (ложь). Логические значения используются в логических операциях, проверках условий и сравнениях. С их помощью программа выбирает, что делать дальше в зависимости от выполнения условия.

### 4.1. Логические значения

Значения `True` и `False` — ключевые слова Python, обозначающие истину и ложь.

Рассмотрим пример.

```python
is_raining = True
is_sunny = False
print(is_raining)
print(is_sunny)
```

### 4.2. Операторы сравнения

В Python есть операторы, которые сравнивают значения и дают логический результат. Вот распространённые операторы сравнения.

- `==` — равно.
- `!=` — не равно.
- `>` — больше.
- `<` — меньше.
- `>=` — больше или равно.
- `<=` — меньше или равно.

Операторы сравнивают два значения и возвращают `True` или `False` в зависимости от результата.

Рассмотрим пример.

```python
x = 10
y = 5

result1 = x > y
result2 = x == y
result3 = x != y
print(result1)
print(result2)
print(result3)
```

### 4.3. Логические операторы

Логические операторы Python позволяют объединять логические значения или условия. В языке есть три таких оператора.

- `and` — логическое И.
- `or` — логическое ИЛИ.
- `not` — логическое НЕ.

В следующих примерах эти операторы применяются к логическим значениям и дают логические результаты.

```python
x = 10
y = 5

result1 = x > 0 and y > 0
result2 = x > 10 or y > 10
result3 = not(x > y)
print(result1)
print(result2)
print(result3)
```

### 4.4. Итоги

Вы познакомились с логическим типом данных в Python. Логические значения обозначают истину и ложь и используются в условных инструкциях, логических операциях и сравнениях. Они помогают принимать решения и управлять ходом выполнения программы.
