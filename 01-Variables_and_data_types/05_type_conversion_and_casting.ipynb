## Часть 5. Преобразование и приведение типов

Преобразование типов позволяет получить значение нужного типа из исходного значения. Python может выполнять некоторые преобразования автоматически, например при арифметических операциях со смешанными числовыми типами.

### 5.1. Неявное преобразование типов

В следующем примере складываются целое число (`int`) и число с плавающей точкой (`float`).

```python
x = 10
y = 3.5

result = x + y  # Неявное преобразование int в float
print("Тип результата:", type(result))
print("Результат:", result)
```

В этом примере при сложении с `y` целое значение `x` неявно преобразуется в число типа `float`.

### 5.2. Явное приведение типов

Чтобы явно преобразовать значение из одного типа в другой, можно использовать встроенные функции преобразования.

- `int()` — преобразует значение в целое число.
- `float()` — преобразует значение в число с плавающей точкой.
- `str()` — преобразует значение в строку.
- `bool()` — преобразует значение в логическое значение.

Рассмотрим пример.

```python
# Преобразуем значение в целое число
value1 = "10"
integer_value = int(value1)
print(integer_value)  # Результат: 10
print(type(integer_value))  # Результат: <class 'int'>

# Преобразуем значение в число с плавающей точкой
value2 = "3.14"
float_value = float(value2)
print(float_value)  # Результат: 3.14
print(type(float_value))  # Результат: <class 'float'>

# Преобразуем значение в строку
value3 = 42
string_value = str(value3)
print(string_value)  # Результат: 42
print(type(string_value))  # Результат: <class 'str'>

# Преобразуем значение в логическое
value4 = 0
bool_value = bool(value4)
print(bool_value)  # Результат: False
print(type(bool_value))  # Результат: <class 'bool'>
```

В этом примере используются значения `"10"`, `"3.14"`, `42` и `0`. С помощью функций `int()`, `float()`, `str()` и `bool()` они преобразуются в нужные типы. Затем функция `print()` выводит преобразованные значения и их типы.

### 5.3. Итоги

Вы познакомились с преобразованием и приведением типов в Python. При неявном преобразовании Python автоматически преобразует значения. Для явного преобразования нужно вызвать подходящую функцию. Эти приёмы позволяют приводить данные к форме, удобной для обработки в программе.
