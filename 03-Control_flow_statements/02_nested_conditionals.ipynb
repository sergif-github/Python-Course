## Часть 2. Вложенные условные инструкции

Вложенная условная инструкция находится внутри другой условной инструкции. Вложенность позволяет проверять несколько условий и выполнять разные блоки кода в зависимости от результатов проверок.

Ниже приведён пример вложенных условных инструкций.

```python
num = 10

if num > 0:
    print("Число положительное.")
    if num % 2 == 0:
        print("Число чётное.")
    else:
        print("Число нечётное.")
else:
    print("Число неположительное.")
```

В этом примере внешняя инструкция `if` проверяет, больше ли `num` нуля. Если условие выполняется, программа выводит «Число положительное.». Затем во вложенной инструкции `if` с ветвью `else` программа проверяет, делится ли `num` на 2, и определяет, чётное оно или нечётное.

Число уровней вложенности зависит от сложности условий, которые нужно проверить. Выравнивайте отступы в блоках кода, чтобы обозначить уровни вложенности.

Следите за отступами, чтобы уровни вложенности были видны, а код легко читался.
