# Модуль 3. Управление ходом выполнения программы

## Часть 5. Упражнения

### Упражнение 1
Напишите программу на Python, которая запрашивает возраст пользователя. Если возраст 18 лет или больше, выведите «У вас есть право голосовать.». В противном случае выведите «У вас нет права голосовать.».

### Упражнение 2
Напишите программу на Python, которая запрашивает число. Если число положительное, выведите «Число положительное.». Если отрицательное, выведите «Число отрицательное.». Если число равно нулю, выведите «Число равно нулю.».

### Упражнение 3
Напишите программу на Python, которая запрашивает два числа. Сравните числа и выведите большее с сообщением «Большее число: [большее число]».

### Упражнение 4
Напишите программу, которая запрашивает два числа и выводит большее из них с помощью тернарного оператора.

### Упражнение 5
Напишите программу на Python, которая запрашивает оценку в виде целого числа от 0 до 100. С помощью вложенных условных инструкций определите буквенную оценку по следующим правилам.

- При 90 баллах и выше выведите `A`.
- При 80–89 баллах выведите `B`.
- При 70–79 баллах выведите `C`.
- При 60–69 баллах выведите `D`.
- При результате ниже 60 баллов выведите `F`.

### Упражнение 6
Напишите программу на Python, которая запрашивает число. С помощью логических операторов в условной инструкции проверьте, делится ли число одновременно на 2 и на 3. Если условие выполняется, выведите «Число делится и на 2, и на 3.». В противном случае выведите «Число не делится одновременно на 2 и на 3.».

### Упражнение 7
Напишите программу, которая запрашивает число. Если число положительное, проверьте, делится ли оно на 2. Если делится, сообщите, что это положительное чётное число. Если не делится, сообщите, что это положительное нечётное число. Если число отрицательное, выведите сообщение об этом.

### Упражнение 8
Напишите программу, которая запрашивает число и с помощью тернарного оператора сообщает, положительное оно или отрицательное.

### Упражнение 9
Напишите программу на Python, которая запрашивает возраст пользователя. С помощью тернарного оператора проверьте, исполнилось ли пользователю 18 лет. Если да, выведите «Вы совершеннолетний человек.». Иначе выведите «Вы несовершеннолетний человек.».

### Упражнение 10
Напишите программу, которая запрашивает возраст пользователя и сообщает, к какой возрастной группе он относится: дети, подростки, взрослые или пожилые люди.
