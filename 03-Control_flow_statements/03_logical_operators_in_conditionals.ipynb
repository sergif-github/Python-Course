## Часть 3. Логические операторы в условных инструкциях

Логические операторы объединяют несколько условий в условной инструкции. В Python операторы `and`, `or` и `not` часто используют для логических операций над логическими значениями. В условиях они работают следующим образом.

- Оператор `and` даёт `True`, если оба условия, слева и справа, равны `True`.
- Оператор `or` даёт `True`, если хотя бы одно из условий слева или справа равно `True`.
- Оператор `not` возвращает логическое отрицание условия. Если условие равно `True`, `not` возвращает `False`, и наоборот.

С помощью логических операторов из нескольких простых условий можно составить более сложное.

Пример применения логических операторов в условной инструкции приведён ниже.

```python
age = 25
country = "USA"

if age >= 18 and country == "USA":
    print("У вас есть право голосовать в США.")
else:
    print("У вас нет права голосовать в США.")
```

В примере оператор `and` объединяет два условия: `age >= 18` и `country == "USA"`. Если оба условия равны `True`, выполняется блок внутри `if` и выводится сообщение «У вас есть право голосовать в США.». Иначе выполняется блок внутри `else` и выводится сообщение «У вас нет права голосовать в США.».

Оператор `or` позволяет составить условие, для которого достаточно, чтобы хотя бы одна из проверок дала `True`.

Пример

```python
score = 85

if score >= 90 or score <= 60:
    print("Вы сдали экзамен.")
else:
    print("Вы не сдали экзамен.")
```

В этом примере оператор `or` объединяет условия `score >= 90` и `score <= 60`. Если хотя бы одно из них равно `True`, выполняется блок внутри `if` и выводится сообщение «Вы сдали экзамен.». Иначе выполняется блок внутри `else` и выводится сообщение «Вы не сдали экзамен.».

Для логического отрицания условия используется оператор `not`.

Пример

```python
is_logged_in = False

if not is_logged_in:
    print("Войдите в систему, чтобы продолжить.")
```

В этом примере оператор `not` проверяет, равно ли `is_logged_in` значению `False`. Если условие `not is_logged_in` истинно, выполняется блок внутри `if` и выводится сообщение «Войдите в систему, чтобы продолжить.».

### Итоги

В этой части вы узнали, как использовать логические операторы `and`, `or` и `not` в условных инструкциях. Они позволяют объединять простые условия в более сложные. Используйте `and`, если оба условия должны быть истинными, `or`, если достаточно хотя бы одного истинного условия, и `not` для логического отрицания условия.
