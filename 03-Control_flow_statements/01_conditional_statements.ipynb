# Модуль 3. Управление ходом выполнения программы

## Часть 1. Условные инструкции (`if`, `else`, `elif`)

Условные инструкции в Python позволяют выполнять разные блоки кода в зависимости от заданных условий. Для ветвления обычно используют `if`, `else` и `elif`.

### 1.1. Инструкция `if`

Инструкция `if` проверяет условие и выполняет блок кода, если условие истинно.

Ниже показан основной синтаксис `if`.

```python
condition = 1 == 1
if condition:
    print("Код, который выполняется, если условие истинно")
```

Пример

```python
age = 20

if age >= 18:
    print("Вы совершеннолетний человек.")
```

В этом примере инструкция `if` проверяет, больше ли значение переменной `age` или равно ли оно 18. Если условие выполняется, программа выполняет блок внутри `if` и выводит сообщение «Вы совершеннолетний человек.».

### 1.2. Ветвь `else`

В ветви `else` указывают блок кода, который выполняется, если условие в `if` ложно.

Ниже показан основной синтаксис `else`.

```python
condition = 1 == 0
if condition:
    print("Код, который выполняется, если условие истинно")
else:
    print("Код, который выполняется, если условие ложно")
```

Пример

```python
age = 15

if age >= 18:
    print("Вы совершеннолетний человек.")
else:
    print("Вы несовершеннолетний человек.")
```

В этом примере, если условие `age >= 18` ложно, выполняется блок внутри `else` и выводится сообщение «Вы несовершеннолетний человек.».

### 1.3. Ветвь `elif`

Ключевое слово `elif` сокращает выражение `else if` («иначе, если»). С её помощью задают дополнительное условие, которое проверяется, если предыдущие условия цепочки оказались ложными.

Ниже показан основной синтаксис `elif`.

```python
condition = 1 == 0
if condition:
    print("Код, который выполняется, если условие истинно")
elif not condition:
    print("Код, который выполняется, если условие ложно")
```

Пример

```python
age = 25

if age < 18:
    print("Вы несовершеннолетний человек.")
elif age >= 18 and age < 65:
    print("Вы совершеннолетний человек.")
else:
    print("Вы пожилой человек.")
```

В этом примере ветвь `elif` проверяет, входит ли возраст в диапазон от 18 до 65 лет, где 65 не включается. Если первое условие ложно, а второе истинно, выполняется блок внутри `elif` и выводится сообщение «Вы совершеннолетний человек.».

### 1.4. Итоги

В этой части вы познакомились с условными инструкциями Python и ветвями `if`, `else` и `elif`. Условные инструкции позволяют выполнять разные блоки кода в зависимости от заданных условий. Используйте `if` для проверки условия, `else` для кода, который выполняется при ложном условии, и `elif` для проверки дополнительных условий, если предыдущие условия ложны.
