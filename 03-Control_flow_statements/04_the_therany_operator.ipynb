## Часть 4. Тернарный оператор

Тернарный оператор, также называемый условным выражением, позволяет кратко записывать простое ветвление в Python. Он проверяет условие и выбирает значение в зависимости от результата проверки.

В следующем примере показано, как использовать тернарный оператор.

```python
num = 10

result = "Положительное" if num > 0 else "Неположительное"

print(result)
```

В этом примере тернарный оператор присваивает переменной `result` значение в зависимости от условия `num > 0`. Если условие истинно, переменной присваивается строка `"Положительное"`, в противном случае — `"Неположительное"`. Затем программа выводит значение `result`. Оно зависит от `num`.

Когда значение нужно выбрать по простому условию, тернарный оператор может сделать код короче и понятнее. Используйте его там, где это уместно. Вложенные тернарные операторы могут затруднить чтение.

### Итоги

В этой части вы познакомились с тернарным оператором в Python. Он позволяет кратко записать простое ветвление и присвоить значение по условию. В некоторых случаях код получается короче и понятнее. Чтобы сохранить ясность, используйте условные выражения умеренно и избегайте глубокой вложенности.
