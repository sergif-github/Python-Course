# Модуль 8. Модули, пакеты и библиотеки

## Часть 2. Создание и организация пакетов

В Python пакеты позволяют объединять связанные модули. Обычный пакет представляет собой каталог, содержащий один или несколько модулей и дополнительный файл `__init__.py`. В этом разделе мы разберём, как создавать и организовывать пакеты.

### 2.1. Создание пакета

Чтобы создать пакет, выполните следующие шаги.

1. Создайте каталог с подходящим именем.
2. В каталоге пакета создайте файл `__init__.py`. Файл может быть пустым или содержать код инициализации пакета.
3. Создайте в каталоге один или несколько модулей Python. В них будет находиться код вашего пакета.

Пример структуры пакета.

```
my_package/
    __init__.py
    module1.py
    module2.py
```

### 2.2. Импорт модулей из пакета

Чтобы импортировать модуль из пакета, укажите имя пакета и имя модуля в инструкции `import`. Например, чтобы импортировать `module1` из пакета `my_package`, напишите следующее.

```python
import my_package.module1
```

После импорта к функциям, классам и переменным модуля можно обращаться по полному имени. Например, функцию `my_function()` из `module1` можно вызвать так.

```python
my_package.module1.my_function()
```

С помощью ключевого слова `from` можно импортировать отдельные объекты из модуля внутри пакета.

```python
from my_package.module1 import my_function
```

После такого импорта функцию `my_function()` можно вызывать напрямую, без имени модуля.

### 2.3. Организация подпакетов

В пакетах могут быть подпакеты, то есть вложенные пакеты. Чтобы создать подпакет, выполните те же шаги, что и для обычного пакета, и поместите каталог подпакета внутрь каталога родительского пакета. У обычного подпакета также есть собственный файл `__init__.py`.

Пример структуры пакета с подпакетом.

```
my_package/
    __init__.py
    module1.py
    subpackage/
        __init__.py
        module3.py
```

Чтобы импортировать модуль из подпакета, используют запись через точку с именами пакета и подпакета. Например, `module3` можно импортировать следующим образом.

```python
import my_package.subpackage.module3
```

### 2.4. Итоги

Пакеты помогают организовывать связанные модули Python. Для обычного пакета создают каталог с файлом `__init__.py` и одним или несколькими модулями. Чтобы импортировать модуль из пакета, указывают имена пакета и модуля. Внутри пакета можно организовать подпакеты. Умение создавать и организовывать пакеты помогает структурировать код, упрощает поддержку и повторное использование.
