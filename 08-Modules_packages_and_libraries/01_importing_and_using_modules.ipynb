# Модуль 8. Модули, пакеты и библиотеки

## Часть 1. Введение в модули

В Python модуль — это файл с кодом Python, в котором определены функции, классы и переменные. Модули позволяют организовать и повторно использовать код, что упрощает поддержку больших программ и совместную работу над ними. Импортируя модуль, вы получаете доступ к его функциям, классам и переменным и можете применять их в своём коде.

Пакет позволяет организовать связанные модули в иерархию каталогов. Библиотека — это набор модулей или пакетов с готовым кодом для конкретных задач или распространённых проблем.

### 1.1. Использование модулей

Чтобы использовать модуль в программе на Python, сначала импортируйте его. В Python это можно сделать несколькими способами.

- Импорт всего модуля

  Чтобы импортировать весь модуль, укажите его имя после инструкции `import`.

    ```python
    import math
    ```

  После импорта к функциям, классам и переменным модуля можно обращаться через его имя. Например, чтобы использовать функцию `sqrt()` модуля `math`, напишите `math.sqrt()`.

- Импорт отдельных объектов

  Если нужны только отдельные функции, классы или переменные модуля, их можно импортировать напрямую с помощью ключевого слова `from`.

    ```python
    from math import sqrt, pi
    ```

  После такого импорта к `sqrt()` и `pi` можно обращаться в коде без имени модуля.

- Импорт с псевдонимом

  Модулю или отдельному объекту можно задать псевдоним с помощью ключевого слова `as`. Это удобно, когда нужно более короткое или более понятное имя.

    ```python
    import math as m
    ```

  После такого импорта можно писать `m.sqrt()` вместо `math.sqrt()`.

### 1.2. Модули стандартной библиотеки

Библиотеки могут входить в стандартную библиотеку Python, которая поставляется вместе с языком, или быть сторонними, созданными другими разработчиками. В стандартную библиотеку Python входит много модулей для разных задач. Они доступны без дополнительной установки и помогают решать типовые задачи с помощью готового кода.

Примеры часто используемых модулей стандартной библиотеки.

- `random`

  Предоставляет функции для генерации случайных чисел, случайного выбора элементов последовательности и перемешивания последовательностей.

- `datetime`

  Помогает работать с датами и временем и задавать формат их представления. В нём есть классы для работы с датами, временем, часовыми поясами и интервалами.

- `os`

  Позволяет взаимодействовать с операционной системой. С его помощью можно работать с файлами и каталогами, управлять процессами и обращаться к переменным окружения.

- `csv`

  Предоставляет средства для чтения и записи CSV-файлов (Comma-Separated Values, значения, разделённые запятыми). Модуль позволяет работать с табличными данными в структурированном виде.

- `math`

  Предоставляет математические функции и константы.

- `json`

  Предоставляет функции для работы с данными в формате JSON.

Это лишь несколько примеров. В стандартной библиотеке есть модули для многих других задач, включая сетевое программирование (`socket`), регулярные выражения (`re`) и сериализацию данных (`pickle`).

Полный список модулей стандартной библиотеки и их возможностей приведён в официальной документации Python по адресу https://docs.python.org/3/library/index.html.

### 1.3. Сторонние модули

Помимо стандартной библиотеки, у Python есть обширная экосистема сторонних модулей, которые можно устанавливать и использовать в своих проектах. Их создаёт сообщество Python, чтобы добавить возможности для веб-разработки, анализа данных, машинного обучения и других задач. Популярные примеры — `numpy`, `pandas`, `matplotlib` и `requests`. Их можно устанавливать с помощью менеджеров пакетов, например `pip` или `conda`.

### 1.4. Итоги

Модули — важная часть программирования на Python. Они позволяют организовывать и повторно использовать код. Импорт даёт доступ к определённым в модулях функциям, классам и переменным, поэтому можно пользоваться готовыми возможностями и создавать более функциональные приложения. В Python можно импортировать модуль целиком, импортировать только отдельные объекты и задавать псевдонимы. Расширить возможности программ помогают модули стандартной библиотеки и многочисленные сторонние модули.
