# Модуль 7. Обработка исключений

## Часть 5. Упражнения

### Упражнение 1
Напишите программу, которая получает от пользователя два целых числа и делит первое на второе. Обработайте возможное исключение `ZeroDivisionError` с помощью блока `try-except`.

### Упражнение 2
Напишите программу, которая запрашивает у пользователя число и преобразует его в целое. Обработайте исключения `ValueError` и `TypeError`, которые могут возникнуть при преобразовании.

### Упражнение 3
Напишите программу, которая получает от пользователя два числа и делит первое на второе. Обработайте исключения `ZeroDivisionError` и `ValueError` в одном блоке `except`.

### Упражнение 4
Напишите программу, которая пытается преобразовать строку в целое число и выполняет деление. Обрабатывайте исключения в следующем порядке: сначала `ValueError`, затем `ZeroDivisionError`, а после них — остальные исключения.

### Упражнение 5
Напишите программу, которая запрашивает два числа и вычисляет их сумму. Обработайте исключение `ValueError` и выведите результат в блоке `else`, если исключения не возникли.

### Упражнение 6
Напишите функцию, которая принимает целое число и проверяет, является ли оно положительным. Если число отрицательное или равно нулю, вызовите исключение `ValueError` с подходящим сообщением.

### Упражнение 7
Напишите функцию, которая принимает строку и проверяет, находятся ли все её буквы в верхнем регистре. Используйте метод `.isupper()`. Если условие не выполнено, вызовите собственное исключение `UppercaseError` с подходящим сообщением.
