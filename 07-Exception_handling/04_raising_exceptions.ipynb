# Модуль 7. Обработка исключений

## Часть 4. Вызов исключений

В Python исключения можно вызывать явно, чтобы сообщать об ошибке или необычной ситуации во время выполнения кода. Это полезно, когда нужно обработать определённый случай или проверить соблюдение заданных условий.

### 4.1. Вызов исключений

Чтобы вызвать исключение, используют инструкцию `raise`, после которой указывают класс исключения или его экземпляр.

Базовый синтаксис выглядит так.

```python
raise ExceptionType("Сообщение об ошибке")
```

Например, следующий код вызывает `ValueError` с собственным сообщением.

```python
raise ValueError("Некорректный ввод. Введите положительное число.")
```

После вызова исключения обычное выполнение прерывается. Исключение передаётся вверх по стеку вызовов, пока его не перехватит подходящий блок `except`.

В следующем примере мы явно вызываем исключение.

````python
def check_positive(value):
    if value <= 0:
        raise ValueError("Value must be positive.")
    return value

try:
    num = int(input("Введите положительное число: "))
    validated_num = check_positive(num)
    print("Число:", validated_num)
except ValueError as ve:
    print("Ошибка:", str(ve))
````

Функция `check_positive` проверяет, является ли переданное число положительным. Если число меньше или равно нулю, мы вызываем `ValueError` с собственным сообщением об ошибке.

В блоке `try` мы просим пользователя ввести положительное число и вызываем функцию `check_positive` для проверки. Если число не проходит проверку, возникает `ValueError`. Соответствующий блок `except` обрабатывает исключение и выводит сообщение об ошибке.

Этот пример показывает, как вызвать исключение при конкретном условии. Если введённое число не соответствует требованиям, пользователь получает понятное сообщение.

### 4.2. Пользовательские исключения

Помимо встроенных исключений, можно создавать собственные классы, описывающие определённый тип ошибки или необычный случай в коде. Пользовательские исключения позволяют выводить более содержательные сообщения и точнее определять, как обрабатывать определённые ситуации.

Чтобы создать пользовательское исключение, обычно определяют класс, который наследуется от `Exception` или одного из его подклассов.

Последний пример использует пользовательское исключение.

````python
class NonPositiveError(Exception):
    def __init__(self, message="Value must be positive."):
        self.message = message
        super().__init__(self.message)

def check_positive(value):
    if value <= 0:
        raise NonPositiveError()
    return value

try:
    num = int(input("Введите положительное число: "))
    validated_num = check_positive(num)
    print("Число:", validated_num)
except NonPositiveError as npe:
    print("Ошибка:", str(npe))
````

В примере мы определяем пользовательское исключение `NonPositiveError`, подкласс `Exception`. Класс `NonPositiveError` переопределяет метод `__init__`, чтобы можно было задавать сообщение об ошибке.

Внутри функции `check_positive` мы вызываем исключение `NonPositiveError`, если значение меньше или равно нулю.

В блоке `try` мы просим пользователя ввести положительное число и вызываем функцию `check_positive`. Если число не проходит проверку, возникает `NonPositiveError`. Соответствующий блок `except` перехватывает именно `NonPositiveError` и выводит связанное с ним сообщение.

Пользовательские исключения позволяют точнее описывать возникшие ошибки. Код становится понятнее, и для каждого типа ошибки можно подобрать подходящую обработку.

### 4.3. Итоги

Вызов исключений позволяет явно обозначать ошибки и особые ситуации в коде. Вызывая встроенные или пользовательские исключения, можно обрабатывать определённые ситуации, проверять соблюдение условий и формировать понятные сообщения об ошибках. При вызове исключения обычное выполнение программы прерывается, а исключение передаётся вверх по стеку вызовов, пока его не перехватит подходящий блок `except`.
