# Модуль 7. Обработка исключений

## Часть 1. Обработка распространённых ошибок

При разработке программ на Python ошибки встречаются часто. Знание распространённых типов ошибок и способов их обработки помогает сделать код надёжнее и устойчивее. В этом разделе разобраны часто встречающиеся ошибки и способы работы с ними.

### 1.1. Распространённые ошибки

#### Синтаксические ошибки

Синтаксические ошибки возникают при нарушении правил синтаксиса Python. Обычно интерпретатор обнаруживает их на этапе синтаксического анализа, и программа не запускается. Типичные причины — пропущенные скобки, неправильные отступы и опечатки в ключевых словах.

#### Ошибки имени

Ошибка `NameError` возникает при обращении к неопределённому имени, например если переменной ещё не присвоено значение или в её имени есть опечатка. Интерпретатор не может найти указанное имя.

#### Ошибки типа

Ошибка `TypeError` возникает при выполнении операции над объектами несовместимых типов. Обычно это означает, что тип объекта не поддерживает способ его использования в этой операции.

#### Ошибки индекса и ключа

Ошибка `IndexError` возникает при обращении по недопустимому индексу в последовательности, например в списке или строке. Ошибка `KeyError` возникает при попытке получить значение по отсутствующему ключу словаря.

#### Ошибки атрибута

Ошибка `AttributeError` возникает при обращении к атрибуту или методу, которого нет у объекта. Такое бывает, если допустить опечатку в имени атрибута или если у объекта отсутствует ожидаемый атрибут.

#### Ошибки значения

Ошибка `ValueError` возникает, когда переданный аргумент имеет подходящий тип, но недопустимое значение. Например, вы передали отрицательное число функции, которая ожидает положительное.

#### Ошибки импорта

Ошибки импорта возникают, если Python не находит импортируемый модуль или при загрузке модуля возникают проблемы с его зависимостями.

#### Ошибки при работе с файлами

Ошибки при работе с файлами могут возникать, например, при открытии файла, чтении или записи данных. Распространённые примеры — `FileNotFoundError`, `PermissionError` и `IOError`.

### 1.2. Обработка исключений с помощью блоков `try-except`

Для обработки ошибок и исключений, которые могут возникнуть во время выполнения кода, в Python есть конструкция `try-except`. С её помощью можно перехватить конкретные исключения и предусмотреть действия для таких ситуаций.

Чтобы обработать ошибку с помощью `try-except`, выполните следующие шаги.

- Поместите код, который может вызвать исключение, в блок `try`.
- В блоке `except` укажите, какие исключения нужно перехватить.
- Внутри блока `except` обработайте ошибку, например покажите сообщение или выполните запасной сценарий.
- При необходимости добавьте общий блок `except` для исключений, которые не перехватили предыдущие блоки, или для общей обработки ошибок.

### 1.3. Итоги

Обработка распространённых ошибок — важная часть программирования на Python. Если знать, какие ошибки могут возникнуть, например синтаксические ошибки, `NameError`, `TypeError`, `IndexError`, `KeyError` и ошибки при работе с файлами, можно подобрать подходящий способ обработки. Для этого читайте сообщения об ошибках и трассировки стека вызовов, проверяйте имена и определения переменных, типы данных и допустимость индексов или ключей. Используйте блоки `try-except` для обработки исключений. Грамотная обработка ошибок делает программы устойчивее и надёжнее.
