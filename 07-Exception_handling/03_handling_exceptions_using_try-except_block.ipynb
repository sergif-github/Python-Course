# Модуль 7. Обработка исключений

## Часть 3. Обработка исключений с помощью блоков `try-except`

В Python для обработки исключений можно использовать конструкцию `try-except`. Она позволяет перехватывать и обрабатывать конкретные исключения, возникающие во время выполнения кода. Это помогает предотвратить аварийное завершение программы, выполнить альтернативное действие или показать пользователю сообщение об ошибке.

### 3.1. Синтаксис блока `try-except`

Конструкция `try-except` состоит из блока `try` с кодом, который может вызвать исключение, и одного или нескольких блоков `except`, определяющих, как обрабатывать конкретные исключения.

Базовый синтаксис выглядит так.

```python
try:
    # Код, который может вызвать исключение
    # ...
except ExceptionType:
    # Код обработки исключения
    # ...
```

В блоке `except` указывают тип обрабатываемого исключения, здесь это `ExceptionType`. Если в блоке `try` возникает исключение этого типа, выполняется соответствующий блок `except`.

Например, исключение `ValueError` можно обработать следующим образом.

````python
try:
    age = int(input("Введите ваш возраст: "))
except ValueError:
    print("Некорректный возраст. Введите целое число.")
````

В этом примере, если пользователь введёт значение возраста, которое нельзя преобразовать в целое число, возникнет исключение `ValueError`. Соответствующий блок `except ValueError` выполнится и покажет сообщение об ошибке.

### 3.2. Обработка нескольких исключений

Чтобы обработать несколько исключений, можно использовать несколько блоков `except`. Каждый блок `except` может обрабатывать свой тип исключения. Выполнится первый блок, подходящий для возникшего исключения, а остальные блоки `except` будут пропущены.

```python
 try:
    # Код, который может вызвать исключение
    # ...
except ValueError:
    # Обработка исключения ValueError
    # ...
except TypeError:
    # Обработка исключения TypeError
    # ...
```

В примере при возникновении `ValueError` выполнится первый блок `except ValueError`, а при возникновении `TypeError` — второй блок `except TypeError`.

Несколько исключений можно обработать следующим образом.

````python
try:
    age = int(input("Введите ваш возраст: "))
    result = 10 / age
except ValueError:
    print("Некорректный возраст. Введите целое число.")
except ZeroDivisionError:
    print("Ошибка: деление на ноль.")
````

В этом примере мы добавили ещё один блок `except`, чтобы обработать исключение `ZeroDivisionError`. Если пользователь введёт возраст, равный 0, при попытке вычислить `10 / age` возникнет `ZeroDivisionError`. Второй блок `except` перехватит это исключение и покажет соответствующее сообщение.

### 3.3. Обработка нескольких исключений в одном блоке `except`

Если несколько типов исключений нужно обработать одинаково, их можно перечислить в одном блоке `except`. Такой блок позволяет использовать общий код для разных типов ошибок.

```python
try:
    # Код, который может вызвать исключение
    # ...
except (ExceptionType1, ExceptionType2):
    # Обработка исключений ExceptionType1 и ExceptionType2
    # ...
```

Перепишем предыдущий пример так, чтобы оба исключения обрабатывались в одном блоке `except`.

````python
try:
    age = int(input("Введите ваш возраст: "))
    result = 10 / age
except (ValueError, ZeroDivisionError):
    print("Некорректный возраст или деление на ноль.")
````

В этом примере после `except` мы указываем в скобках кортеж типов исключений `(ValueError, ZeroDivisionError)`. Это позволяет обрабатывать `ValueError` и `ZeroDivisionError` в одном блоке. Если возникнет любое из этих исключений, код блока `except` выполнится и выведет соответствующее сообщение об ошибке.

### 3.4. Общая обработка исключений

Чтобы обработать исключения, которые не перехватили блоки `except` для конкретных типов, можно использовать общий блок `except` без указания типа. Как правило, лучше перехватывать конкретные исключения, когда это возможно. Так проще подобрать действия и сообщение для соответствующей ошибки.

Пример общего блока `except`.

```python
try:
    # Код, который может вызвать исключение
    # ...
except:
    # Общий код обработки исключений
    # ...
```

Если возникнет исключение, которое не перехватили предыдущие блоки `except`, выполнится общий блок.

````python
try:
    age = int(input("Введите ваш возраст: "))
    result = 10 / age
except Exception as e:
    print("Произошла ошибка:", str(e))
````

В этом примере общий обработчик `except Exception as e` перехватывает исключения класса `Exception` и его подклассов, возникшие внутри блока `try`.

Класс `Exception` — базовый для большинства встроенных исключений. Такие исключения, как `KeyboardInterrupt` и `SystemExit`, наследуются от `BaseException` вне ветви `Exception`.

### 3.5. Обработка исключений с учётом иерархии

При обработке конкретных исключений важно учитывать их иерархию. В Python одни классы исключений являются подклассами других. Если нужно обработать и исключение подкласса, и исключение базового класса, начните с обработчика для более конкретного типа, а затем добавьте обработчик для общего типа.

Например, чтобы по отдельности обработать `FileNotFoundError` и более общий тип `OSError`, можно написать следующее.

```python
try:
    # Код, который может вызвать исключение
    # ...
except FileNotFoundError:
    # Обработка исключения FileNotFoundError
    # ...
except OSError:
    # Обработка исключения OSError
    # ...
```

В примере при возникновении `FileNotFoundError` выполнится первый блок `except FileNotFoundError`. При другой системной ошибке ввода-вывода выполнится второй блок `except OSError`.

Следующий пример показывает обработку исключений от конкретных типов к общим.

````python
try:
    # Код, который может вызвать исключение
    x = int(input("Введите число: "))
    y = int(input("Введите ещё одно число: "))
    result = x / y
except ValueError:
    # Сначала обрабатываем конкретные типы исключений
    print("Некорректный ввод. Введите целые числа.")
except ZeroDivisionError:
    # Затем обрабатываем другие конкретные типы исключений
    print("Ошибка: деление на ноль.")
except Exception:
    # В конце обрабатываем более общие типы исключений
    print("Произошла ошибка.")
````

Сначала мы обрабатываем `ValueError`, которое возникает, когда пользователь вводит значение, не преобразуемое в целое число. Для этого типа предусмотрено своё сообщение об ошибке.

Затем мы обрабатываем `ZeroDivisionError`, которое возникает, если пользователь вводит 0 в качестве делителя. Для этого типа также выводится отдельное сообщение.

В конце расположен более общий блок `except Exception`, который обрабатывает остальные исключения этой ветви иерархии, пропущенные предыдущими блоками. Он позволяет обработать неожиданные или неизвестные ошибки.

Такой порядок позволяет сначала реагировать на конкретные типы исключений и при необходимости переходить к более общим. Для разных сценариев ошибок можно задать более точную обработку.

### 3.6. Блок `else`

После всех блоков `except` можно добавить необязательный блок `else`. Код внутри него выполняется, когда блок `try` завершается без исключений. В `else` часто выполняют завершающие или дополнительные действия, которые нужны только после успешного выполнения блока `try`.

```python
try:
    # Код, который может вызвать исключение
    # ...
except ExceptionType:
    # Код обработки исключения
    # ...
else:
    # Код, который выполняется при отсутствии исключений
    # ...
```

````python
try:
    # Код, который может вызвать исключение
    x = int(input("Введите число: "))
    y = int(input("Введите ещё одно число: "))
    result = x / y
except ValueError:
    # Сначала обрабатываем конкретные типы исключений
    print("Некорректный ввод. Введите целые числа.")
except ZeroDivisionError:
    # Затем обрабатываем другие конкретные типы исключений
    print("Ошибка: деление на ноль.")
except Exception:
    # В конце обрабатываем более общие типы исключений
    print("Произошла ошибка.")
else:
    # Код, который выполняется при отсутствии исключений
    print("Результат деления:", result)
````

В обновлённом коде мы добавили блок `else` после блоков `except`. Код в блоке `else` выполнится только при отсутствии исключений в `try`. В данном случае он выводит результат деления.

Блок `else` отделяет код, который должен выполниться при отсутствии исключений, и делает программу понятнее. Он особенно полезен, когда определённые действия нужно выполнить только после успешного завершения блока `try`.

Блок `else` необязателен. Если при отсутствии исключений дополнительных действий не требуется, его можно опустить.

### 3.7. Блок `finally`

Ещё один необязательный блок конструкции `try-except` — `finally`. Код внутри `finally` выполняется независимо от того, возникло исключение или нет. Этот блок часто используют, чтобы освободить ресурсы, закрыть файлы или выполнить другие завершающие действия.

```python
try:
    # Код, который может вызвать исключение
    # ...
except ExceptionType:
    # Код обработки исключения
    # ...
finally:
    # Код, который выполняется в любом случае
    # ...
```

````python
try:
    # Код, который может вызвать исключение
    x = int(input("Введите число: "))
    y = int(input("Введите ещё одно число: "))
    result = x / y
except ValueError:
    # Сначала обрабатываем конкретные типы исключений
    print("Некорректный ввод. Введите целые числа.")
except ZeroDivisionError:
    # Затем обрабатываем другие конкретные типы исключений
    print("Ошибка: деление на ноль.")
except Exception:
    # В конце обрабатываем более общие типы исключений
    print("Произошла ошибка.")
else:
    # Код, который выполняется при отсутствии исключений
    print("Результат деления:", result)
finally:
    # Код, который выполняется независимо от исключений
    print("Блок finally выполнен.")
````

В обновлённом коде мы добавили блок `finally` после блоков `except` и `else`. Код внутри блока `finally` выполняется независимо от того, возникло исключение или нет.

Блок `finally` нужен для освобождения ресурсов или других действий, которые должны произойти при любом исходе выполнения блока `try`. Такие действия выполняются, даже если возникло исключение и обработчик его перехватил.

В этом примере блок `finally` выводит сообщение о том, что он выполнился.

Блок `finally` необязателен. Его можно опустить, если нет ресурсов или действий, которыми нужно заняться независимо от исключений.

### 3.8. Итоги

Конструкция `try-except` позволяет учитывать возможные ошибки и поддерживать нормальную работу программы. Перехват и обработка исключений помогают показывать более понятные сообщения, восстанавливаться после неожиданных ситуаций и предотвращать аварийное завершение программы.

Перехватывая конкретные исключения, можно задавать для разных типов ошибок разные действия. Если указать тип исключения в `except`, обработку можно настроить для определённого сценария. Можно обработать исключение одного типа, задать несколько обработчиков или обработать несколько типов в одном блоке `except`. Чтобы каждый обработчик срабатывал в нужных случаях, учитывайте иерархию исключений и размещайте более конкретные обработчики перед более общими.
