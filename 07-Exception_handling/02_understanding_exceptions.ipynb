# Модуль 7. Обработка исключений

## Часть 2. Что такое исключения

В Python исключение сигнализирует о ситуации, которая возникла во время выполнения программы. Исключение нарушает обычный ход выполнения, и программа может завершиться, если его не обработать. Понимание исключений необходимо, чтобы писать код, устойчивый к ошибкам.

### 2.1. Что такое исключение?

Исключение — это объект Python, который представляет ошибочную или неожиданную ситуацию. Исключение может вызвать как интерпретатор, так и сам программист. Причинами бывают, например, некорректный ввод, деление на ноль, отсутствие файла или ошибка сети.

### 2.2. Обработка исключений

Обработка исключений включает их перехват и выполнение действий в ответ. Она помогает корректно реагировать на ошибки, предотвращать аварийное завершение и показывать пользователю понятные сообщения. Программа может восстановиться после ошибки или выполнить альтернативные действия и продолжить работу.

### 2.3. Иерархия исключений

В Python исключения образуют иерархию. На её вершине находится `BaseException` — базовый класс всех встроенных исключений. От него напрямую или через другие классы наследуются `Exception`, `TypeError`, `ValueError`, `FileNotFoundError` и другие встроенные исключения. Вы можете создавать собственные классы исключений, наследуясь от встроенных классов.

### 2.4. Блок `try-except`

Для обработки исключений используют конструкцию `try-except`. В блоке `try` находится код, который может вызвать исключение, а блок `except` задаёт действия для конкретного исключения. Можно определить несколько блоков `except` для разных типов исключений. Необязательный блок `else` выполняется, когда блок `try` завершается без исключений, а блок `finally` выполняется независимо от того, возникло исключение или нет.

```python
try:
    # Код, который может вызвать исключение
    # ...
except ExceptionType1:
    # Обработка исключения ExceptionType1
    # ...
except ExceptionType2:
    # Обработка исключения ExceptionType2
    # ...
else:
    # Код, который выполняется при отсутствии исключений
    # ...
finally:
    # Код, который выполняется в любом случае
    # ...
```

### 2.5. Вызов исключений

Можно обрабатывать исключения, вызванные интерпретатором или сторонними библиотеками, а также самостоятельно вызывать исключения инструкцией `raise`. Так вы явно сообщаете, что при выполнении кода возникла определённая ошибка.

```python
raise ExceptionType("Сообщение об ошибке")
```

### 2.6. Итоги

Чтобы писать надёжные программы на Python, нужно понимать исключения и уметь их обрабатывать. Исключения позволяют реагировать на ошибки, восстанавливаться после непредвиденных ситуаций и продолжать выполнение программы. Перехват и обработка исключений с помощью `try-except` помогают предотвратить аварийное завершение и улучшить работу программы для пользователя.
